```python
import numpy as np
import cv2
import random
from typing import Dict, List, Tuple
import pickle
import os
from scene_geometry_generator import SceneGeometryGenerator
from scene_reference_generator import SceneReferenceGenerator

class SceneTrainingDataGenerator:
    def __init__(self, canvas_size=256, ref_size=64, num_references=12):
        self.canvas_size = canvas_size
        self.ref_size = ref_size
        self.num_references = num_references
        
        self.scene_generator = SceneGeometryGenerator(canvas_size, ref_size)
        self.ref_generator = SceneReferenceGenerator(ref_size, canvas_size)
        
        # Scene complexity levels for curriculum learning
        self.complexity_levels = {
            "simple": {"chars": (1, 2), "assets": (1, 3), "interactions": False},
            "medium": {"chars": (2, 3), "assets": (2, 5), "interactions": True},
            "complex": {"chars": (3, 5), "assets": (3, 8), "interactions": True}
        }
    
    def encode_scene_prompt(self, scene_layout: Dict) -> np.ndarray:
        """Encode scene layout into structured prompt vector"""
        encoding = []
        
        # Scene metadata
        num_characters = len(scene_layout["characters"])
        num_assets = len(scene_layout["assets"])
        bg_type = scene_layout["background"]["type"]
        
        # Encode number of characters (0-5, one-hot)
        char_count_encoding = [1 if i == min(num_characters-1, 4) else 0 for i in range(5)]
        encoding.extend(char_count_encoding)
        
        # Encode number of assets (0-8, one-hot)
        asset_count_encoding = [1 if i == min(num_assets-1, 7) else 0 for i in range(8)]
        encoding.extend(asset_count_encoding)
        
        # Encode background type
        bg_types = list(self.scene_generator.background_types.keys())
        bg_encoding = [1 if bg == bg_type else 0 for bg in bg_types]
        encoding.extend(bg_encoding)
        
        # Encode character types present
        char_types = list(self.scene_generator.character_types.keys())
        for char_type in char_types:
            has_char = any(char["type"] == char_type for char in scene_layout["characters"])
            encoding.append(1 if has_char else 0)
        
        # Encode asset types present
        asset_types = list(self.scene_generator.asset_types.keys())
        for asset_type in asset_types:
            has_asset = any(asset["type"] == asset_type for asset in scene_layout["assets"])
            encoding.append(1 if has_asset else 0)
        
        # Encode dominant interaction type
        interactions = ["facing_each_other", "playing_together", "chasing", "following", "standing_apart"]
        if scene_layout["relationships"]:
            dominant_interaction = scene_layout["relationships"][0]["interaction"]
        else:
            dominant_interaction = "standing_apart"
        interaction_encoding = [1 if interaction == dominant_interaction else 0 for interaction in interactions]
        encoding.extend(interaction_encoding)
        
        # Encode scene density (sparse/medium/dense based on total elements)
        total_elements = num_characters + num_assets
        if total_elements <= 3:
            density = "sparse"
        elif total_elements <= 6:
            density = "medium"
        else:
            density = "dense"
        
        density_encoding = [1 if d == density else 0 for d in ["sparse", "medium", "dense"]]
        encoding.extend(density_encoding)
        
        return np.array(encoding, dtype=np.float32)
    
    def generate_scene_references(self, scene_layout: Dict) -> List[np.ndarray]:
        """Generate reference images for scene generation"""
        references = []
        
        # Category 1-2: Character references (heads and bodies)
        unique_char_types = list(set(char["type"] for char in scene_layout["characters"]))
        if len(unique_char_types) >= 1:
            char_type = unique_char_types[0]
            references.append(self.ref_generator.create_noisy_character_reference(char_type, "head"))
            references.append(self.ref_generator.create_noisy_character_reference(char_type, "body"))
        else:
            # Default character references
            references.append(self.ref_generator.create_noisy_character_reference("dog", "head"))
            references.append(self.ref_generator.create_noisy_character_reference("dog", "body"))
        
        # Category 3: Pose/interaction reference
        if scene_layout["relationships"]:
            interaction = scene_layout["relationships"][0]["interaction"]
            references.append(self.ref_generator.create_interaction_reference(interaction))
        else:
            references.append(self.ref_generator.create_pose_reference("standing"))
        
        # Category 4-5: Asset references
        unique_asset_types = list(set(asset["type"] for asset in scene_layout["assets"]))
        for i in range(2):
            if i < len(unique_asset_types):
                asset_type = unique_asset_types[i]
                references.append(self.ref_generator.create_asset_reference(asset_type))
            else:
                # Default asset
                references.append(self.ref_generator.create_asset_reference("tree"))
        
        # Category 6: Background reference
        bg_type = scene_layout["background"]["type"]
        references.append(self.ref_generator.create_background_reference(bg_type))
        
        # Category 7: Layout reference
        num_elements = len(scene_layout["characters"]) + len(scene_layout["assets"])
        layout_type = random.choice(["linear", "circular", "grouped", "scattered"])
        references.append(self.ref_generator.create_layout_reference(num_elements, layout_type))
        
        # Category 8: Secondary character (if exists)
        if len(unique_char_types) >= 2:
            char_type = unique_char_types[1]
            references.append(self.ref_generator.create_noisy_character_reference(char_type, "full_character"))
        else:
            # Add another pose reference
            pose = random.choice(self.scene_generator.poses)
            references.append(self.ref_generator.create_pose_reference(pose))
        
        # Category 9-12: Additional context references
        remaining_slots = self.num_references - len(references)
        
        for i in range(remaining_slots):
            ref_type = random.choice([
                "extra_asset", "extra_character", "extra_layout", "extra_background"
            ])
            
            if ref_type == "extra_asset" and unique_asset_types:
                asset_type = random.choice(unique_asset_types)
                references.append(self.ref_generator.create_asset_reference(asset_type))
            elif ref_type == "extra_character" and unique_char_types:
                char_type = random.choice(unique_char_types)
                focus = random.choice(["head", "body", "full_character"])
                references.append(self.ref_generator.create_noisy_character_reference(char_type, focus))
            elif ref_type == "extra_layout":
                layout_type = random.choice(["linear", "circular", "grouped", "scattered"])
                references.append(self.ref_generator.create_layout_reference(num_elements, layout_type))
            else:
                # Extra background
                references.append(self.ref_generator.create_background_reference(bg_type))
        
        # Ensure we have exactly num_references
        while len(references) < self.num_references:
            references.append(self.ref_generator.create_background_reference("park"))
        
        return references[:self.num_references]
    
    def generate_training_sample(self, complexity: str = "simple") -> Tuple[List[np.ndarray], np.ndarray, np.ndarray, Dict]:
        """Generate one complete training sample"""
        
        # Get complexity parameters
        complexity_params = self.complexity_levels[complexity]
        
        # Generate scene parameters
        num_characters = random.randint(*complexity_params["chars"])
        num_assets = random.randint(*complexity_params["assets"])
        background_type = random.choice(list(self.scene_generator.background_types.keys()))
        
        # Generate scene layout
        scene_layout = self.scene_generator.generate_scene_layout(
            num_characters, num_assets, background_type
        )
        
        # Create scene sketch (target)
        scene_sketch = self.scene_generator.create_scene_sketch(scene_layout)
        
        # Generate reference images
        reference_images = self.generate_scene_references(scene_layout)
        
        # Encode scene prompt
        scene_prompt = self.encode_scene_prompt(scene_layout)
        
        return reference_images, scene_prompt, scene_sketch, scene_layout
    
    def generate_curriculum_dataset(self, samples_per_level: int = 2000, save_path: str = "scene_dataset"):
        """Generate curriculum learning dataset with increasing complexity"""
        
        if not os.path.exists(save_path):
            os.makedirs(save_path)
        
        all_ref_images = []
        all_prompts = []
        all_sketches = []
        all_metadata = []
        
        total_samples = 0
        
        # Generate samples for each complexity level
        for complexity in ["simple", "medium", "complex"]:
            print(f"Generating {samples_per_level} {complexity} scenes...")
            
            for i in range(samples_per_level):
                if i % 500 == 0:
                    print(f"  Generated {i}/{samples_per_level} {complexity} samples")
                
                ref_images, prompt, sketch, metadata = self.generate_training_sample(complexity)
                
                ref_images, prompt, sketch, metadata = self.generate_training_sample(complexity)
                
                all_ref_images.append(np.stack(ref_images))  # Stack reference images
                all_prompts.append(prompt)
                all_sketches.append(sketch)
                
                # Add complexity level to metadata
                metadata["complexity"] = complexity
                metadata["sample_id"] = total_samples
                all_metadata.append(metadata)
                
                total_samples += 1
        
        # Convert to numpy arrays
        all_ref_images = np.array(all_ref_images)  # [N, num_refs, ref_size, ref_size]
        all_prompts = np.array(all_prompts)        # [N, prompt_dim]
        all_sketches = np.array(all_sketches)      # [N, canvas_size, canvas_size]
        
        # Save dataset
        np.save(os.path.join(save_path, "reference_images.npy"), all_ref_images)
        np.save(os.path.join(save_path, "scene_prompts.npy"), all_prompts)
        np.save(os.path.join(save_path, "scene_sketches.npy"), all_sketches)
        
        with open(os.path.join(save_path, "scene_metadata.pkl"), "wb") as f:
            pickle.dump(all_metadata, f)
        
        # Save dataset info
        dataset_info = {
            "total_samples": total_samples,
            "samples_per_complexity": samples_per_level,
            "complexity_levels": list(self.complexity_levels.keys()),
            "num_references": self.num_references,
            "canvas_size": self.canvas_size,
            "ref_size": self.ref_size,
            "prompt_dim": all_prompts.shape[1]
        }
        
        with open(os.path.join(save_path, "dataset_info.json"), "w") as f:
            import json
            json.dump(dataset_info, f, indent=2)
        
        print(f"\nDataset generation complete!")
        print(f"Total samples: {total_samples}")
        print(f"Reference images shape: {all_ref_images.shape}")
        print(f"Scene prompts shape: {all_prompts.shape}")
        print(f"Scene sketches shape: {all_sketches.shape}")
        print(f"Dataset saved to: {save_path}")
        
        return all_ref_images, all_prompts, all_sketches, all_metadata
    
    def generate_test_visualizations(self, num_samples: int = 5, save_path: str = "scene_visualizations"):
        """Generate test samples for visualization"""
        
        if not os.path.exists(save_path):
            os.makedirs(save_path)
        
        for complexity in ["simple", "medium", "complex"]:
            for sample_idx in range(num_samples):
                # Generate sample
                ref_images, prompt, sketch, metadata = self.generate_training_sample(complexity)
                
                # Create visualization
                self._create_sample_visualization(
                    ref_images, prompt, sketch, metadata, 
                    f"{save_path}/{complexity}_sample_{sample_idx}.png"
                )
        
        print(f"Test visualizations saved to {save_path}")
    
    def _create_sample_visualization(self, ref_images: List[np.ndarray], prompt: np.ndarray, 
                                   sketch: np.ndarray, metadata: Dict, save_path: str):
        """Create visualization of a training sample"""
        try:
            import matplotlib.pyplot as plt
            
            # Create figure with subplots
            fig, axes = plt.subplots(4, 4, figsize=(16, 16))
            
            # Show reference images (3x4 grid)
            ref_titles = [
                "Char Head", "Char Body", "Interaction", "Asset 1", 
                "Asset 2", "Background", "Layout", "Char 2",
                "Extra 1", "Extra 2", "Extra 3", "Extra 4"
            ]
            
            for i in range(min(12, len(ref_images))):
                row = i // 4
                col = i % 4
                axes[row, col].imshow(ref_images[i], cmap='gray')
                axes[row, col].set_title(ref_titles[i], fontsize=10)
                axes[row, col].axis('off')
            
            # Show target sketch
            axes[3, 0].imshow(sketch, cmap='gray')
            axes[3, 0].set_title('Target Scene Sketch', fontsize=12)
            axes[3, 0].axis('off')
            
            # Show scene metadata
            scene_info = f"""Scene: {metadata.get('complexity', 'Unknown')}
Characters: {len(metadata.get('characters', []))}
Assets: {len(metadata.get('assets', []))}
Background: {metadata.get('background', {}).get('type', 'Unknown')}
Relationships: {len(metadata.get('relationships', []))}"""
            
            axes[3, 1].text(0.1, 0.5, scene_info, fontsize=10, verticalalignment='center')
            axes[3, 1].set_title('Scene Info', fontsize=12)
            axes[3, 1].axis('off')
            
            # Show prompt encoding info
            prompt_info = f"""Prompt Dimensions: {len(prompt)}
Char Count Encoding: {prompt[:5]}
Asset Count Encoding: {prompt[5:13]}
Background Encoding: {prompt[13:18]}
Interaction Encoding: {prompt[-8:-3]}"""
            
            axes[3, 2].text(0.1, 0.5, prompt_info, fontsize=8, verticalalignment='center', family='monospace')
            axes[3, 2].set_title('Prompt Encoding', fontsize=12)
            axes[3, 2].axis('off')
            
            # Hide empty subplot
            axes[3, 3].axis('off')
            
            plt.tight_layout()
            plt.savefig(save_path, dpi=150, bbox_inches='tight')
            plt.close()
            
        except ImportError:
            print("Matplotlib not available - skipping visualization")
    
    def analyze_dataset_statistics(self, dataset_path: str):
        """Analyze generated dataset statistics"""
        
        # Load dataset
        ref_images = np.load(os.path.join(dataset_path, "reference_images.npy"))
        prompts = np.load(os.path.join(dataset_path, "scene_prompts.npy"))
        sketches = np.load(os.path.join(dataset_path, "scene_sketches.npy"))
        
        with open(os.path.join(dataset_path, "scene_metadata.pkl"), "rb") as f:
            metadata = pickle.load(f)
        
        print("=== Dataset Statistics ===")
        print(f"Total samples: {len(ref_images)}")
        print(f"Reference images shape: {ref_images.shape}")
        print(f"Prompts shape: {prompts.shape}")
        print(f"Sketches shape: {sketches.shape}")
        
        # Analyze complexity distribution
        complexity_counts = {}
        for meta in metadata:
            complexity = meta.get("complexity", "unknown")
            complexity_counts[complexity] = complexity_counts.get(complexity, 0) + 1
        
        print(f"\nComplexity distribution:")
        for complexity, count in complexity_counts.items():
            print(f"  {complexity}: {count} samples")
        
        # Analyze character and asset distributions
        char_counts = {}
        asset_counts = {}
        bg_counts = {}
        
        for meta in metadata:
            # Character types
            for char in meta.get("characters", []):
                char_type = char["type"]
                char_counts[char_type] = char_counts.get(char_type, 0) + 1
            
            # Asset types
            for asset in meta.get("assets", []):
                asset_type = asset["type"]
                asset_counts[asset_type] = asset_counts.get(asset_type, 0) + 1
            
            # Background types
            bg_type = meta.get("background", {}).get("type", "unknown")
            bg_counts[bg_type] = bg_counts.get(bg_type, 0) + 1
        
        print(f"\nCharacter type distribution:")
        for char_type, count in sorted(char_counts.items()):
            print(f"  {char_type}: {count}")
        
        print(f"\nAsset type distribution:")
        for asset_type, count in sorted(asset_counts.items()):
            print(f"  {asset_type}: {count}")
        
        print(f"\nBackground type distribution:")
        for bg_type, count in sorted(bg_counts.items()):
            print(f"  {bg_type}: {count}")
        
        # Analyze prompt encoding statistics
        print(f"\nPrompt encoding statistics:")
        print(f"  Min value: {prompts.min():.3f}")
        print(f"  Max value: {prompts.max():.3f}")
        print(f"  Mean value: {prompts.mean():.3f}")
        print(f"  Non-zero rate: {(prompts > 0).mean():.3f}")
        
        # Analyze sketch statistics
        print(f"\nSketch statistics:")
        print(f"  Min value: {sketches.min():.3f}")
        print(f"  Max value: {sketches.max():.3f}")
        print(f"  Mean value: {sketches.mean():.3f}")
        print(f"  Non-zero pixels per sketch: {(sketches > 0).sum(axis=(1,2)).mean():.1f}")
        
        return {
            "complexity_counts": complexity_counts,
            "char_counts": char_counts,
            "asset_counts": asset_counts,
            "bg_counts": bg_counts
        }

# Test the complete pipeline
def main():
    print("=== Scene Training Pipeline Test ===")
    
    # Initialize generator
    generator = SceneTrainingDataGenerator(
        canvas_size=256, 
        ref_size=64, 
        num_references=12
    )
    
    # Test single sample generation
    print("\n1. Testing single sample generation...")
    for complexity in ["simple", "medium", "complex"]:
        print(f"\nGenerating {complexity} scene...")
        ref_images, prompt, sketch, metadata = generator.generate_training_sample(complexity)
        
        print(f"  Reference images: {len(ref_images)} x {ref_images[0].shape}")
        print(f"  Prompt shape: {prompt.shape}")
        print(f"  Sketch shape: {sketch.shape}")
        print(f"  Characters: {len(metadata['characters'])}")
        print(f"  Assets: {len(metadata['assets'])}")
        print(f"  Background: {metadata['background']['type']}")
    
    # Generate test visualizations
    print("\n2. Generating test visualizations...")
    generator.generate_test_visualizations(num_samples=2)
    
    # Generate small dataset for testing
    print("\n3. Generating small test dataset...")
    ref_images, prompts, sketches, metadata = generator.generate_curriculum_dataset(
        samples_per_level=100,  # Small dataset for testing
        save_path="test_scene_dataset"
    )
    
    # Analyze dataset
    print("\n4. Analyzing dataset...")
    stats = generator.analyze_dataset_statistics("test_scene_dataset")
    
    print("\n=== Pipeline Test Complete ===")
    print("Ready for model training!")

if __name__ == "__main__":
    main()
```

```python
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader
import numpy as np
import matplotlib.pyplot as plt
import os
import json
import pickle
from scene_model_architecture import SceneGenerationModel, SceneDataset, SceneLoss, count_parameters

class SceneModelTrainer:
    def __init__(self, model, device='cuda', learning_rate=0.001):  # Increased LR from 0.0001
        self.model = model.to(device)
        self.device = device
        
        # Simplified optimizer with higher learning rate
        self.optimizer = optim.Adam(
            self.model.parameters(), 
            lr=learning_rate, 
            weight_decay=1e-4  # Reduced weight decay
        )
        
        # Learning rate scheduler
        self.scheduler = optim.lr_scheduler.StepLR(
            self.optimizer, step_size=20, gamma=0.5  # Reduce LR every 20 epochs
        )
        
        # Fixed loss function
        self.criterion = SceneLoss(sketch_weight=1.0, attention_weight=0.01)
        
        # Training history
        self.train_losses = []
        self.val_losses = []
        self.best_val_loss = float('inf')
        
    def train_epoch(self, train_loader, epoch):
        """Train for one epoch"""
        self.model.train()
        total_loss = 0.0
        total_sketch_loss = 0.0
        num_batches = len(train_loader)
        
        for batch_idx, (ref_images, prompts, target_sketches) in enumerate(train_loader):
            ref_images = ref_images.to(self.device)
            prompts = prompts.to(self.device)
            target_sketches = target_sketches.to(self.device)
            
            # Forward pass
            self.optimizer.zero_grad()
            pred_sketches, global_weights, spatial_maps = self.model(ref_images, prompts)
            
            # Compute loss
            total_loss_batch, sketch_loss_batch = self.criterion(
                pred_sketches, target_sketches, global_weights
            )
            
            # Backward pass
            total_loss_batch.backward()
            
            # Gradient clipping
            torch.nn.utils.clip_grad_norm_(self.model.parameters(), max_norm=1.0)
            
            self.optimizer.step()
            
            total_loss += total_loss_batch.item()
            total_sketch_loss += sketch_loss_batch.item()
            
            # Debug output for first few batches
            if epoch == 0 and batch_idx < 3:
                print(f"  Batch {batch_idx}: Loss = {total_loss_batch.item():.4f}")
                print(f"    Pred sketch range: [{pred_sketches.min():.3f}, {pred_sketches.max():.3f}]")
                print(f"    Target sketch range: [{target_sketches.min():.3f}, {target_sketches.max():.3f}]")
                print(f"    Global attention shape: {global_weights.shape}")
            
            # Progress update
            if batch_idx % (num_batches // 5) == 0:
                print(f"  Batch {batch_idx}/{num_batches}, Loss: {total_loss_batch.item():.4f}")
        
        avg_loss = total_loss / num_batches
        avg_sketch_loss = total_sketch_loss / num_batches
        
        return avg_loss, avg_sketch_loss
    
    def validate_epoch(self, val_loader):
        """Validate for one epoch"""
        self.model.eval()
        total_loss = 0.0
        total_sketch_loss = 0.0
        num_batches = len(val_loader)
        
        with torch.no_grad():
            for ref_images, prompts, target_sketches in val_loader:
                ref_images = ref_images.to(self.device)
                prompts = prompts.to(self.device)
                target_sketches = target_sketches.to(self.device)
                
                # Forward pass
                pred_sketches, global_weights, spatial_maps = self.model(ref_images, prompts)
                
                # Compute loss
                total_loss_batch, sketch_loss_batch = self.criterion(
                    pred_sketches, target_sketches, global_weights
                )
                
                total_loss += total_loss_batch.item()
                total_sketch_loss += sketch_loss_batch.item()
        
        avg_loss = total_loss / num_batches
        avg_sketch_loss = total_sketch_loss / num_batches
        
        return avg_loss, avg_sketch_loss
    
    def train(self, train_loader, val_loader, num_epochs=80, save_path="scene_model_checkpoints"):
        """Full training loop"""
        
        if not os.path.exists(save_path):
            os.makedirs(save_path)
        
        print(f"Starting training for {num_epochs} epochs...")
        print(f"Model parameters: {count_parameters(self.model):,}")
        print(f"Device: {self.device}")
        
        for epoch in range(num_epochs):
            print(f"\nEpoch {epoch+1}/{num_epochs}")
            
            # Training
            train_loss, train_sketch_loss = self.train_epoch(train_loader, epoch)
            
            # Validation
            val_loss, val_sketch_loss = self.validate_epoch(val_loader)
            
            # Update learning rate
            self.scheduler.step()  # Changed from scheduler.step(val_loss)
            
            # Save losses
            self.train_losses.append(train_loss)
            self.val_losses.append(val_loss)
            
            print(f"  Train Loss: {train_loss:.4f} (Sketch: {train_sketch_loss:.4f})")
            print(f"  Val Loss: {val_loss:.4f} (Sketch: {val_sketch_loss:.4f})")
            print(f"  LR: {self.optimizer.param_groups[0]['lr']:.6f}")
            
            # Save best model
            if val_loss < self.best_val_loss:
                self.best_val_loss = val_loss
                torch.save({
                    'epoch': epoch,
                    'model_state_dict': self.model.state_dict(),
                    'optimizer_state_dict': self.optimizer.state_dict(),
                    'train_losses': self.train_losses,
                    'val_losses': self.val_losses,
                    'best_val_loss': self.best_val_loss
                }, os.path.join(save_path, 'best_scene_model.pth'))
                print(f"  ✓ Best model saved (Val Loss: {val_loss:.4f})")
            
            # Save checkpoint every 10 epochs
            if (epoch + 1) % 10 == 0:
                torch.save({
                    'epoch': epoch,
                    'model_state_dict': self.model.state_dict(),
                    'optimizer_state_dict': self.optimizer.state_dict(),
                    'train_losses': self.train_losses,
                    'val_losses': self.val_losses
                }, os.path.join(save_path, f'scene_model_epoch_{epoch+1}.pth'))
                
                # Plot training curves
                self.plot_training_curves(os.path.join(save_path, f'training_curves_epoch_{epoch+1}.png'))
        
        print(f"\nTraining completed!")
        print(f"Best validation loss: {self.best_val_loss:.4f}")
        
        return self.train_losses, self.val_losses
    
    def plot_training_curves(self, save_path):
        """Plot and save training curves"""
        plt.figure(figsize=(12, 5))
        
        plt.subplot(1, 2, 1)
        plt.plot(self.train_losses, label='Train Loss')
        plt.plot(self.val_losses, label='Validation Loss')
        plt.xlabel('Epoch')
        plt.ylabel('Loss')
        plt.title('Training Progress')
        plt.legend()
        plt.grid(True)
        
        plt.subplot(1, 2, 2)
        # Plot recent epochs only for better detail
        recent_epochs = min(20, len(self.train_losses))
        recent_train = self.train_losses[-recent_epochs:]
        recent_val = self.val_losses[-recent_epochs:]
        
        plt.plot(range(len(self.train_losses)-recent_epochs, len(self.train_losses)), recent_train, label='Train Loss (Recent)')
        plt.plot(range(len(self.val_losses)-recent_epochs, len(self.val_losses)), recent_val, label='Val Loss (Recent)')
        plt.xlabel('Epoch')
        plt.ylabel('Loss')
        plt.title('Recent Training Progress')
        plt.legend()
        plt.grid(True)
        
        plt.tight_layout()
        plt.savefig(save_path, dpi=150, bbox_inches='tight')
        plt.close()

def visualize_scene_predictions(model, val_loader, device, num_samples=3, save_path="scene_predictions"):
    """Visualize model predictions on validation data"""
    model.eval()
    
    if not os.path.exists(save_path):
        os.makedirs(save_path)
    
    with torch.no_grad():
        ref_images, prompts, target_sketches = next(iter(val_loader))
        ref_images = ref_images.to(device)
        prompts = prompts.to(device)
        target_sketches = target_sketches.to(device)
        
        pred_sketches, global_weights, spatial_maps = model(ref_images, prompts)
        
        for sample_idx in range(min(num_samples, len(ref_images))):
            fig, axes = plt.subplots(4, 4, figsize=(16, 16))
            
            # Show reference images (top 3 rows)
            ref_titles = [
                "Char Head", "Char Body", "Interaction", "Asset 1",
                "Asset 2", "Background", "Layout", "Char 2", 
                "Extra 1", "Extra 2", "Extra 3", "Extra 4"
            ]
            
            for ref_idx in range(12):
                row = ref_idx // 4
                col = ref_idx % 4
                axes[row, col].imshow(ref_images[sample_idx, ref_idx].cpu().numpy(), cmap='gray')
                
                # Show attention weight in title
                if ref_idx < global_weights.shape[2]:
                    weight = global_weights[sample_idx, 0, ref_idx].cpu().item()
                    axes[row, col].set_title(f'{ref_titles[ref_idx]}\n(Attn: {weight:.3f})', fontsize=10)
                else:
                    axes[row, col].set_title(ref_titles[ref_idx], fontsize=10)
                
                axes[row, col].axis('off')
            
            # Show predictions vs targets (bottom row)
            axes[3, 0].imshow(target_sketches[sample_idx, 0].cpu().numpy(), cmap='gray')
            axes[3, 0].set_title('Target Scene Sketch', fontsize=12)
            axes[3, 0].axis('off')
            
            axes[3, 1].imshow(pred_sketches[sample_idx, 0].cpu().numpy(), cmap='gray')
            axes[3, 1].set_title('Predicted Scene Sketch', fontsize=12)
            axes[3, 1].axis('off')
            
            # Show difference map
            diff = torch.abs(target_sketches[sample_idx, 0] - pred_sketches[sample_idx, 0])
            axes[3, 2].imshow(diff.cpu().numpy(), cmap='hot')
            axes[3, 2].set_title('Difference Map', fontsize=12)
            axes[3, 2].axis('off')
            
            # Show global attention weights
            weights = global_weights[sample_idx, 0].cpu().numpy()
            axes[3, 3].bar(range(len(weights)), weights)
            axes[3, 3].set_title('Global Attention Weights', fontsize=12)
            axes[3, 3].set_xlabel('Reference Image')
            axes[3, 3].set_ylabel('Attention Weight')
            axes[3, 3].set_xticks(range(0, len(weights), 2))
            
            plt.tight_layout()
            plt.savefig(f'{save_path}/scene_prediction_sample_{sample_idx}.png', dpi=150, bbox_inches='tight')
            plt.close()
    
    print(f"Scene predictions saved to {save_path}")

def load_scene_dataset(dataset_path):
    """Load scene dataset from saved files"""
    print(f"Loading dataset from {dataset_path}...")
    
    # Load arrays
    ref_images = np.load(os.path.join(dataset_path, "reference_images.npy"))
    prompts = np.load(os.path.join(dataset_path, "scene_prompts.npy"))
    sketches = np.load(os.path.join(dataset_path, "scene_sketches.npy"))
    
    # Load metadata
    with open(os.path.join(dataset_path, "scene_metadata.pkl"), "rb") as f:
        metadata = pickle.load(f)
    
    # Load dataset info
    with open(os.path.join(dataset_path, "dataset_info.json"), "r") as f:
        dataset_info = json.load(f)
    
    print(f"Dataset loaded:")
    print(f"  Reference images: {ref_images.shape}")
    print(f"  Prompts: {prompts.shape}")
    print(f"  Sketches: {sketches.shape}")
    print(f"  Total samples: {len(metadata)}")
    
    return ref_images, prompts, sketches, metadata, dataset_info

def create_curriculum_dataloaders(ref_images, prompts, sketches, metadata, batch_size=8, val_split=0.2):
    """Create data loaders with curriculum learning organization"""
    
    # Organize by complexity
    simple_indices = []
    medium_indices = []
    complex_indices = []
    
    for i, meta in enumerate(metadata):
        complexity = meta.get("complexity", "simple")
        if complexity == "simple":
            simple_indices.append(i)
        elif complexity == "medium":
            medium_indices.append(i)
        else:
            complex_indices.append(i)
    
    print(f"Dataset split by complexity:")
    print(f"  Simple: {len(simple_indices)} samples")
    print(f"  Medium: {len(medium_indices)} samples")
    print(f"  Complex: {len(complex_indices)} samples")
    
    # For initial training, we'll use all complexities mixed
    # In practice, you might want to train curriculum: simple -> medium -> complex
    all_indices = simple_indices + medium_indices + complex_indices
    np.random.shuffle(all_indices)
    
    # Train/validation split
    n_val = int(len(all_indices) * val_split)
    val_indices = all_indices[:n_val]
    train_indices = all_indices[n_val:]
    
    print(f"Train/Val split:")
    print(f"  Train: {len(train_indices)} samples")
    print(f"  Validation: {len(val_indices)} samples")
    
    # Create datasets
    train_dataset = SceneDataset(
        ref_images[train_indices],
        prompts[train_indices],
        sketches[train_indices]
    )
    
    val_dataset = SceneDataset(
        ref_images[val_indices],
        prompts[val_indices],
        sketches[val_indices]
    )
    
    # Create data loaders (single-threaded to avoid multiprocessing issues)
    train_loader = DataLoader(
        train_dataset, 
        batch_size=batch_size, 
        shuffle=True, 
        num_workers=0,
        pin_memory=False
    )
    
    val_loader = DataLoader(
        val_dataset, 
        batch_size=batch_size, 
        shuffle=False, 
        num_workers=0,
        pin_memory=False
    )
    
    return train_loader, val_loader

def main():
    print("=== Scene Model Training ===")
    
    # Configuration
    device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
    print(f"Using device: {device}")
    
    dataset_path = "test_scene_dataset"  # Use the test dataset we generated
    batch_size = 6  # Slightly larger batch size for better gradient estimates
    learning_rate = 0.001  # Much higher learning rate for faster initial learning
    num_epochs = 80  # Enough epochs to see convergence
    
    # Load dataset
    ref_images, prompts, sketches, metadata, dataset_info = load_scene_dataset(dataset_path)
    
    # Create data loaders
    train_loader, val_loader = create_curriculum_dataloaders(
        ref_images, prompts, sketches, metadata, 
        batch_size=batch_size, val_split=0.2
    )
    
    # Create model
    prompt_dim = dataset_info["prompt_dim"]
    num_refs = dataset_info["num_references"]
    
    model = SceneGenerationModel(
        prompt_dim=prompt_dim,
        feature_dim=256,
        num_refs=num_refs
    )
    
    print(f"Model created with {count_parameters(model):,} parameters")
    
    # Create trainer
    trainer = SceneModelTrainer(
        model=model,
        device=device,
        learning_rate=learning_rate
    )
    
    # Test model with one batch first
    print("\nTesting model with one batch...")
    test_ref, test_prompts, test_sketches = next(iter(train_loader))
    test_ref = test_ref.to(device)
    test_prompts = test_prompts.to(device)
    test_sketches = test_sketches.to(device)
    
    with torch.no_grad():
        pred_sketches, global_weights, spatial_maps = model(test_ref, test_prompts)
        print(f"Test forward pass successful!")
        print(f"  Input: {test_ref.shape} -> Output: {pred_sketches.shape}")
        print(f"  Global attention: {global_weights.shape}")
        print(f"  Spatial maps: {len(spatial_maps)} batches")
    
    # Train model
    print(f"\nStarting training...")
    train_losses, val_losses = trainer.train(
        train_loader=train_loader,
        val_loader=val_loader,
        num_epochs=num_epochs,
        save_path="scene_model_checkpoints"
    )
    
    # Load best model for evaluation
    print("\nLoading best model for evaluation...")
    checkpoint = torch.load("scene_model_checkpoints/best_scene_model.pth")
    model.load_state_dict(checkpoint['model_state_dict'])
    
    # Generate predictions
    print("Generating prediction visualizations...")
    visualize_scene_predictions(
        model=model,
        val_loader=val_loader,
        device=device,
        num_samples=5,
        save_path="scene_predictions"
    )
    
    # Final training curves
    trainer.plot_training_curves("final_training_curves.png")
    
    print("=== Training Complete ===")
    print(f"Best validation loss: {trainer.best_val_loss:.4f}")
    print("Check 'scene_predictions/' for generated scene sketches!")

def test_curriculum_training():
    """Test curriculum training approach: simple -> medium -> complex"""
    print("=== Testing Curriculum Training ===")
    
    device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
    dataset_path = "test_scene_dataset"
    
    # Load dataset
    ref_images, prompts, sketches, metadata, dataset_info = load_scene_dataset(dataset_path)
    
    # Separate by complexity
    complexity_data = {"simple": [], "medium": [], "complex": []}
    
    for i, meta in enumerate(metadata):
        complexity = meta.get("complexity", "simple")
        complexity_data[complexity].append(i)
    
    # Create model
    model = SceneGenerationModel(
        prompt_dim=dataset_info["prompt_dim"],
        feature_dim=256,
        num_refs=dataset_info["num_references"]
    )
    
    trainer = SceneModelTrainer(model=model, device=device, learning_rate=0.0002)
    
    # Stage 1: Train on simple scenes
    print("\n=== Stage 1: Training on Simple Scenes ===")
    simple_indices = complexity_data["simple"]
    train_split = int(0.8 * len(simple_indices))
    
    simple_train_dataset = SceneDataset(
        ref_images[simple_indices[:train_split]],
        prompts[simple_indices[:train_split]],
        sketches[simple_indices[:train_split]]
    )
    simple_val_dataset = SceneDataset(
        ref_images[simple_indices[train_split:]],
        prompts[simple_indices[train_split:]],
        sketches[simple_indices[train_split:]]
    )
    
    simple_train_loader = DataLoader(simple_train_dataset, batch_size=4, shuffle=True, num_workers=0)
    simple_val_loader = DataLoader(simple_val_dataset, batch_size=4, shuffle=False, num_workers=0)
    
    # Train on simple for fewer epochs
    trainer.train(simple_train_loader, simple_val_loader, num_epochs=20, save_path="curriculum_stage1")
    
    # Stage 2: Fine-tune on medium complexity
    print("\n=== Stage 2: Fine-tuning on Medium Scenes ===")
    # Reduce learning rate for fine-tuning
    for param_group in trainer.optimizer.param_groups:
        param_group['lr'] *= 0.5
    
    medium_indices = complexity_data["medium"]
    train_split = int(0.8 * len(medium_indices))
    
    medium_train_dataset = SceneDataset(
        ref_images[medium_indices[:train_split]],
        prompts[medium_indices[:train_split]],
        sketches[medium_indices[:train_split]]
    )
    medium_val_dataset = SceneDataset(
        ref_images[medium_indices[train_split:]],
        prompts[medium_indices[train_split:]],
        sketches[medium_indices[train_split:]]
    )
    
    medium_train_loader = DataLoader(medium_train_dataset, batch_size=4, shuffle=True, num_workers=0)
    medium_val_loader = DataLoader(medium_val_dataset, batch_size=4, shuffle=False, num_workers=0)
    
    trainer.train(medium_train_loader, medium_val_loader, num_epochs=15, save_path="curriculum_stage2")
    
    # Stage 3: Fine-tune on complex scenes
    print("\n=== Stage 3: Fine-tuning on Complex Scenes ===")
    for param_group in trainer.optimizer.param_groups:
        param_group['lr'] *= 0.5
    
    complex_indices = complexity_data["complex"]
    train_split = int(0.8 * len(complex_indices))
    
    complex_train_dataset = SceneDataset(
        ref_images[complex_indices[:train_split]],
        prompts[complex_indices[:train_split]],
        sketches[complex_indices[:train_split]]
    )
    complex_val_dataset = SceneDataset(
        ref_images[complex_indices[train_split:]],
        prompts[complex_indices[train_split:]],
        sketches[complex_indices[train_split:]]
    )
    
    complex_train_loader = DataLoader(complex_train_dataset, batch_size=4, shuffle=True, num_workers=0)
    complex_val_loader = DataLoader(complex_val_dataset, batch_size=4, shuffle=False, num_workers=0)
    
    trainer.train(complex_train_loader, complex_val_loader, num_epochs=20, save_path="curriculum_stage3")
    
    print("=== Curriculum Training Complete ===")

if __name__ == "__main__":
    # Run main training
    # main()
    
    # Optionally test curriculum approach
    test_curriculum_training()
```

```python
import numpy as np
import cv2
import random
from typing import Dict, List, Tuple, Optional
import os
import json
import pickle
from dataclasses import dataclass
from enum import Enum

class SceneType(Enum):
    PARK = "park"
    CITY = "city" 
    KITCHEN = "kitchen"
    GARDEN = "garden"
    OFFICE = "office"
    PLAYGROUND = "playground"

class ObjectType(Enum):
    # Furniture/Structures
    TABLE = "table"
    CHAIR = "chair"
    BUILDING = "building"
    HOUSE = "house"
    BENCH = "bench"
    
    # Nature
    TREE = "tree"
    FLOWER = "flower"
    CLOUD = "cloud"
    SUN = "sun"
    ROCK = "rock"
    
    # Vehicles/Objects
    CAR = "car"
    BIKE = "bike"
    BALL = "ball"
    UMBRELLA = "umbrella"
    LAMP = "lamp"

class PatternType(Enum):
    SOLID = "solid"
    STRIPES = "stripes"
    DOTS = "dots"
    GRID = "grid"
    WAVES = "waves"
    ZIGZAG = "zigzag"

class ArrangementType(Enum):
    LINEAR = "linear"
    CIRCULAR = "circular"
    SCATTERED = "scattered"
    CLUSTERED = "clustered"
    GRID_LAYOUT = "grid_layout"

@dataclass
class SceneObject:
    obj_type: ObjectType
    position: Tuple[int, int]
    size: int
    pattern: PatternType
    arrangement_role: str  # "main", "secondary", "background"
    color_intensity: float  # 0.3 to 1.0 for different shades

@dataclass
class Scene:
    scene_type: SceneType
    objects: List[SceneObject]
    overall_arrangement: ArrangementType
    dominant_pattern: PatternType
    scene_style: str  # "minimal", "dense", "balanced"
    canvas_size: int

class FullSceneGenerator:
    def __init__(self, canvas_size=256):
        self.canvas_size = canvas_size
        
        # Define scene templates with typical objects
        self.scene_templates = {
            SceneType.PARK: {
                "main_objects": [ObjectType.TREE, ObjectType.BENCH, ObjectType.FLOWER],
                "secondary_objects": [ObjectType.CLOUD, ObjectType.SUN, ObjectType.ROCK],
                "typical_arrangements": [ArrangementType.SCATTERED, ArrangementType.LINEAR],
                "common_patterns": [PatternType.SOLID, PatternType.DOTS, PatternType.WAVES]
            },
            SceneType.CITY: {
                "main_objects": [ObjectType.BUILDING, ObjectType.CAR, ObjectType.HOUSE],
                "secondary_objects": [ObjectType.LAMP, ObjectType.UMBRELLA, ObjectType.BENCH],
                "typical_arrangements": [ArrangementType.GRID_LAYOUT, ArrangementType.LINEAR],
                "common_patterns": [PatternType.STRIPES, PatternType.GRID, PatternType.SOLID]
            },
            SceneType.KITCHEN: {
                "main_objects": [ObjectType.TABLE, ObjectType.CHAIR],
                "secondary_objects": [ObjectType.LAMP, ObjectType.BALL],
                "typical_arrangements": [ArrangementType.CLUSTERED, ArrangementType.LINEAR],
                "common_patterns": [PatternType.GRID, PatternType.SOLID, PatternType.STRIPES]
            },
            SceneType.GARDEN: {
                "main_objects": [ObjectType.FLOWER, ObjectType.TREE, ObjectType.ROCK],
                "secondary_objects": [ObjectType.BENCH, ObjectType.SUN, ObjectType.UMBRELLA],
                "typical_arrangements": [ArrangementType.SCATTERED, ArrangementType.CIRCULAR],
                "common_patterns": [PatternType.DOTS, PatternType.WAVES, PatternType.SOLID]
            },
            SceneType.OFFICE: {
                "main_objects": [ObjectType.TABLE, ObjectType.CHAIR, ObjectType.LAMP],
                "secondary_objects": [ObjectType.BUILDING, ObjectType.BALL],
                "typical_arrangements": [ArrangementType.GRID_LAYOUT, ArrangementType.LINEAR],
                "common_patterns": [PatternType.SOLID, PatternType.STRIPES, PatternType.GRID]
            },
            SceneType.PLAYGROUND: {
                "main_objects": [ObjectType.BENCH, ObjectType.BALL, ObjectType.TREE],
                "secondary_objects": [ObjectType.UMBRELLA, ObjectType.FLOWER, ObjectType.ROCK],
                "typical_arrangements": [ArrangementType.SCATTERED, ArrangementType.CIRCULAR],
                "common_patterns": [PatternType.DOTS, PatternType.STRIPES, PatternType.WAVES]
            }
        }
        
        # Object drawing parameters
        self.object_base_sizes = {
            ObjectType.TREE: 25,
            ObjectType.BUILDING: 35,
            ObjectType.HOUSE: 30,
            ObjectType.CAR: 20,
            ObjectType.TABLE: 25,
            ObjectType.CHAIR: 15,
            ObjectType.BENCH: 20,
            ObjectType.FLOWER: 8,
            ObjectType.BALL: 10,
            ObjectType.ROCK: 12,
            ObjectType.CLOUD: 20,
            ObjectType.SUN: 18,
            ObjectType.BIKE: 18,
            ObjectType.UMBRELLA: 15,
            ObjectType.LAMP: 12
        }
    
    def generate_scene(self, scene_type: SceneType = None, style: str = None) -> Scene:
        """Generate a complete scene with multiple objects"""
        
        if scene_type is None:
            scene_type = random.choice(list(SceneType))
        
        if style is None:
            style = random.choice(["minimal", "dense", "balanced"])
        
        template = self.scene_templates[scene_type]
        
        # Determine number of objects based on style
        if style == "minimal":
            num_main = random.randint(1, 2)
            num_secondary = random.randint(0, 2)
        elif style == "balanced":
            num_main = random.randint(2, 3)
            num_secondary = random.randint(1, 3)
        else:  # dense
            num_main = random.randint(3, 4)
            num_secondary = random.randint(2, 4)
        
        # Select objects
        main_objects = random.sample(template["main_objects"], 
                                   min(num_main, len(template["main_objects"])))
        secondary_objects = random.sample(template["secondary_objects"], 
                                        min(num_secondary, len(template["secondary_objects"])))
        
        # Choose scene-wide properties
        overall_arrangement = random.choice(template["typical_arrangements"])
        dominant_pattern = random.choice(template["common_patterns"])
        
        # Generate object positions based on arrangement
        all_object_types = main_objects + secondary_objects
        positions = self._generate_positions(all_object_types, overall_arrangement)
        
        # Create scene objects
        scene_objects = []
        
        for i, obj_type in enumerate(main_objects):
            scene_obj = SceneObject(
                obj_type=obj_type,
                position=positions[i],
                size=self.object_base_sizes[obj_type] + random.randint(-5, 8),
                pattern=self._choose_object_pattern(dominant_pattern, is_main=True),
                arrangement_role="main",
                color_intensity=random.uniform(0.7, 1.0)
            )
            scene_objects.append(scene_obj)
        
        for i, obj_type in enumerate(secondary_objects):
            scene_obj = SceneObject(
                obj_type=obj_type,
                position=positions[len(main_objects) + i],
                size=self.object_base_sizes[obj_type] + random.randint(-3, 5),
                pattern=self._choose_object_pattern(dominant_pattern, is_main=False),
                arrangement_role="secondary",
                color_intensity=random.uniform(0.4, 0.8)
            )
            scene_objects.append(scene_obj)
        
        return Scene(
            scene_type=scene_type,
            objects=scene_objects,
            overall_arrangement=overall_arrangement,
            dominant_pattern=dominant_pattern,
            scene_style=style,
            canvas_size=self.canvas_size
        )
    
    def _generate_positions(self, object_types: List[ObjectType], 
                          arrangement: ArrangementType) -> List[Tuple[int, int]]:
        """Generate positions for objects based on arrangement type"""
        num_objects = len(object_types)
        positions = []
        
        # Define safe zones (avoid edges)
        margin = 40
        safe_width = self.canvas_size - 2 * margin
        safe_height = self.canvas_size - 2 * margin
        
        if arrangement == ArrangementType.LINEAR:
            # Objects in a line (horizontal or vertical)
            if random.random() < 0.7:  # Horizontal line
                y = margin + safe_height // 2 + random.randint(-20, 20)
                x_spacing = safe_width // (num_objects + 1)
                for i in range(num_objects):
                    x = margin + x_spacing * (i + 1) + random.randint(-10, 10)
                    positions.append((x, y))
            else:  # Vertical line
                x = margin + safe_width // 2 + random.randint(-20, 20)
                y_spacing = safe_height // (num_objects + 1)
                for i in range(num_objects):
                    y = margin + y_spacing * (i + 1) + random.randint(-10, 10)
                    positions.append((x, y))
        
        elif arrangement == ArrangementType.CIRCULAR:
            # Objects arranged in a circle
            center_x = margin + safe_width // 2
            center_y = margin + safe_height // 2
            radius = min(safe_width, safe_height) // 3
            
            for i in range(num_objects):
                angle = 2 * np.pi * i / num_objects + random.uniform(-0.3, 0.3)
                x = center_x + int(radius * np.cos(angle))
                y = center_y + int(radius * np.sin(angle))
                positions.append((x, y))
        
        elif arrangement == ArrangementType.GRID_LAYOUT:
            # Objects in a grid pattern
            cols = int(np.ceil(np.sqrt(num_objects)))
            rows = int(np.ceil(num_objects / cols))
            
            x_spacing = safe_width // (cols + 1)
            y_spacing = safe_height // (rows + 1)
            
            for i in range(num_objects):
                row = i // cols
                col = i % cols
                x = margin + x_spacing * (col + 1) + random.randint(-15, 15)
                y = margin + y_spacing * (row + 1) + random.randint(-15, 15)
                positions.append((x, y))
        
        elif arrangement == ArrangementType.CLUSTERED:
            # Objects grouped in 1-2 clusters
            num_clusters = random.randint(1, min(2, num_objects))
            cluster_centers = []
            
            for _ in range(num_clusters):
                center_x = margin + random.randint(safe_width // 4, 3 * safe_width // 4)
                center_y = margin + random.randint(safe_height // 4, 3 * safe_height // 4)
                cluster_centers.append((center_x, center_y))
            
            objects_per_cluster = num_objects // num_clusters
            remaining_objects = num_objects % num_clusters
            
            obj_idx = 0
            for cluster_idx, (center_x, center_y) in enumerate(cluster_centers):
                cluster_size = objects_per_cluster + (1 if cluster_idx < remaining_objects else 0)
                
                for _ in range(cluster_size):
                    if obj_idx < num_objects:
                        x = center_x + random.randint(-30, 30)
                        y = center_y + random.randint(-30, 30)
                        x = max(margin, min(self.canvas_size - margin, x))
                        y = max(margin, min(self.canvas_size - margin, y))
                        positions.append((x, y))
                        obj_idx += 1
        
        else:  # SCATTERED
            # Random positions avoiding overlap
            positions = []
            for _ in range(num_objects):
                attempts = 0
                while attempts < 20:
                    x = margin + random.randint(0, safe_width)
                    y = margin + random.randint(0, safe_height)
                    
                    # Check for minimum distance from existing positions
                    min_distance = 35
                    too_close = False
                    for existing_x, existing_y in positions:
                        if np.sqrt((x - existing_x)**2 + (y - existing_y)**2) < min_distance:
                            too_close = True
                            break
                    
                    if not too_close:
                        positions.append((x, y))
                        break
                    
                    attempts += 1
                
                # If we couldn't find a good position, just place it randomly
                if attempts == 20:
                    x = margin + random.randint(0, safe_width)
                    y = margin + random.randint(0, safe_height)
                    positions.append((x, y))
        
        return positions
    
    def _choose_object_pattern(self, dominant_pattern: PatternType, is_main: bool) -> PatternType:
        """Choose pattern for an object based on scene's dominant pattern"""
        if is_main:
            # Main objects more likely to use dominant pattern
            if random.random() < 0.7:
                return dominant_pattern
            else:
                return random.choice([p for p in PatternType if p != dominant_pattern])
        else:
            # Secondary objects more varied
            if random.random() < 0.4:
                return dominant_pattern
            else:
                return random.choice(list(PatternType))
    
    def render_scene(self, scene: Scene) -> np.ndarray:
        """Render a scene to a numpy array"""
        canvas = np.zeros((self.canvas_size, self.canvas_size), dtype=np.float32)
        
        # Add subtle background texture based on scene type
        self._add_background(canvas, scene.scene_type)
        
        # Sort objects by rendering order (background to foreground)
        sorted_objects = sorted(scene.objects, 
                              key=lambda obj: (obj.arrangement_role == "main", obj.position[1]))
        
        # Render each object
        for obj in sorted_objects:
            self._draw_object(canvas, obj)
        
        return canvas
    
    def _add_background(self, canvas: np.ndarray, scene_type: SceneType):
        """Add subtle background elements"""
        if scene_type == SceneType.PARK:
            # Ground line
            ground_y = int(self.canvas_size * 0.85)
            cv2.line(canvas, (0, ground_y), (self.canvas_size, ground_y), 0.3, 1)
            
            # Grass texture
            for _ in range(15):
                x = random.randint(0, self.canvas_size)
                y = random.randint(ground_y, self.canvas_size)
                cv2.line(canvas, (x, y), (x, y-3), 0.2, 1)
        
        elif scene_type == SceneType.CITY:
            # City grid background
            for i in range(0, self.canvas_size, 50):
                cv2.line(canvas, (i, 0), (i, self.canvas_size), 0.15, 1)
                cv2.line(canvas, (0, i), (self.canvas_size, i), 0.15, 1)
        
        elif scene_type == SceneType.KITCHEN:
            # Floor tiles
            tile_size = 25
            for i in range(0, self.canvas_size, tile_size):
                for j in range(0, self.canvas_size, tile_size):
                    if (i // tile_size + j // tile_size) % 2 == 0:
                        cv2.rectangle(canvas, (i, j), (i+tile_size, j+tile_size), 0.1, -1)
        
        elif scene_type == SceneType.GARDEN:
            # Organic curved paths
            pts = []
            for x in range(0, self.canvas_size, 20):
                y = self.canvas_size // 2 + int(20 * np.sin(x * 0.02))
                pts.append([x, y])
            if len(pts) > 1:
                pts = np.array(pts, np.int32)
                cv2.polylines(canvas, [pts], False, 0.2, 2)
    
    def _draw_object(self, canvas: np.ndarray, obj: SceneObject):
        """Draw a single object on the canvas"""
        x, y = obj.position
        size = obj.size
        pattern = obj.pattern
        intensity = obj.color_intensity
        
        # Draw base shape based on object type
        if obj.obj_type == ObjectType.TREE:
            self._draw_tree(canvas, x, y, size, pattern, intensity)
        elif obj.obj_type == ObjectType.BUILDING:
            self._draw_building(canvas, x, y, size, pattern, intensity)
        elif obj.obj_type == ObjectType.HOUSE:
            self._draw_house(canvas, x, y, size, pattern, intensity)
        elif obj.obj_type == ObjectType.CAR:
            self._draw_car(canvas, x, y, size, pattern, intensity)
        elif obj.obj_type == ObjectType.TABLE:
            self._draw_table(canvas, x, y, size, pattern, intensity)
        elif obj.obj_type == ObjectType.CHAIR:
            self._draw_chair(canvas, x, y, size, pattern, intensity)
        elif obj.obj_type == ObjectType.BENCH:
            self._draw_bench(canvas, x, y, size, pattern, intensity)
        elif obj.obj_type == ObjectType.FLOWER:
            self._draw_flower(canvas, x, y, size, pattern, intensity)
        elif obj.obj_type == ObjectType.BALL:
            self._draw_ball(canvas, x, y, size, pattern, intensity)
        elif obj.obj_type == ObjectType.ROCK:
            self._draw_rock(canvas, x, y, size, pattern, intensity)
        elif obj.obj_type == ObjectType.CLOUD:
            self._draw_cloud(canvas, x, y, size, pattern, intensity)
        elif obj.obj_type == ObjectType.SUN:
            self._draw_sun(canvas, x, y, size, pattern, intensity)
        elif obj.obj_type == ObjectType.BIKE:
            self._draw_bike(canvas, x, y, size, pattern, intensity)
        elif obj.obj_type == ObjectType.UMBRELLA:
            self._draw_umbrella(canvas, x, y, size, pattern, intensity)
        elif obj.obj_type == ObjectType.LAMP:
            self._draw_lamp(canvas, x, y, size, pattern, intensity)
    
    def _apply_pattern(self, canvas: np.ndarray, x: int, y: int, width: int, height: int, 
                      pattern: PatternType, intensity: float):
        """Apply pattern to a rectangular area"""
        if pattern == PatternType.STRIPES:
            for i in range(0, height, 4):
                if i % 8 < 4:  # Every other stripe
                    cv2.line(canvas, (x-width//2, y-height//2+i), 
                            (x+width//2, y-height//2+i), intensity, 2)
        
        elif pattern == PatternType.DOTS:
            for i in range(-width//2, width//2, 8):
                for j in range(-height//2, height//2, 8):
                    cv2.circle(canvas, (x+i, y+j), 1, intensity, -1)
        
        elif pattern == PatternType.GRID:
            for i in range(-width//2, width//2, 6):
                cv2.line(canvas, (x+i, y-height//2), (x+i, y+height//2), intensity, 1)
            for j in range(-height//2, height//2, 6):
                cv2.line(canvas, (x-width//2, y+j), (x+width//2, y+j), intensity, 1)
        
        elif pattern == PatternType.WAVES:
            for j in range(-height//2, height//2, 3):
                pts = []
                for i in range(-width//2, width//2, 2):
                    wave_y = y + j + int(2 * np.sin(i * 0.3))
                    pts.append([x + i, wave_y])
                if len(pts) > 1:
                    pts = np.array(pts, np.int32)
                    cv2.polylines(canvas, [pts], False, intensity, 1)
        
        elif pattern == PatternType.ZIGZAG:
            for j in range(-height//2, height//2, 6):
                for i in range(-width//2, width//2, 4):
                    if (i // 4) % 2 == 0:
                        cv2.line(canvas, (x+i, y+j), (x+i+2, y+j+2), intensity, 1)
                    else:
                        cv2.line(canvas, (x+i, y+j+2), (x+i+2, y+j), intensity, 1)
    
    # Object drawing methods
    def _draw_tree(self, canvas: np.ndarray, x: int, y: int, size: int, 
                  pattern: PatternType, intensity: float):
        # Trunk
        trunk_width = size // 4
        trunk_height = size // 2
        cv2.rectangle(canvas, (x-trunk_width//2, y), (x+trunk_width//2, y+trunk_height), 
                     intensity * 0.8, 2)
        
        # Crown
        crown_radius = size // 2
        cv2.circle(canvas, (x, y-trunk_height//2), crown_radius, intensity, 2)
        
        # Apply pattern to crown area
        if pattern != PatternType.SOLID:
            self._apply_pattern(canvas, x, y-trunk_height//2, crown_radius*2, crown_radius*2, 
                              pattern, intensity * 0.6)
    
    def _draw_building(self, canvas: np.ndarray, x: int, y: int, size: int, 
                      pattern: PatternType, intensity: float):
        # Main structure
        building_width = size
        building_height = int(size * 1.5)
        cv2.rectangle(canvas, (x-building_width//2, y-building_height), 
                     (x+building_width//2, y), intensity, 2)
        
        # Apply pattern
        if pattern != PatternType.SOLID:
            self._apply_pattern(canvas, x, y-building_height//2, building_width, building_height, 
                              pattern, intensity * 0.7)
        
        # Windows
        for i in range(3):
            for j in range(4):
                win_x = x - building_width//3 + i * building_width//3
                win_y = y - building_height + j * building_height//5 + building_height//10
                cv2.rectangle(canvas, (win_x-2, win_y-2), (win_x+2, win_y+2), intensity * 0.9, 1)
    
    def _draw_house(self, canvas: np.ndarray, x: int, y: int, size: int, 
                   pattern: PatternType, intensity: float):
        # Base
        house_width = size
        house_height = size // 2
        cv2.rectangle(canvas, (x-house_width//2, y-house_height), 
                     (x+house_width//2, y), intensity, 2)
        
        # Roof
        roof_pts = np.array([[x, y-house_height-size//3], 
                           [x-house_width//2, y-house_height],
                           [x+house_width//2, y-house_height]], np.int32)
        cv2.polylines(canvas, [roof_pts], True, intensity, 2)
        
        # Apply pattern to walls
        if pattern != PatternType.SOLID:
            self._apply_pattern(canvas, x, y-house_height//2, house_width, house_height, 
                              pattern, intensity * 0.6)
        
        # Door
        cv2.rectangle(canvas, (x-3, y-house_height//3), (x+3, y), intensity * 0.9, 2)
    
    def _draw_car(self, canvas: np.ndarray, x: int, y: int, size: int, 
                 pattern: PatternType, intensity: float):
        # Body
        car_width = size
        car_height = size // 2
        cv2.rectangle(canvas, (x-car_width//2, y-car_height), 
                     (x+car_width//2, y), intensity, 2)
        
        # Apply pattern to body
        if pattern != PatternType.SOLID:
            self._apply_pattern(canvas, x, y-car_height//2, car_width, car_height, 
                              pattern, intensity * 0.7)
        
        # Wheels
        wheel_radius = size // 6
        cv2.circle(canvas, (x-car_width//3, y), wheel_radius, intensity, 2)
        cv2.circle(canvas, (x+car_width//3, y), wheel_radius, intensity, 2)
        
        # Windows
        cv2.rectangle(canvas, (x-car_width//3, y-car_height+3), 
                     (x+car_width//3, y-car_height//2), intensity * 0.8, 1)
    
    def _draw_table(self, canvas: np.ndarray, x: int, y: int, size: int, 
                   pattern: PatternType, intensity: float):
        # Table top
        table_width = size
        table_height = size // 6
        cv2.rectangle(canvas, (x-table_width//2, y-table_height), 
                     (x+table_width//2, y), intensity, 2)
        
        # Apply pattern to table top
        if pattern != PatternType.SOLID:
            self._apply_pattern(canvas, x, y-table_height//2, table_width, table_height, 
                              pattern, intensity * 0.7)
        
        # Legs
        leg_height = size // 3
        for leg_x in [x-table_width//3, x+table_width//3]:
            cv2.line(canvas, (leg_x, y), (leg_x, y+leg_height), intensity, 2)
    
    def _draw_chair(self, canvas: np.ndarray, x: int, y: int, size: int, 
                   pattern: PatternType, intensity: float):
        # Seat
        seat_width = size
        seat_height = size // 8
        cv2.rectangle(canvas, (x-seat_width//2, y-seat_height), 
                     (x+seat_width//2, y), intensity, 2)
        
        # Backrest
        back_height = size // 2
        cv2.line(canvas, (x-seat_width//2, y-seat_height), 
                (x-seat_width//2, y-seat_height-back_height), intensity, 2)
        cv2.line(canvas, (x+seat_width//2, y-seat_height), 
                (x+seat_width//2, y-seat_height-back_height), intensity, 2)
        cv2.line(canvas, (x-seat_width//2, y-seat_height-back_height), 
                (x+seat_width//2, y-seat_height-back_height), intensity, 2)
        
        # Apply pattern to backrest
        if pattern != PatternType.SOLID:
            self._apply_pattern(canvas, x, y-seat_height-back_height//2, seat_width, back_height, 
                              pattern, intensity * 0.6)
        
        # Legs
        leg_height = size // 3
        for leg_x in [x-seat_width//3, x+seat_width//3]:
            cv2.line(canvas, (leg_x, y), (leg_x, y+leg_height), intensity, 2)
    
    def _draw_bench(self, canvas: np.ndarray, x: int, y: int, size: int, 
                   pattern: PatternType, intensity: float):
        # Seat
        bench_width = int(size * 1.2)
        bench_height = size // 6
        cv2.rectangle(canvas, (x-bench_width//2, y-bench_height), 
                     (x+bench_width//2, y), intensity, 2)
        
        # Apply pattern
        if pattern != PatternType.SOLID:
            self._apply_pattern(canvas, x, y-bench_height//2, bench_width, bench_height, 
                              pattern, intensity * 0.7)
        
        # Support legs
        cv2.line(canvas, (x-bench_width//3, y), (x-bench_width//3, y+size//4), intensity, 2)
        cv2.line(canvas, (x+bench_width//3, y), (x+bench_width//3, y+size//4), intensity, 2)
    
    def _draw_flower(self, canvas: np.ndarray, x: int, y: int, size: int, 
                    pattern: PatternType, intensity: float):
        # Center
        cv2.circle(canvas, (x, y), size//4, intensity, 2)
        
        # Petals
        petal_radius = size//3
        for i in range(6):
            angle = i * np.pi / 3
            petal_x = x + int(petal_radius * np.cos(angle))
            petal_y = y + int(petal_radius * np.sin(angle))
            cv2.circle(canvas, (petal_x, petal_y), size//6, intensity, 1)
        
    def _draw_flower(self, canvas: np.ndarray, x: int, y: int, size: int, 
                    pattern: PatternType, intensity: float):
        # Center
        cv2.circle(canvas, (x, y), size//4, intensity, 2)
        
        # Petals
        petal_radius = size//3
        for i in range(6):
            angle = i * np.pi / 3
            petal_x = x + int(petal_radius * np.cos(angle))
            petal_y = y + int(petal_radius * np.sin(angle))
            cv2.circle(canvas, (petal_x, petal_y), size//6, intensity, 1)
        
        # Apply pattern to petals area
        if pattern != PatternType.SOLID:
            self._apply_pattern(canvas, x, y, size, size, pattern, intensity * 0.5)
        
        # Stem
        cv2.line(canvas, (x, y+size//4), (x, y+size), intensity, 2)
    
    def _draw_ball(self, canvas: np.ndarray, x: int, y: int, size: int, 
                  pattern: PatternType, intensity: float):
        # Main circle
        cv2.circle(canvas, (x, y), size//2, intensity, 2)
        
        # Apply pattern
        if pattern != PatternType.SOLID:
            self._apply_pattern(canvas, x, y, size, size, pattern, intensity * 0.6)
    
    def _draw_rock(self, canvas: np.ndarray, x: int, y: int, size: int, 
                  pattern: PatternType, intensity: float):
        # Irregular shape using polygon
        angles = np.linspace(0, 2*np.pi, 8)
        points = []
        for angle in angles:
            radius = size//2 + random.randint(-size//6, size//6)
            px = x + int(radius * np.cos(angle))
            py = y + int(radius * np.sin(angle))
            points.append([px, py])
        
        points = np.array(points, np.int32)
        cv2.polylines(canvas, [points], True, intensity, 2)
        
        # Apply pattern
        if pattern != PatternType.SOLID:
            self._apply_pattern(canvas, x, y, size, size, pattern, intensity * 0.5)
    
    def _draw_cloud(self, canvas: np.ndarray, x: int, y: int, size: int, 
                   pattern: PatternType, intensity: float):
        # Multiple overlapping circles for cloud shape
        cloud_circles = [
            (x-size//3, y, size//3),
            (x, y-size//6, size//2),
            (x+size//3, y, size//3),
            (x, y+size//6, size//4)
        ]
        
        for cx, cy, radius in cloud_circles:
            cv2.circle(canvas, (cx, cy), radius, intensity, 2)
        
        # Apply pattern
        if pattern != PatternType.SOLID:
            self._apply_pattern(canvas, x, y, size, size, pattern, intensity * 0.4)
    
    def _draw_sun(self, canvas: np.ndarray, x: int, y: int, size: int, 
                 pattern: PatternType, intensity: float):
        # Central circle
        cv2.circle(canvas, (x, y), size//3, intensity, 2)
        
        # Rays
        for i in range(8):
            angle = i * np.pi / 4
            ray_start_x = x + int((size//3 + 2) * np.cos(angle))
            ray_start_y = y + int((size//3 + 2) * np.sin(angle))
            ray_end_x = x + int((size//2 + 5) * np.cos(angle))
            ray_end_y = y + int((size//2 + 5) * np.sin(angle))
            cv2.line(canvas, (ray_start_x, ray_start_y), (ray_end_x, ray_end_y), intensity, 2)
        
        # Apply pattern to center
        if pattern != PatternType.SOLID:
            self._apply_pattern(canvas, x, y, size//2, size//2, pattern, intensity * 0.6)
    
    def _draw_bike(self, canvas: np.ndarray, x: int, y: int, size: int, 
                  pattern: PatternType, intensity: float):
        # Wheels
        wheel_radius = size//3
        cv2.circle(canvas, (x-size//2, y), wheel_radius, intensity, 2)
        cv2.circle(canvas, (x+size//2, y), wheel_radius, intensity, 2)
        
        # Frame
        cv2.line(canvas, (x-size//2, y), (x, y-size//3), intensity, 2)  # Front fork
        cv2.line(canvas, (x, y-size//3), (x+size//2, y), intensity, 2)  # Top tube
        cv2.line(canvas, (x, y-size//3), (x, y), intensity, 2)          # Seat post
        cv2.line(canvas, (x-size//4, y), (x+size//4, y), intensity, 2)  # Bottom tube
        
        # Apply pattern to frame
        if pattern != PatternType.SOLID:
            self._apply_pattern(canvas, x, y-size//6, size, size//2, pattern, intensity * 0.6)
    
    def _draw_umbrella(self, canvas: np.ndarray, x: int, y: int, size: int, 
                      pattern: PatternType, intensity: float):
        # Handle
        cv2.line(canvas, (x, y), (x, y+size), intensity, 2)
        
        # Umbrella top (semicircle)
        cv2.ellipse(canvas, (x, y), (size//2, size//4), 0, 0, 180, intensity, 2)
        
        # Umbrella ribs
        for i in range(5):
            angle = np.pi * i / 4
            rib_x = x + int((size//2) * np.cos(angle))
            rib_y = y - int((size//4) * np.sin(angle))
            cv2.line(canvas, (x, y), (rib_x, rib_y), intensity, 1)
        
        # Apply pattern to umbrella top
        if pattern != PatternType.SOLID:
            self._apply_pattern(canvas, x, y-size//8, size, size//2, pattern, intensity * 0.6)
    
    def _draw_lamp(self, canvas: np.ndarray, x: int, y: int, size: int, 
                  pattern: PatternType, intensity: float):
        # Base
        cv2.circle(canvas, (x, y+size//2), size//6, intensity, 2)
        
        # Pole
        cv2.line(canvas, (x, y+size//2), (x, y-size//2), intensity, 2)
        
        # Lampshade
        shade_width = size//2
        shade_height = size//3
        cv2.ellipse(canvas, (x, y-size//2), (shade_width, shade_height//2), 0, 0, 180, intensity, 2)
        cv2.line(canvas, (x-shade_width, y-size//2), (x+shade_width, y-size//2), intensity, 2)
        
        # Apply pattern to lampshade
        if pattern != PatternType.SOLID:
            self._apply_pattern(canvas, x, y-size//2, shade_width*2, shade_height, 
                              pattern, intensity * 0.6)


class ConceptualBlendingDataGenerator:
    def __init__(self, canvas_size=256):
        self.canvas_size = canvas_size
        self.scene_generator = FullSceneGenerator(canvas_size)
        
        # Define blending instruction templates
        self.instruction_templates = {
            "shape_transfer": [
                "Take the shape/form of {object1} from scene 1 and apply it to scene 2",
                "Use the geometric structure of {object1} but in the context of scene 2",
                "Adapt the {object1} shape from scene 1 to fit scene 2's environment"
            ],
            "pattern_transfer": [
                "Apply the pattern from {object1} in scene 1 to the objects in scene 2",
                "Use the visual texture/pattern of scene 1 on the main objects of scene 2",
                "Transfer the surface pattern from scene 1 to scene 2 while keeping the shapes"
            ],
            "arrangement_transfer": [
                "Arrange the objects in scene 2 using the layout pattern from scene 1",
                "Apply the spatial arrangement from scene 1 to the objects in scene 2",
                "Use scene 1's object positioning to reorganize scene 2"
            ],
            "style_transfer": [
                "Combine the overall style of scene 1 with the content of scene 2",
                "Apply scene 1's visual characteristics to scene 2's objects",
                "Blend the aesthetic approach of scene 1 with scene 2's elements"
            ],
            "size_transfer": [
                "Use the size relationships from scene 1 for the objects in scene 2",
                "Apply scene 1's scale proportions to scene 2",
                "Adapt the relative sizing from scene 1 to scene 2's objects"
            ],
            "hybrid_creation": [
                "Create a hybrid scene combining key elements from both scenes",
                "Merge the most prominent features from scene 1 and scene 2",
                "Blend scenes 1 and 2 into a unified composition"
            ],
            "context_adaptation": [
                "Place scene 1's main objects into scene 2's environment",
                "Adapt scene 1's objects to function in scene 2's context",
                "Transplant key elements from scene 1 into scene 2's setting"
            ]
        }
    
    def generate_blending_task(self) -> Dict:
        """Generate a conceptual blending task with 4 reference scenes"""
        
        # Generate 4 different scenes
        scenes = []
        scene_types_used = []
        
        # Ensure diversity in scene types
        available_scene_types = list(SceneType)
        
        for i in range(4):
            # Choose scene type (avoid repetition for first 4)
            if i < len(available_scene_types):
                scene_type = available_scene_types[i]
            else:
                scene_type = random.choice(list(SceneType))
            
            scene_style = random.choice(["minimal", "balanced", "dense"])
            scene = self.scene_generator.generate_scene(scene_type, scene_style)
            scenes.append(scene)
            scene_types_used.append(scene_type)
        
        # Choose blending instruction type
        instruction_type = random.choice(list(self.instruction_templates.keys()))
        
        # Select target scenes for the instruction
        primary_scene_idx = random.randint(0, 3)
        secondary_scene_idx = random.randint(0, 3)
        while secondary_scene_idx == primary_scene_idx:
            secondary_scene_idx = random.randint(0, 3)
        
        # Generate specific instruction
        primary_scene = scenes[primary_scene_idx]
        secondary_scene = scenes[secondary_scene_idx]
        
        # Choose specific object from primary scene if relevant
        if primary_scene.objects:
            primary_object = random.choice(primary_scene.objects)
            object_name = primary_object.obj_type.value
        else:
            object_name = "main element"
        
        instruction_template = random.choice(self.instruction_templates[instruction_type])
        instruction = instruction_template.format(object1=object_name)
        
        # Create the target blended scene
        target_scene = self._create_blended_scene(
            scenes, instruction_type, primary_scene_idx, secondary_scene_idx
        )
        
        return {
            "reference_scenes": scenes,
            "instruction": instruction,
            "instruction_type": instruction_type,
            "primary_scene_idx": primary_scene_idx,
            "secondary_scene_idx": secondary_scene_idx,
            "target_scene": target_scene,
            "scene_types": [s.scene_type.value for s in scenes]
        }
    
    def _create_blended_scene(self, scenes: List[Scene], instruction_type: str, 
                            primary_idx: int, secondary_idx: int) -> Scene:
        """Create target scene based on blending instruction"""
        
        primary_scene = scenes[primary_idx]
        secondary_scene = scenes[secondary_idx]
        
        # Start with secondary scene as base
        blended_scene = Scene(
            scene_type=secondary_scene.scene_type,
            objects=[],
            overall_arrangement=secondary_scene.overall_arrangement,
            dominant_pattern=secondary_scene.dominant_pattern,
            scene_style=secondary_scene.scene_style,
            canvas_size=self.canvas_size
        )
        
        if instruction_type == "shape_transfer":
            # Take shapes from primary, everything else from secondary
            for i, sec_obj in enumerate(secondary_scene.objects):
                if i < len(primary_scene.objects):
                    prim_obj = primary_scene.objects[i]
                    # Use primary object's type but secondary's other attributes
                    blended_obj = SceneObject(
                        obj_type=prim_obj.obj_type,
                        position=sec_obj.position,
                        size=sec_obj.size,
                        pattern=sec_obj.pattern,
                        arrangement_role=sec_obj.arrangement_role,
                        color_intensity=sec_obj.color_intensity
                    )
                else:
                    blended_obj = sec_obj
                blended_scene.objects.append(blended_obj)
        
        elif instruction_type == "pattern_transfer":
            # Take patterns from primary, shapes from secondary
            primary_pattern = primary_scene.dominant_pattern
            for sec_obj in secondary_scene.objects:
                blended_obj = SceneObject(
                    obj_type=sec_obj.obj_type,
                    position=sec_obj.position,
                    size=sec_obj.size,
                    pattern=primary_pattern,
                    arrangement_role=sec_obj.arrangement_role,
                    color_intensity=sec_obj.color_intensity
                )
                blended_scene.objects.append(blended_obj)
            blended_scene.dominant_pattern = primary_pattern
        
        elif instruction_type == "arrangement_transfer":
            # Use primary arrangement with secondary objects
            blended_scene.overall_arrangement = primary_scene.overall_arrangement
            
            # Regenerate positions based on primary arrangement
            if secondary_scene.objects:
                object_types = [obj.obj_type for obj in secondary_scene.objects]
                new_positions = self.scene_generator._generate_positions(
                    object_types, primary_scene.overall_arrangement
                )
                
                for i, sec_obj in enumerate(secondary_scene.objects):
                    new_pos = new_positions[i] if i < len(new_positions) else sec_obj.position
                    blended_obj = SceneObject(
                        obj_type=sec_obj.obj_type,
                        position=new_pos,
                        size=sec_obj.size,
                        pattern=sec_obj.pattern,
                        arrangement_role=sec_obj.arrangement_role,
                        color_intensity=sec_obj.color_intensity
                    )
                    blended_scene.objects.append(blended_obj)
        
        elif instruction_type == "style_transfer":
            # Apply primary's pattern and intensity to secondary's objects
            for sec_obj in secondary_scene.objects:
                blended_obj = SceneObject(
                    obj_type=sec_obj.obj_type,
                    position=sec_obj.position,
                    size=sec_obj.size,
                    pattern=primary_scene.dominant_pattern,
                    arrangement_role=sec_obj.arrangement_role,
                    color_intensity=primary_scene.objects[0].color_intensity if primary_scene.objects else sec_obj.color_intensity
                )
                blended_scene.objects.append(blended_obj)
            blended_scene.dominant_pattern = primary_scene.dominant_pattern
        
        elif instruction_type == "size_transfer":
            # Apply primary's size relationships to secondary's objects
            if primary_scene.objects and secondary_scene.objects:
                size_ratios = [obj.size for obj in primary_scene.objects]
                avg_primary_size = np.mean(size_ratios)
                avg_secondary_size = np.mean([obj.size for obj in secondary_scene.objects])
                scale_factor = avg_primary_size / avg_secondary_size if avg_secondary_size > 0 else 1.0
                
                for sec_obj in secondary_scene.objects:
                    new_size = int(sec_obj.size * scale_factor)
                    new_size = max(8, min(50, new_size))  # Clamp size
                    
                    blended_obj = SceneObject(
                        obj_type=sec_obj.obj_type,
                        position=sec_obj.position,
                        size=new_size,
                        pattern=sec_obj.pattern,
                        arrangement_role=sec_obj.arrangement_role,
                        color_intensity=sec_obj.color_intensity
                    )
                    blended_scene.objects.append(blended_obj)
        
        elif instruction_type == "hybrid_creation":
            # Combine objects from both scenes
            combined_objects = []
            
            # Add main objects from both scenes
            main_objects_primary = [obj for obj in primary_scene.objects if obj.arrangement_role == "main"]
            main_objects_secondary = [obj for obj in secondary_scene.objects if obj.arrangement_role == "main"]
            
            all_main_objects = main_objects_primary + main_objects_secondary
            if all_main_objects:
                # Regenerate positions for combined objects
                object_types = [obj.obj_type for obj in all_main_objects]
                new_positions = self.scene_generator._generate_positions(
                    object_types, ArrangementType.SCATTERED
                )
                
                for i, obj in enumerate(all_main_objects):
                    new_pos = new_positions[i] if i < len(new_positions) else obj.position
                    blended_obj = SceneObject(
                        obj_type=obj.obj_type,
                        position=new_pos,
                        size=obj.size,
                        pattern=obj.pattern,
                        arrangement_role="main",
                        color_intensity=obj.color_intensity
                    )
                    combined_objects.append(blended_obj)
            
            blended_scene.objects = combined_objects
        
        elif instruction_type == "context_adaptation":
            # Place primary objects in secondary's context
            blended_scene.scene_type = secondary_scene.scene_type
            
            # Use primary objects but adapt to secondary's environment
            if primary_scene.objects:
                object_types = [obj.obj_type for obj in primary_scene.objects]
                new_positions = self.scene_generator._generate_positions(
                    object_types, secondary_scene.overall_arrangement
                )
                
                for i, prim_obj in enumerate(primary_scene.objects):
                    new_pos = new_positions[i] if i < len(new_positions) else prim_obj.position
                    blended_obj = SceneObject(
                        obj_type=prim_obj.obj_type,
                        position=new_pos,
                        size=prim_obj.size,
                        pattern=secondary_scene.dominant_pattern,  # Adapt to new context
                        arrangement_role=prim_obj.arrangement_role,
                        color_intensity=prim_obj.color_intensity
                    )
                    blended_scene.objects.append(blended_obj)
        
        return blended_scene
    
    def generate_training_dataset(self, num_samples: int = 5000, save_path: str = "concept_blend_scenes"):
        """Generate complete training dataset"""
        
        if not os.path.exists(save_path):
            os.makedirs(save_path)
        
        all_reference_images = []
        all_instructions = []
        all_target_sketches = []
        all_metadata = []
        
        print(f"Generating {num_samples} conceptual blending scene samples...")
        
        for i in range(num_samples):
            if i % 500 == 0:
                print(f"  Generated {i}/{num_samples} samples")
            
            # Generate blending task
            blend_task = self.generate_blending_task()
            
            # Render reference scenes
            reference_images = []
            for scene in blend_task["reference_scenes"]:
                scene_image = self.scene_generator.render_scene(scene)
                reference_images.append(scene_image)
            
            # Render target scene
            target_sketch = self.scene_generator.render_scene(blend_task["target_scene"])
            
            # Store data
            all_reference_images.append(np.stack(reference_images))
            all_instructions.append(blend_task["instruction"])
            all_target_sketches.append(target_sketch)
            all_metadata.append(blend_task)
        
        # Convert to numpy arrays
        all_reference_images = np.array(all_reference_images)  # [N, 4, 256, 256]
        all_target_sketches = np.array(all_target_sketches)    # [N, 256, 256]
        
        # Save dataset
        np.save(os.path.join(save_path, "reference_scenes.npy"), all_reference_images)
        np.save(os.path.join(save_path, "target_sketches.npy"), all_target_sketches)
        
        # Save instructions as text file
        with open(os.path.join(save_path, "instructions.txt"), "w") as f:
            for instruction in all_instructions:
                f.write(instruction + "\n")
        
        # Save metadata
        with open(os.path.join(save_path, "metadata.pkl"), "wb") as f:
            pickle.dump(all_metadata, f)
        
        # Save dataset info
        dataset_info = {
            "total_samples": num_samples,
            "num_reference_scenes": 4,
            "canvas_size": self.canvas_size,
            "instruction_types": list(self.instruction_templates.keys()),
            "scene_types": [st.value for st in SceneType],
            "object_types": [ot.value for ot in ObjectType],
            "pattern_types": [pt.value for pt in PatternType],
            "arrangement_types": [at.value for at in ArrangementType]
        }
        
        with open(os.path.join(save_path, "dataset_info.json"), "w") as f:
            json.dump(dataset_info, f, indent=2)
        
        print(f"\nDataset generation complete!")
        print(f"Reference scenes shape: {all_reference_images.shape}")
        print(f"Target sketches shape: {all_target_sketches.shape}")
        print(f"Instructions: {len(all_instructions)} text samples")
        print(f"Dataset saved to: {save_path}")
        
        return all_reference_images, all_instructions, all_target_sketches, all_metadata
    
    def create_test_visualizations(self, num_samples: int = 6, save_path: str = "scene_blend_visualizations"):
        """Create test visualizations of the blending process"""
        
        if not os.path.exists(save_path):
            os.makedirs(save_path)
        
        for sample_idx in range(num_samples):
            # Generate sample
            blend_task = self.generate_blending_task()
            
            # Render scenes
            reference_images = []
            for scene in blend_task["reference_scenes"]:
                scene_image = self.scene_generator.render_scene(scene)
                reference_images.append(scene_image)
            
            target_sketch = self.scene_generator.render_scene(blend_task["target_scene"])
            
            # Create visualization
            self._create_blend_visualization(
                reference_images, blend_task["instruction"], target_sketch, blend_task,
                f"{save_path}/scene_blend_sample_{sample_idx}.png"
            )
        
        print(f"Test visualizations saved to {save_path}")
    
    def _create_blend_visualization(self, reference_images: List[np.ndarray], 
                                  instruction: str, target_sketch: np.ndarray, 
                                  metadata: Dict, save_path: str):
        """Create visualization of a scene blending sample"""
        try:
            import matplotlib.pyplot as plt
            
            fig, axes = plt.subplots(2, 3, figsize=(18, 12))
            
            # Show 4 reference scenes (top row + bottom left)
            scene_titles = ["Reference Scene 1", "Reference Scene 2", "Reference Scene 3", "Reference Scene 4"]
            positions = [(0, 0), (0, 1), (0, 2), (1, 0)]
            
            for i in range(4):
                row, col = positions[i]
                axes[row, col].imshow(reference_images[i], cmap='gray')
                
                # Add scene type and style info
                scene_info = f"{metadata['scene_types'][i].title()}"
                if i < len(metadata['reference_scenes']):
                    scene_style = metadata['reference_scenes'][i].scene_style
                    scene_info += f"\n({scene_style})"
                
                axes[row, col].set_title(f'{scene_titles[i]}\n{scene_info}', fontsize=11)
                axes[row, col].axis('off')
            
            # Show target sketch (bottom middle)
            axes[1, 1].imshow(target_sketch, cmap='gray')
            axes[1, 1].set_title('Target Blended Scene', fontsize=14, weight='bold')
            axes[1, 1].axis('off')
            
            # Show instruction and metadata (bottom right)
            blend_info = f"""Instruction:
{instruction}

Blend Type: {metadata['instruction_type']}
Primary Scene: {metadata['primary_scene_idx'] + 1}
Secondary Scene: {metadata['secondary_scene_idx'] + 1}

Scene Types:
1. {metadata['scene_types'][0]}
2. {metadata['scene_types'][1]}
3. {metadata['scene_types'][2]}
4. {metadata['scene_types'][3]}"""
            
            axes[1, 2].text(0.05, 0.95, blend_info, fontsize=10, verticalalignment='top',
                           transform=axes[1, 2].transAxes, family='monospace')
            axes[1, 2].set_title('Blending Task', fontsize=12)
            axes[1, 2].axis('off')
            
            plt.tight_layout()
            plt.savefig(save_path, dpi=150, bbox_inches='tight')
            plt.close()
            
        except ImportError:
            print("Matplotlib not available - skipping visualization")


def main():
    """Test the scene conceptual blending data generator"""
    print("=== Scene Conceptual Blending Data Generator Test ===")
    
    # Initialize generator
    generator = ConceptualBlendingDataGenerator(canvas_size=256)
    
    # Test single task generation
    print("\n1. Testing single blending task generation...")
    blend_task = generator.generate_blending_task()
    
    print(f"Scene types: {blend_task['scene_types']}")
    print(f"Instruction: {blend_task['instruction']}")
    print(f"Instruction type: {blend_task['instruction_type']}")
    print(f"Primary scene: {blend_task['primary_scene_idx'] + 1}")
    print(f"Secondary scene: {blend_task['secondary_scene_idx'] + 1}")
    
    # Test variety
    print("\n2. Testing instruction variety...")
    instruction_types = set()
    scene_combinations = set()
    
    for i in range(20):
        task = generator.generate_blending_task()
        instruction_types.add(task['instruction_type'])
        scene_combo = tuple(sorted(task['scene_types']))
        scene_combinations.add(scene_combo)
    
    print(f"Instruction types seen: {sorted(instruction_types)}")
    print(f"Scene combinations: {len(scene_combinations)} unique combinations")
    
    # Generate test visualizations
    print("\n3. Generating test visualizations...")
    generator.create_test_visualizations(num_samples=5)
    
    # Generate small dataset for testing
    print("\n4. Generating small test dataset...")
    ref_images, instructions, target_sketches, metadata = generator.generate_training_dataset(
        num_samples=200,  # Small test dataset
        save_path="test_scene_concept_dataset"
    )
    
    # Analyze dataset
    print("\n5. Dataset analysis:")
    instruction_type_counts = {}
    for meta in metadata:
        inst_type = meta['instruction_type']
        instruction_type_counts[inst_type] = instruction_type_counts.get(inst_type, 0) + 1
    
    print("Instruction type distribution:")
    for inst_type, count in sorted(instruction_type_counts.items()):
        print(f"  {inst_type}: {count}")
    
    print("\n=== Scene Conceptual Blending Test Complete ===")
    print("Ready for model training!")


if __name__ == "__main__":
    main()
```

```python
import numpy as np
import cv2
import random
import os
import json
import pickle
import time
from typing import Dict, List, Tuple, Optional
from dataclasses import dataclass
import itertools
from tqdm import tqdm
import multiprocessing as mp
from functools import partial
import gc
import shutil

class InstructPix2PixDatasetGenerator:
    def __init__(self, canvas_size=256):
        self.canvas_size = canvas_size
        # Note: These would need to be imported from your scene generator
        # self.scene_generator = FullSceneGenerator(canvas_size)
        # self.blend_generator = ConceptualBlendingDataGenerator(canvas_size)
        
        # Extended instruction templates for more variety - add more for 10k dataset
        self.detailed_instructions = {
            "shape_transfer": [
                "Transfer the {source_object} shape from scene {source_idx} to replace {target_object} in scene {target_idx}",
                "Take the geometric form of {source_object} and apply it to {target_object} in the target scene",
                "Replace {target_object} with the shape style of {source_object} from the reference scene",
                "Adapt the {source_object} form to create a new {target_object} in the scene",
                "Use {source_object}'s structure but keep it as {target_object} functionally",
                "Copy the outline of {source_object} and use it for {target_object}",
                "Make {target_object} have the same silhouette as {source_object}",
                "Transform {target_object} to match the contour of {source_object}"
            ],
            "pattern_application": [
                "Apply the surface pattern from {source_object} to all objects in the target scene",
                "Transfer the texture/pattern of {source_object} to {target_object}",
                "Use the visual pattern from scene {source_idx} on the {target_object}",
                "Copy the decorative pattern and apply it to the main objects",
                "Make {target_object} have the same surface treatment as {source_object}",
                "Paint {target_object} with the same design as {source_object}",
                "Give {target_object} the visual texture of {source_object}",
                "Replicate the surface details from {source_object} onto {target_object}"
            ],
            "size_modification": [
                "Make {target_object} the same size as {source_object} from the reference",
                "Scale {target_object} to match the proportions of {source_object}",
                "Resize all objects to match the scale shown in scene {source_idx}",
                "Adjust {target_object} to have {source_object}'s dimensions",
                "Change the size relationship to match the reference scene",
                "Enlarge {target_object} to be as big as {source_object}",
                "Shrink {target_object} to match {source_object}'s size",
                "Modify the scale of {target_object} based on {source_object}"
            ],
            "arrangement_change": [
                "Rearrange objects using the layout pattern from scene {source_idx}",
                "Move {target_object} to match the positioning of {source_object}",
                "Apply the spatial arrangement from the reference to the target scene",
                "Reorganize the scene using the reference layout as a guide",
                "Position objects in the same pattern as shown in scene {source_idx}",
                "Reposition {target_object} to mirror {source_object}'s placement",
                "Adjust the spacing between objects like in scene {source_idx}",
                "Create the same spatial composition as the reference scene"
            ],
            "style_blend": [
                "Combine the visual style of scene {source_idx} with scene {target_idx}",
                "Apply the aesthetic approach from the reference to the target",
                "Merge the design style while keeping the target scene's content",
                "Use the reference's visual treatment on the target scene",
                "Blend the artistic style from reference into the target composition",
                "Fuse the visual characteristics of both scenes",
                "Harmonize the appearance of both scenes into one",
                "Create a unified look combining elements from both scenes"
            ],
            "object_replacement": [
                "Replace {target_object} with {source_object} while keeping the scene context",
                "Substitute {target_object} for {source_object} in the scene",
                "Change {target_object} to {source_object} but maintain the layout",
                "Put {source_object} where {target_object} is located",
                "Swap {target_object} with {source_object} from the reference",
                "Remove {target_object} and add {source_object} instead",
                "Exchange {target_object} for {source_object} in the same position",
                "Transform {target_object} into {source_object}"
            ],
            "context_adaptation": [
                "Move {source_object} from scene {source_idx} into the environment of scene {target_idx}",
                "Place {source_object} into the target scene's setting",
                "Adapt {source_object} to fit in the target scene's context",
                "Transplant {source_object} into the new environment",
                "Integrate {source_object} into the target scene naturally",
                "Insert {source_object} so it belongs in the target setting",
                "Add {source_object} to complement the target scene",
                "Incorporate {source_object} seamlessly into the new context"
            ],
            "creative_fusion": [
                "Create a fusion of {source_object} and {target_object} combining their characteristics",
                "Merge {source_object} and {target_object} into a hybrid design",
                "Combine elements from both scenes into a unified composition",
                "Blend the key features from both reference scenes",
                "Create something new using inspiration from both scenes",
                "Mix the attributes of {source_object} and {target_object}",
                "Generate a hybrid that incorporates both objects' qualities",
                "Synthesize {source_object} and {target_object} into one entity"
            ],
            "attribute_transfer": [
                "Give {target_object} the visual qualities of {source_object}",
                "Make {target_object} look more like {source_object} in style",
                "Transfer the appearance characteristics from {source_object} to {target_object}",
                "Apply {source_object}'s visual properties to {target_object}",
                "Modify {target_object} to have {source_object}'s visual attributes",
                "Copy the color scheme from {source_object} to {target_object}",
                "Make {target_object} adopt the visual mood of {source_object}",
                "Give {target_object} the same visual personality as {source_object}"
            ],
            "environmental_change": [
                "Change the scene setting from {target_scene_type} to {source_scene_type}",
                "Adapt the objects to fit a {source_scene_type} environment",
                "Transform the background context while keeping main objects",
                "Place the objects in a {source_scene_type} setting instead",
                "Change the environmental context to match the reference scene",
                "Relocate the scene to a {source_scene_type} atmosphere",
                "Modify the surroundings to create a {source_scene_type} feeling",
                "Alter the backdrop to match the {source_scene_type} theme"
            ]
        }
        
        # Complexity levels for curriculum learning
        self.complexity_levels = {
            "simple": {
                "max_objects_per_scene": 3,
                "instruction_types": ["shape_transfer", "pattern_application", "size_modification"],
                "scene_styles": ["minimal", "balanced"],
                "pattern_complexity": ["solid", "stripes", "dots"]
            },
            "medium": {
                "max_objects_per_scene": 5,
                "instruction_types": ["arrangement_change", "style_blend", "object_replacement"],
                "scene_styles": ["minimal", "balanced", "dense"],
                "pattern_complexity": ["stripes", "dots", "grid", "waves"]
            },
            "complex": {
                "max_objects_per_scene": 7,
                "instruction_types": ["creative_fusion", "attribute_transfer", "environmental_change"],
                "scene_styles": ["balanced", "dense"],
                "pattern_complexity": ["grid", "waves", "zigzag"]
            }
        }
    
    def check_disk_space(self, path: str, required_gb: float = 15.0) -> bool:
        """Check if there's enough disk space"""
        try:
            total, used, free = shutil.disk_usage(path)
            free_gb = free / (1024**3)
            print(f"Available disk space: {free_gb:.1f} GB")
            
            if free_gb < required_gb:
                print(f"Warning: Only {free_gb:.1f} GB available. Need at least {required_gb} GB.")
                return False
            return True
        except Exception as e:
            print(f"Could not check disk space: {e}")
            return True  # Assume OK if can't check
    
    def estimate_dataset_size(self, num_samples: int, canvas_size: int = 256) -> float:
        """Estimate dataset size in GB"""
        # Each sample: 4 reference images + 1 target = 5 images
        # Each image: canvas_size x canvas_size x 1 byte (grayscale)
        bytes_per_sample = 5 * canvas_size * canvas_size
        total_bytes = num_samples * bytes_per_sample
        return total_bytes / (1024**3)  # Convert to GB
    
    def save_batch_safely(self, batch_data: Dict, batch_idx: int, save_path: str) -> bool:
        """Save a batch with comprehensive error handling"""
        try:
            # Create batch files
            batch_ref_file = os.path.join(save_path, f"batch_{batch_idx:04d}_ref.npz")
            batch_target_file = os.path.join(save_path, f"batch_{batch_idx:04d}_target.npz")
            batch_inst_file = os.path.join(save_path, f"batch_{batch_idx:04d}_instructions.txt")
            batch_meta_file = os.path.join(save_path, f"batch_{batch_idx:04d}_metadata.pkl")
            
            # Save reference images
            np.savez_compressed(batch_ref_file, data=batch_data['ref_images'])
            
            # Save target sketches  
            np.savez_compressed(batch_target_file, data=batch_data['target_sketches'])
            
            # Save instructions
            with open(batch_inst_file, 'w') as f:
                for instruction in batch_data['instructions']:
                    f.write(instruction + '\n')
            
            # Save metadata
            with open(batch_meta_file, 'wb') as f:
                pickle.dump(batch_data['metadata'], f)
            
            # Verify files were written correctly
            if not all(os.path.exists(f) for f in [batch_ref_file, batch_target_file, batch_inst_file, batch_meta_file]):
                print(f"Error: Not all batch {batch_idx} files were created")
                return False
            
            # Verify file sizes are reasonable
            ref_size = os.path.getsize(batch_ref_file)
            target_size = os.path.getsize(batch_target_file)
            
            if ref_size == 0 or target_size == 0:
                print(f"Error: Batch {batch_idx} files are empty")
                return False
            
            print(f"  + Saved batch {batch_idx}: {len(batch_data['ref_images'])} samples")
            return True
            
        except Exception as e:
            print(f"  - Error saving batch {batch_idx}: {e}")
            return False
    
    def load_and_combine_batches(self, save_path: str, num_batches: int) -> Optional[Tuple]:
        """Load and combine all batch files with validation"""
        try:
            print(f"Loading and combining {num_batches} batches...")
            
            all_ref_images = []
            all_target_sketches = []
            all_instructions = []
            all_metadata = []
            
            valid_batches = 0
            
            for batch_idx in range(num_batches):
                try:
                    # File paths
                    batch_ref_file = os.path.join(save_path, f"batch_{batch_idx:04d}_ref.npz")
                    batch_target_file = os.path.join(save_path, f"batch_{batch_idx:04d}_target.npz")
                    batch_inst_file = os.path.join(save_path, f"batch_{batch_idx:04d}_instructions.txt")
                    batch_meta_file = os.path.join(save_path, f"batch_{batch_idx:04d}_metadata.pkl")
                    
                    # Check if all files exist
                    if not all(os.path.exists(f) for f in [batch_ref_file, batch_target_file, batch_inst_file, batch_meta_file]):
                        print(f"  Skipping batch {batch_idx}: missing files")
                        continue
                    
                    # Load data
                    ref_data = np.load(batch_ref_file)['data']
                    target_data = np.load(batch_target_file)['data']
                    
                    with open(batch_inst_file, 'r') as f:
                        instructions = [line.strip() for line in f.readlines()]
                    
                    with open(batch_meta_file, 'rb') as f:
                        metadata = pickle.load(f)
                    
                    # Validate data consistency
                    if len(ref_data) != len(target_data) or len(ref_data) != len(instructions) or len(ref_data) != len(metadata):
                        print(f"  Skipping batch {batch_idx}: inconsistent data lengths")
                        continue
                    
                    # Validate array shapes
                    if ref_data.ndim != 4 or target_data.ndim != 3:
                        print(f"  Skipping batch {batch_idx}: incorrect array dimensions")
                        continue
                    
                    # Add to collections
                    all_ref_images.append(ref_data)
                    all_target_sketches.append(target_data)
                    all_instructions.extend(instructions)
                    all_metadata.extend(metadata)
                    
                    valid_batches += 1
                    print(f"  + Loaded batch {batch_idx}: {len(ref_data)} samples")
                    
                    # Clean up batch data from memory
                    del ref_data, target_data, instructions, metadata
                    gc.collect()
                    
                except Exception as e:
                    print(f"  - Error loading batch {batch_idx}: {e}")
                    continue
            
            if valid_batches == 0:
                print("No valid batches found!")
                return None
            
            print(f"Successfully loaded {valid_batches}/{num_batches} batches")
            
            # Combine arrays
            if all_ref_images and all_target_sketches:
                combined_ref = np.concatenate(all_ref_images, axis=0)
                combined_target = np.concatenate(all_target_sketches, axis=0)
                
                print(f"Combined dataset shape: ref={combined_ref.shape}, target={combined_target.shape}")
                return combined_ref, combined_target, all_instructions, all_metadata
            else:
                print("No data to combine")
                return None
                
        except Exception as e:
            print(f"Error combining batches: {e}")
            return None
    
    def save_final_dataset(self, ref_images: np.ndarray, target_sketches: np.ndarray, 
                          instructions: List[str], metadata: List[Dict], save_path: str) -> bool:
        """Save final combined dataset with validation"""
        try:
            # Validate input consistency
            if len(ref_images) != len(target_sketches) or len(ref_images) != len(instructions) or len(ref_images) != len(metadata):
                print("Error: Inconsistent data lengths in final dataset")
                return False
            
            print(f"Saving final dataset: {len(ref_images)} samples")
            
            # Save arrays
            ref_file = os.path.join(save_path, "reference_scenes.npz")
            target_file = os.path.join(save_path, "target_sketches.npz")
            
            np.savez_compressed(ref_file, data=ref_images)
            np.savez_compressed(target_file, data=target_sketches)
            
            # Save instructions
            inst_file = os.path.join(save_path, "instructions.txt")
            with open(inst_file, 'w') as f:
                for instruction in instructions:
                    f.write(instruction + '\n')
            
            # Save metadata
            meta_file = os.path.join(save_path, "metadata.pkl")
            with open(meta_file, 'wb') as f:
                pickle.dump(metadata, f)
            
            # Verify files
            for file_path in [ref_file, target_file, inst_file, meta_file]:
                if not os.path.exists(file_path) or os.path.getsize(file_path) == 0:
                    print(f"Error: File {file_path} was not saved correctly")
                    return False
            
            print("+ Final dataset saved successfully")
            return True
            
        except Exception as e:
            print(f"Error saving final dataset: {e}")
            return False
    
    def create_train_val_split_safe(self, save_path: str, total_samples: int, val_split: float = 0.15) -> bool:
        """Create train/validation split with comprehensive error handling"""
        try:
            print(f"Creating train/val split for {total_samples} samples...")
            
            # Calculate split sizes
            val_size = int(total_samples * val_split)
            train_size = total_samples - val_size
            
            # Create random indices
            np.random.seed(42)  # For reproducible splits
            all_indices = np.arange(total_samples)
            np.random.shuffle(all_indices)
            
            train_indices = all_indices[:train_size]
            val_indices = all_indices[train_size:]
            
            # Create directories
            train_path = os.path.join(save_path, "train")
            val_path = os.path.join(save_path, "val")
            os.makedirs(train_path, exist_ok=True)
            os.makedirs(val_path, exist_ok=True)
            
            # Save indices
            train_idx_file = os.path.join(train_path, "indices.txt")
            val_idx_file = os.path.join(val_path, "indices.txt")
            
            with open(train_idx_file, 'w') as f:
                for idx in train_indices:
                    f.write(f"{idx}\n")
            
            with open(val_idx_file, 'w') as f:
                for idx in val_indices:
                    f.write(f"{idx}\n")
            
            # Create info files
            train_info = {
                "num_samples": len(train_indices),
                "percentage": len(train_indices) / total_samples * 100,
                "indices_file": "indices.txt"
            }
            
            val_info = {
                "num_samples": len(val_indices),
                "percentage": len(val_indices) / total_samples * 100,
                "indices_file": "indices.txt"
            }
            
            with open(os.path.join(train_path, "split_info.json"), 'w') as f:
                json.dump(train_info, f, indent=2)
            
            with open(os.path.join(val_path, "split_info.json"), 'w') as f:
                json.dump(val_info, f, indent=2)
            
            print(f"+ Train/val split created:")
            print(f"  Train: {len(train_indices)} samples ({len(train_indices)/total_samples*100:.1f}%)")
            print(f"  Val: {len(val_indices)} samples ({len(val_indices)/total_samples*100:.1f}%)")
            
            return True
            
        except Exception as e:
            print(f"Error creating train/val split: {e}")
            return False
    
    def generate_safe_dataset(self, 
                             num_samples: int = 1000,
                             save_path: str = "instructpix2pix_dataset",
                             batch_size: int = 50,
                             complexity_distribution: Dict[str, float] = None) -> bool:
        """Generate dataset with comprehensive safety checks and error handling"""
        
        if complexity_distribution is None:
            complexity_distribution = {"simple": 0.4, "medium": 0.4, "complex": 0.2}
        
        # Validate inputs
        if num_samples <= 0 or batch_size <= 0:
            print("Error: Invalid sample count or batch size")
            return False
        
        if not 0.99 <= sum(complexity_distribution.values()) <= 1.01:
            print("Error: Complexity distribution doesn't sum to 1.0")
            return False
        
        # Check disk space for 10k dataset
        estimated_size = self.estimate_dataset_size(num_samples, self.canvas_size)
        print(f"Estimated dataset size: {estimated_size:.2f} GB")
        
        if not self.check_disk_space(save_path, estimated_size + 2):  # +2 GB buffer for 10k dataset
            response = input("Continue anyway? (y/N): ")
            if response.lower() != 'y':
                return False
        
        # Create output directory
        os.makedirs(save_path, exist_ok=True)
        
        # Calculate samples per complexity
        samples_per_complexity = {
            complexity: int(num_samples * ratio)
            for complexity, ratio in complexity_distribution.items()
        }
        
        # Adjust for rounding errors
        total_assigned = sum(samples_per_complexity.values())
        if total_assigned < num_samples:
            samples_per_complexity["simple"] += (num_samples - total_assigned)
        
        print(f"Sample distribution: {samples_per_complexity}")
        
        # Initialize batch tracking
        batch_counter = 0
        current_batch = {
            'ref_images': [],
            'target_sketches': [],
            'instructions': [],
            'metadata': []
        }
        
        # Generate samples by complexity with better progress tracking for 10k
        total_generated = 0
        start_time = time.time()
        
        for complexity, num_complexity_samples in samples_per_complexity.items():
            if num_complexity_samples == 0:
                continue
                
            print(f"\nGenerating {num_complexity_samples} {complexity} samples...")
            complexity_start_time = time.time()
            
            for i in tqdm(range(num_complexity_samples), desc=f"{complexity} samples"):
                try:
                    # Generate sample (placeholder - would use your actual generation logic)
                    sample = self.generate_sample_placeholder(complexity)
                    
                    # Add to batch
                    current_batch['ref_images'].append(sample['ref_images'])
                    current_batch['target_sketches'].append(sample['target_sketch'])
                    current_batch['instructions'].append(sample['instruction'])
                    current_batch['metadata'].append(sample['metadata'])
                    
                    # Save batch when full
                    if len(current_batch['ref_images']) >= batch_size:
                        # Convert to arrays
                        current_batch['ref_images'] = np.array(current_batch['ref_images'])
                        current_batch['target_sketches'] = np.array(current_batch['target_sketches'])
                        
                        # Save batch
                        if self.save_batch_safely(current_batch, batch_counter, save_path):
                            total_generated += len(current_batch['ref_images'])
                            batch_counter += 1
                        else:
                            print(f"Failed to save batch {batch_counter}")
                            return False
                        
                        # Reset batch
                        current_batch = {
                            'ref_images': [],
                            'target_sketches': [],
                            'instructions': [],
                            'metadata': []
                        }
                        
                        # Force garbage collection for memory management
                        gc.collect()
                        
                        # Print progress every 1000 samples
                        if total_generated % 1000 == 0:
                            elapsed = time.time() - start_time
                            rate = total_generated / elapsed
                            eta = (num_samples - total_generated) / rate if rate > 0 else 0
                            print(f"    Progress: {total_generated}/{num_samples} samples ({total_generated/num_samples*100:.1f}%) - "
                                  f"Rate: {rate:.1f} samples/sec - ETA: {eta/60:.1f} min")
                
                except Exception as e:
                    print(f"Error generating {complexity} sample {i}: {e}")
                    continue
            
            complexity_elapsed = time.time() - complexity_start_time
            print(f"Completed {complexity}: {num_complexity_samples} samples in {complexity_elapsed/60:.1f} minutes")
        
        # Save remaining batch
        if current_batch['ref_images']:
            current_batch['ref_images'] = np.array(current_batch['ref_images'])
            current_batch['target_sketches'] = np.array(current_batch['target_sketches'])
            
            if self.save_batch_safely(current_batch, batch_counter, save_path):
                total_generated += len(current_batch['ref_images'])
                batch_counter += 1
        
        print(f"\nGenerated {total_generated} samples in {batch_counter} batches")
        
        # Combine batches
        combined_data = self.load_and_combine_batches(save_path, batch_counter)
        if combined_data is None:
            print("Failed to combine batches")
            return False
        
        ref_images, target_sketches, instructions, metadata = combined_data
        
        # Save final dataset
        if not self.save_final_dataset(ref_images, target_sketches, instructions, metadata, save_path):
            return False
        
        # Create train/val split
        if not self.create_train_val_split_safe(save_path, len(ref_images)):
            return False
        
        # Generate analysis
        self.generate_dataset_analysis(metadata, save_path)
        
        # Clean up batch files
        self.cleanup_batch_files(save_path, batch_counter)
        
        print(f"\n[SUCCESS] Dataset generation completed successfully!")
        print(f"Final stats: {len(ref_images)} samples")
        print(f"Saved to: {save_path}")
        
        return True
    
    def generate_sample_placeholder(self, complexity: str) -> Dict:
        """Enhanced placeholder for sample generation with more variety for 10k dataset"""
        canvas_size = self.canvas_size
        
        # More varied dummy data generation
        # Generate reference images with different patterns
        ref_images = []
        for i in range(4):
            # Create more varied reference images
            if i == 0:
                # Geometric shapes
                img = np.zeros((canvas_size, canvas_size), dtype=np.uint8)
                cv2.circle(img, (128, 128), 50, 255, -1)
            elif i == 1:
                # Stripes pattern
                img = np.zeros((canvas_size, canvas_size), dtype=np.uint8)
                for y in range(0, canvas_size, 20):
                    img[y:y+10, :] = 255
            elif i == 2:
                # Random noise pattern
                img = np.random.randint(100, 200, (canvas_size, canvas_size), dtype=np.uint8)
            else:
                # Grid pattern
                img = np.zeros((canvas_size, canvas_size), dtype=np.uint8)
                for x in range(0, canvas_size, 30):
                    img[:, x:x+3] = 255
                for y in range(0, canvas_size, 30):
                    img[y:y+3, :] = 255
            
            ref_images.append(img)
        
        ref_images = np.array(ref_images)
        
        # Generate target sketch with complexity-based variation
        if complexity == "simple":
            # Simple geometric target
            target_sketch = np.zeros((canvas_size, canvas_size), dtype=np.uint8)
            cv2.rectangle(target_sketch, (64, 64), (192, 192), 255, 3)
        elif complexity == "medium":
            # Medium complexity with multiple shapes
            target_sketch = np.zeros((canvas_size, canvas_size), dtype=np.uint8)
            cv2.circle(target_sketch, (100, 100), 30, 255, 2)
            cv2.rectangle(target_sketch, (140, 140), (200, 200), 255, 2)
        else:  # complex
            # Complex pattern combination
            target_sketch = np.zeros((canvas_size, canvas_size), dtype=np.uint8)
            # Multiple overlapping shapes
            for i in range(5):
                x, y = np.random.randint(50, 200, 2)
                r = np.random.randint(20, 40)
                cv2.circle(target_sketch, (x, y), r, 150, 2)
        
        # More varied instruction generation
        object_types = ["circle", "square", "triangle", "line", "star", "house", "tree", "car", "flower", "moon"]
        scene_types = ["park", "city", "garden", "room", "forest", "beach", "mountain", "desert"]
        
        source_object = random.choice(object_types)
        target_object = random.choice([obj for obj in object_types if obj != source_object])
        source_scene = random.choice(scene_types)
        target_scene = random.choice([scene for scene in scene_types if scene != source_scene])
        
        instruction_type = random.choice(list(self.detailed_instructions.keys()))
        templates = self.detailed_instructions[instruction_type]
        template = random.choice(templates)
        
        try:
            instruction = template.format(
                source_object=source_object,
                target_object=target_object,
                source_idx=random.randint(1, 4),
                target_idx=random.randint(1, 4),
                source_scene_type=source_scene,
                target_scene_type=target_scene
            )
        except KeyError:
            # Fallback instruction
            instruction = f"Transform the {target_object} to look like the {source_object}"
        
        metadata = {
            "complexity": complexity,
            "instruction_type": instruction_type,
            "combination_type": "random",
            "source_object": source_object,
            "target_object": target_object,
            "source_scene": source_scene,
            "target_scene": target_scene
        }
        
        return {
            'ref_images': ref_images,
            'target_sketch': target_sketch,
            'instruction': instruction,
            'metadata': metadata
        }
    
    def cleanup_batch_files(self, save_path: str, num_batches: int):
        """Clean up batch files"""
        try:
            print("Cleaning up batch files...")
            for batch_idx in range(num_batches):
                files_to_remove = [
                    f"batch_{batch_idx:04d}_ref.npz",
                    f"batch_{batch_idx:04d}_target.npz", 
                    f"batch_{batch_idx:04d}_instructions.txt",
                    f"batch_{batch_idx:04d}_metadata.pkl"
                ]
                
                for filename in files_to_remove:
                    file_path = os.path.join(save_path, filename)
                    if os.path.exists(file_path):
                        os.remove(file_path)
            
            print("+ Batch files cleaned up")
        except Exception as e:
            print(f"Error cleaning up batch files: {e}")
    
    def generate_dataset_analysis(self, metadata: List[Dict], save_path: str):
        """Generate dataset analysis"""
        try:
            analysis = {
                "total_samples": len(metadata),
                "instruction_type_distribution": {},
                "complexity_distribution": {},
                "combination_type_distribution": {}
            }
            
            # Count distributions
            for meta in metadata:
                # Instruction types
                inst_type = meta.get("instruction_type", "unknown")
                analysis["instruction_type_distribution"][inst_type] = \
                    analysis["instruction_type_distribution"].get(inst_type, 0) + 1
                
                # Complexity levels
                complexity = meta.get("complexity", "unknown")
                analysis["complexity_distribution"][complexity] = \
                    analysis["complexity_distribution"].get(complexity, 0) + 1
                
                # Combination types
                combo_type = meta.get("combination_type", "unknown")
                analysis["combination_type_distribution"][combo_type] = \
                    analysis["combination_type_distribution"].get(combo_type, 0) + 1
            
            # Save analysis
            with open(os.path.join(save_path, "dataset_analysis.json"), "w") as f:
                json.dump(analysis, f, indent=2)
            
            print("+ Dataset analysis saved")
            
        except Exception as e:
            print(f"Error generating analysis: {e}")


def create_pytorch_dataset_loader(dataset_path: str):
    """Create a robust PyTorch dataset loader"""
    
    dataset_code = '''
import torch
import numpy as np
from torch.utils.data import Dataset
import os
import pickle

class InstructPix2PixDataset(Dataset):
    def __init__(self, data_path: str, split: str = "train", load_to_memory: bool = False):
        """
        Args:
            data_path: Path to dataset directory
            split: 'train' or 'val'
            load_to_memory: If True, load all data to memory (faster but uses more RAM)
        """
        self.data_path = data_path
        self.split = split
        self.load_to_memory = load_to_memory
        
        # Load main dataset files
        self.ref_file = os.path.join(data_path, "reference_scenes.npz")
        self.target_file = os.path.join(data_path, "target_sketches.npz")
        self.inst_file = os.path.join(data_path, "instructions.txt")
        
        # Load split indices
        split_path = os.path.join(data_path, split)
        indices_file = os.path.join(split_path, "indices.txt")
        
        if os.path.exists(indices_file):
            with open(indices_file, 'r') as f:
                self.indices = [int(line.strip()) for line in f.readlines()]
        else:
            # Fallback: use all data
            print(f"Warning: No split indices found, using all data")
            # Load a small sample to get dataset size
            ref_data = np.load(self.ref_file)['data']
            self.indices = list(range(len(ref_data)))
            del ref_data
        
        # Load instructions
        with open(self.inst_file, 'r') as f:
            all_instructions = [line.strip() for line in f.readlines()]
        self.instructions = [all_instructions[i] for i in self.indices]
        
        if self.load_to_memory:
            print(f"Loading {split} data to memory...")
            ref_data = np.load(self.ref_file)['data']
            target_data = np.load(self.target_file)['data']
            
            self.ref_images = ref_data[self.indices]
            self.target_sketches = target_data[self.indices]
            
            del ref_data, target_data
            print(f"Loaded {len(self.indices)} {split} samples to memory")
        else:
            self.ref_images = None
            self.target_sketches = None
            print(f"Loaded {split} dataset: {len(self.indices)} samples (disk-based)")
    
    def __len__(self):
        return len(self.indices)
    
    def __getitem__(self, idx):
        actual_idx = self.indices[idx]
        
        if self.load_to_memory:
            # Data already in memory
            ref_images = torch.FloatTensor(self.ref_images[idx]) / 255.0
            target_sketch = torch.FloatTensor(self.target_sketches[idx]).unsqueeze(0) / 255.0
        else:
            # Load from disk
            if not hasattr(self, '_ref_data'):
                self._ref_data = np.load(self.ref_file)['data']
                self._target_data = np.load(self.target_file)['data']
            
            ref_images = torch.FloatTensor(self._ref_data[actual_idx]) / 255.0
            target_sketch = torch.FloatTensor(self._target_data[actual_idx]).unsqueeze(0) / 255.0
        
        instruction = self.instructions[idx]
        
        return {
            "reference_images": ref_images,  # [4, H, W]
            "instruction": instruction,
            "target_sketch": target_sketch,  # [1, H, W]
            "idx": actual_idx
        }

# Usage example:
# train_dataset = InstructPix2PixDataset("instructpix2pix_dataset", "train", load_to_memory=False)
# val_dataset = InstructPix2PixDataset("instructpix2pix_dataset", "val", load_to_memory=False)
# 
# train_loader = torch.utils.data.DataLoader(
#     train_dataset, 
#     batch_size=4, 
#     shuffle=True,
#     num_workers=2,
#     pin_memory=True
# )
'''
    
    with open(os.path.join(dataset_path, "pytorch_dataset.py"), "w", encoding='utf-8') as f:
        f.write(dataset_code)
    
    print(f"PyTorch dataset class saved to {dataset_path}/pytorch_dataset.py")


def main():
    """Main function with improved error handling"""
    print("=" * 80)
    print("IMPROVED INSTRUCTPIX2PIX DATASET GENERATOR")
    print("=" * 80)
    
    # Configuration for 10k examples
    config = {
        "num_samples": 10000,  # Generate 10k examples
        "save_path": "instructpix2pix_dataset_10k",
        "batch_size": 100,     # Larger batch size for efficiency
        "canvas_size": 256,
        "complexity_distribution": {
            "simple": 0.3,     # 3,000 samples
            "medium": 0.5,     # 5,000 samples  
            "complex": 0.2     # 2,000 samples
        }
    }
    
    print("Configuration:")
    for key, value in config.items():
        print(f"  {key}: {value}")
    print()
    
    # Initialize generator
    generator = InstructPix2PixDatasetGenerator(canvas_size=config["canvas_size"])
    
    # Generate dataset
    success = generator.generate_safe_dataset(
        num_samples=config["num_samples"],
        save_path=config["save_path"],
        batch_size=config["batch_size"],
        complexity_distribution=config["complexity_distribution"]
    )
    
    if success:
        # Create PyTorch dataset loader
        create_pytorch_dataset_loader(config["save_path"])
        
        print("\n" + "=" * 80)
        print("🎉 DATASET GENERATION SUCCESSFUL!")
        print("=" * 80)
        print(f"📁 Dataset saved to: {config['save_path']}")
        print(f"📊 Total samples: {config['num_samples']}")
        print(f"🐍 PyTorch dataset loader created")
        print("=" * 80)
        
        # Print usage instructions
        print("\n📋 USAGE INSTRUCTIONS:")
        print("1. Import the dataset:")
        print("   from pytorch_dataset import InstructPix2PixDataset")
        print("2. Create datasets:")
        print("   train_ds = InstructPix2PixDataset('instructpix2pix_dataset', 'train')")
        print("   val_ds = InstructPix2PixDataset('instructpix2pix_dataset', 'val')")
        print("3. Create data loaders:")
        print("   train_loader = torch.utils.data.DataLoader(train_ds, batch_size=4)")
        
    else:
        print("\n❌ Dataset generation failed!")
        return False
    
    return True


if __name__ == "__main__":
    # Set random seed for reproducibility
    import random
    random.seed(42)
    np.random.seed(42)
    
    # Run main function
    main()
```

```python
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader
import torch.nn.functional as F
from torchvision import transforms
import numpy as np
import os
import json
import pickle
from tqdm import tqdm
import matplotlib.pyplot as plt
from typing import Dict, List, Tuple, Optional
import time
import random
import logging

# Set up logging
logging.basicConfig(level=logging.INFO)
logger = logging.getLogger(__name__)

class SimpleInstructPix2PixDataset(Dataset):
    """Simplified dataset for InstructPix2Pix training"""
    
    def __init__(self, data_path: str, split: str = "train", image_size: int = 256):
        self.data_path = data_path
        self.split = split
        self.image_size = image_size
        
        # Load main dataset files
        self.ref_file = os.path.join(data_path, "reference_scenes.npz")
        self.target_file = os.path.join(data_path, "target_sketches.npz")
        self.inst_file = os.path.join(data_path, "instructions.txt")
        
        # Load split indices
        split_path = os.path.join(data_path, split)
        indices_file = os.path.join(split_path, "indices.txt")
        
        if os.path.exists(indices_file):
            with open(indices_file, 'r') as f:
                self.indices = [int(line.strip()) for line in f.readlines()]
        else:
            # Fallback: use all data
            print(f"Warning: No split indices found, using all data")
            ref_data = np.load(self.ref_file)['data']
            self.indices = list(range(min(len(ref_data), 1000)))  # Limit for testing
            del ref_data
        
        # Load instructions
        with open(self.inst_file, 'r') as f:
            all_instructions = [line.strip() for line in f.readlines()]
        self.instructions = [all_instructions[i] for i in self.indices]
        
        # Simple vocabulary for text encoding
        self.vocab = self.build_vocabulary()
        
        print(f"Loaded {split} dataset: {len(self.indices)} samples")
    
    def build_vocabulary(self) -> Dict[str, int]:
        """Build simple vocabulary from instructions"""
        vocab = {"<PAD>": 0, "<UNK>": 1}
        vocab_idx = 2
        
        for instruction in self.instructions:
            words = instruction.lower().split()
            for word in words:
                if word not in vocab:
                    vocab[word] = vocab_idx
                    vocab_idx += 1
        
        print(f"Built vocabulary with {len(vocab)} words")
        return vocab
    
    def encode_text(self, text: str, max_length: int = 32) -> torch.Tensor:
        """Simple text encoding"""
        words = text.lower().split()
        tokens = [self.vocab.get(word, 1) for word in words]  # 1 is <UNK>
        
        # Pad or truncate
        if len(tokens) < max_length:
            tokens.extend([0] * (max_length - len(tokens)))  # 0 is <PAD>
        else:
            tokens = tokens[:max_length]
        
        return torch.tensor(tokens, dtype=torch.long)
    
    def __len__(self):
        return len(self.indices)
    
    def __getitem__(self, idx):
        actual_idx = self.indices[idx]
        
        # Load from disk (lazy loading)
        if not hasattr(self, '_ref_data'):
            print("Loading reference images...")
            self._ref_data = np.load(self.ref_file)['data']
            print("Loading target sketches...")
            self._target_data = np.load(self.target_file)['data']
        
        # Get reference images (use first one as input image)
        ref_images = self._ref_data[actual_idx]  # [4, H, W]
        input_image = ref_images[0]  # Use first reference as input
        
        # Get target image
        target_image = self._target_data[actual_idx]  # [H, W]
        
        # Get instruction
        instruction = self.instructions[idx]
        text_tokens = self.encode_text(instruction)
        
        # Convert to tensors and normalize to [-1, 1]
        input_image = torch.from_numpy(input_image).float() / 127.5 - 1.0
        target_image = torch.from_numpy(target_image).float() / 127.5 - 1.0
        
        # Add channel dimension
        input_image = input_image.unsqueeze(0)
        target_image = target_image.unsqueeze(0)
        
        return {
            "input_image": input_image,
            "target_image": target_image,
            "text_tokens": text_tokens,
            "instruction": instruction,
            "idx": actual_idx
        }

class SimpleUNet(nn.Module):
    """Simplified U-Net for image-to-image translation with text conditioning"""
    
    def __init__(self, vocab_size: int, text_embed_dim: int = 128, image_channels: int = 1):
        super().__init__()
        
        # Text embedding
        self.text_embedding = nn.Embedding(vocab_size, text_embed_dim)
        self.text_encoder = nn.LSTM(text_embed_dim, text_embed_dim, batch_first=True)
        
        # U-Net encoder
        self.enc1 = self.conv_block(image_channels, 64)
        self.enc2 = self.conv_block(64, 128)
        self.enc3 = self.conv_block(128, 256)
        self.enc4 = self.conv_block(256, 512)
        
        # Bottleneck with text conditioning
        self.bottleneck = self.conv_block(512 + text_embed_dim, 1024)
        
        # U-Net decoder
        self.dec4 = self.conv_block(1024 + 512, 512)
        self.dec3 = self.conv_block(512 + 256, 256)
        self.dec2 = self.conv_block(256 + 128, 128)
        self.dec1 = self.conv_block(128 + 64, 64)
        
        # Output layer
        self.output = nn.Conv2d(64, image_channels, 1)
        
        # Pooling and upsampling
        self.pool = nn.MaxPool2d(2)
        self.upsample = nn.Upsample(scale_factor=2, mode='bilinear', align_corners=True)
    
    def conv_block(self, in_channels: int, out_channels: int) -> nn.Module:
        """Convolutional block with batch norm and ReLU"""
        return nn.Sequential(
            nn.Conv2d(in_channels, out_channels, 3, padding=1),
            nn.BatchNorm2d(out_channels),
            nn.ReLU(inplace=True),
            nn.Conv2d(out_channels, out_channels, 3, padding=1),
            nn.BatchNorm2d(out_channels),
            nn.ReLU(inplace=True)
        )
    
    def forward(self, x: torch.Tensor, text_tokens: torch.Tensor) -> torch.Tensor:
        # Encode text
        text_embeds = self.text_embedding(text_tokens)  # [B, seq_len, embed_dim]
        text_output, (h_n, c_n) = self.text_encoder(text_embeds)
        text_features = h_n[-1]  # [B, embed_dim]
        
        # U-Net encoder with pooling
        e1 = self.enc1(x)  # [B, 64, 256, 256]
        e1_pooled = self.pool(e1)  # [B, 64, 128, 128]
        
        e2 = self.enc2(e1_pooled)  # [B, 128, 128, 128]
        e2_pooled = self.pool(e2)  # [B, 128, 64, 64]
        
        e3 = self.enc3(e2_pooled)  # [B, 256, 64, 64]
        e3_pooled = self.pool(e3)  # [B, 256, 32, 32]
        
        e4 = self.enc4(e3_pooled)  # [B, 512, 32, 32]
        e4_pooled = self.pool(e4)  # [B, 512, 16, 16]
        
        # Bottleneck with text conditioning
        # Broadcast text features to match the spatial dimensions of e4_pooled
        text_spatial = text_features.unsqueeze(-1).unsqueeze(-1)  # [B, embed_dim, 1, 1]
        text_spatial = text_spatial.expand(-1, -1, e4_pooled.size(2), e4_pooled.size(3))  # [B, embed_dim, 16, 16]
        
        bottleneck_input = torch.cat([e4_pooled, text_spatial], dim=1)  # [B, 512+embed_dim, 16, 16]
        bottleneck = self.bottleneck(bottleneck_input)  # [B, 1024, 16, 16]
        
        # U-Net decoder with skip connections
        d4 = self.dec4(torch.cat([self.upsample(bottleneck), e4], dim=1))  # [B, 512, 32, 32]
        d3 = self.dec3(torch.cat([self.upsample(d4), e3], dim=1))  # [B, 256, 64, 64]
        d2 = self.dec2(torch.cat([self.upsample(d3), e2], dim=1))  # [B, 128, 128, 128]
        d1 = self.dec1(torch.cat([self.upsample(d2), e1], dim=1))  # [B, 64, 256, 256]
        
        # Output
        output = torch.tanh(self.output(d1))  # [B, 1, 256, 256]
        return output

class SimpleInstructPix2PixTrainer:
    """Simplified trainer for InstructPix2Pix"""
    
    def __init__(self, config: Dict):
        self.config = config
        self.device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
        print(f"Using device: {self.device}")
        
        # Initialize datasets
        self.train_dataset = SimpleInstructPix2PixDataset(
            config["dataset_path"], 
            "train", 
            config.get("image_size", 256)
        )
        self.val_dataset = SimpleInstructPix2PixDataset(
            config["dataset_path"], 
            "val", 
            config.get("image_size", 256)
        )
        
        # Initialize model
        vocab_size = len(self.train_dataset.vocab)
        self.model = SimpleUNet(
            vocab_size=vocab_size,
            text_embed_dim=config.get("text_embed_dim", 128),
            image_channels=1
        ).to(self.device)
        
        print(f"Model initialized with {sum(p.numel() for p in self.model.parameters()):,} parameters")
        
        # Test model with dummy input to verify shapes
        try:
            with torch.no_grad():
                dummy_image = torch.randn(1, 1, 256, 256).to(self.device)
                dummy_text = torch.randint(0, vocab_size, (1, 32)).to(self.device)
                dummy_output = self.model(dummy_image, dummy_text)
                print(f"Model test successful! Output shape: {dummy_output.shape}")
        except Exception as e:
            print(f"Model test failed: {e}")
            raise e
        
        # Initialize data loaders
        self.train_loader = DataLoader(
            self.train_dataset,
            batch_size=config["batch_size"],
            shuffle=True,
            num_workers=0,  # Set to 0 for Windows compatibility
            pin_memory=True if torch.cuda.is_available() else False
        )
        self.val_loader = DataLoader(
            self.val_dataset,
            batch_size=config["batch_size"],
            shuffle=False,
            num_workers=0,
            pin_memory=True if torch.cuda.is_available() else False
        )
        
        # Initialize optimizer and loss
        self.optimizer = optim.Adam(
            self.model.parameters(),
            lr=config["learning_rate"],
            weight_decay=config.get("weight_decay", 1e-4)
        )
        
        self.criterion = nn.MSELoss()
        
        # Training state
        self.global_step = 0
        self.best_val_loss = float('inf')
        
        # Create output directory
        os.makedirs(config["output_dir"], exist_ok=True)
    
    def train_step(self, batch: Dict) -> float:
        """Single training step"""
        self.model.train()
        
        try:
            input_images = batch["input_image"].to(self.device)
            target_images = batch["target_image"].to(self.device)
            text_tokens = batch["text_tokens"].to(self.device)
            
            # Forward pass
            predicted_images = self.model(input_images, text_tokens)
            
            # Compute loss
            loss = self.criterion(predicted_images, target_images)
            
            # Backward pass
            self.optimizer.zero_grad()
            loss.backward()
            
            # Gradient clipping
            if self.config.get("max_grad_norm", None):
                torch.nn.utils.clip_grad_norm_(self.model.parameters(), self.config["max_grad_norm"])
            
            self.optimizer.step()
            
            return loss.item()
            
        except RuntimeError as e:
            if "out of memory" in str(e):
                print(f"WARNING: GPU out of memory. Try reducing batch size. Error: {e}")
                if torch.cuda.is_available():
                    torch.cuda.empty_cache()
                return 0.0
            else:
                raise e
    
    def validate(self) -> float:
        """Validation step"""
        self.model.eval()
        val_losses = []
        
        with torch.no_grad():
            for batch in tqdm(self.val_loader, desc="Validating", leave=False):
                input_images = batch["input_image"].to(self.device)
                target_images = batch["target_image"].to(self.device)
                text_tokens = batch["text_tokens"].to(self.device)
                
                # Forward pass
                predicted_images = self.model(input_images, text_tokens)
                
                # Compute loss
                loss = self.criterion(predicted_images, target_images)
                val_losses.append(loss.item())
        
        return np.mean(val_losses)
    
    def generate_samples(self, num_samples: int = 4) -> List[Dict]:
        """Generate sample images for visualization"""
        self.model.eval()
        samples = []
        
        with torch.no_grad():
            # Get some validation samples
            val_batch = next(iter(self.val_loader))
            input_images = val_batch["input_image"][:num_samples].to(self.device)
            target_images = val_batch["target_image"][:num_samples].to(self.device)
            text_tokens = val_batch["text_tokens"][:num_samples].to(self.device)
            instructions = val_batch["instruction"][:num_samples]
            
            # Generate images
            generated_images = self.model(input_images, text_tokens)
            
            # Convert to numpy for visualization
            for i in range(num_samples):
                input_np = (input_images[i].cpu().numpy()[0] + 1) / 2  # Denormalize to [0,1]
                target_np = (target_images[i].cpu().numpy()[0] + 1) / 2
                generated_np = (generated_images[i].cpu().numpy()[0] + 1) / 2
                
                samples.append({
                    "input": input_np,
                    "target": target_np,
                    "generated": generated_np,
                    "instruction": instructions[i]
                })
        
        return samples
    
    def save_checkpoint(self, epoch: int, val_loss: float):
        """Save model checkpoint"""
        checkpoint = {
            "epoch": epoch,
            "global_step": self.global_step,
            "model_state_dict": self.model.state_dict(),
            "optimizer_state_dict": self.optimizer.state_dict(),
            "val_loss": val_loss,
            "config": self.config,
            "vocab": self.train_dataset.vocab
        }
        
        checkpoint_path = os.path.join(self.config["output_dir"], f"checkpoint_epoch_{epoch}.pt")
        torch.save(checkpoint, checkpoint_path)
        
        # Save best model
        if val_loss < self.best_val_loss:
            self.best_val_loss = val_loss
            best_path = os.path.join(self.config["output_dir"], "best_model.pt")
            torch.save(checkpoint, best_path)
            logger.info(f"New best model saved with val_loss: {val_loss:.4f}")
    
    def save_samples(self, samples: List[Dict], epoch: int):
        """Save sample images to disk"""
        samples_dir = os.path.join(self.config["output_dir"], "samples")
        os.makedirs(samples_dir, exist_ok=True)
        
        for i, sample in enumerate(samples):
            fig, axes = plt.subplots(1, 3, figsize=(15, 5))
            
            # Input image
            axes[0].imshow(sample["input"], cmap='gray', vmin=0, vmax=1)
            axes[0].set_title("Input")
            axes[0].axis('off')
            
            # Target image
            axes[1].imshow(sample["target"], cmap='gray', vmin=0, vmax=1)
            axes[1].set_title("Target")
            axes[1].axis('off')
            
            # Generated image
            axes[2].imshow(sample["generated"], cmap='gray', vmin=0, vmax=1)
            axes[2].set_title("Generated")
            axes[2].axis('off')
            
            plt.suptitle(f"Instruction: {sample['instruction'][:60]}...")
            plt.tight_layout()
            
            sample_path = os.path.join(samples_dir, f"epoch_{epoch}_sample_{i}.png")
            plt.savefig(sample_path, dpi=150, bbox_inches='tight')
            plt.close(fig)
    
    def train(self):
        """Main training loop"""
        logger.info("Starting lightweight InstructPix2Pix training...")
        logger.info(f"Training samples: {len(self.train_dataset)}")
        logger.info(f"Validation samples: {len(self.val_dataset)}")
        logger.info(f"Batch size: {self.config['batch_size']}")
        logger.info(f"Learning rate: {self.config['learning_rate']}")
        logger.info(f"Model parameters: {sum(p.numel() for p in self.model.parameters()):,}")
        
        start_time = time.time()
        
        for epoch in range(self.config["num_epochs"]):
            # Training
            self.model.train()
            train_losses = []
            
            progress_bar = tqdm(
                self.train_loader,
                desc=f"Epoch {epoch+1}/{self.config['num_epochs']}",
                leave=True
            )
            
            epoch_start_time = time.time()
            
            for batch_idx, batch in enumerate(progress_bar):
                loss = self.train_step(batch)
                train_losses.append(loss)
                self.global_step += 1
                
                # Update progress bar
                avg_loss = np.mean(train_losses[-50:])  # Average of last 50 batches
                progress_bar.set_postfix({
                    "loss": f"{avg_loss:.4f}",
                    "lr": f"{self.optimizer.param_groups[0]['lr']:.2e}"
                })
                
                # Quick validation every 500 steps
                if self.global_step % 500 == 0:
                    val_loss = self.validate()
                    logger.info(f"Step {self.global_step} - Val Loss: {val_loss:.4f}")
                    
                    # Clear GPU memory
                    if torch.cuda.is_available():
                        torch.cuda.empty_cache()
            
            # End of epoch validation
            val_loss = self.validate()
            train_loss = np.mean(train_losses)
            epoch_time = time.time() - epoch_start_time
            
            logger.info(f"Epoch {epoch+1} completed in {epoch_time:.1f}s:")
            logger.info(f"  Train Loss: {train_loss:.4f}")
            logger.info(f"  Val Loss: {val_loss:.4f}")
            
            # Generate and save sample images
            if self.config.get("generate_samples", True):
                samples = self.generate_samples(num_samples=4)
                self.save_samples(samples, epoch)
                logger.info(f"  Sample images saved to {self.config['output_dir']}/samples/")
            
            # Save checkpoint
            self.save_checkpoint(epoch, val_loss)
            
            # Learning rate decay
            if epoch > 0 and epoch % 5 == 0:
                for param_group in self.optimizer.param_groups:
                    param_group['lr'] *= 0.8
                logger.info(f"  Learning rate decayed to {self.optimizer.param_groups[0]['lr']:.2e}")
        
        total_time = time.time() - start_time
        logger.info(f"Training completed in {total_time:.1f}s ({total_time/60:.1f} minutes)!")
        logger.info(f"Best validation loss: {self.best_val_loss:.4f}")


def main():
    """Main training function for lightweight InstructPix2Pix"""
    
    # Lightweight training configuration
    config = {
        # Dataset
        "dataset_path": "instructpix2pix_dataset_10k",
        "image_size": 256,
        
        # Model
        "text_embed_dim": 128,
        
        # Training - Much faster settings
        "batch_size": 4,  # Further reduced batch size for stability
        "num_epochs": 20,  # More epochs since each epoch is much faster
        "learning_rate": 1e-3,  # Higher learning rate
        "weight_decay": 1e-4,
        "max_grad_norm": 1.0,
        
        # Validation and logging
        "generate_samples": True,
        
        # System
        "num_workers": 0,  # Windows compatibility
        "output_dir": "./lightweight_instructpix2pix_checkpoints"
    }
    
    # Check if dataset exists
    if not os.path.exists(config["dataset_path"]):
        print(f"Error: Dataset not found at {config['dataset_path']}")
        print("Please run the dataset generator first!")
        return
    
    print("=" * 80)
    print("LIGHTWEIGHT INSTRUCTPIX2PIX TRAINING FROM SCRATCH")
    print("=" * 80)
    print(f"Dataset: {config['dataset_path']}")
    print(f"Batch size: {config['batch_size']}")
    print(f"Learning rate: {config['learning_rate']}")
    print(f"Epochs: {config['num_epochs']}")
    print(f"Output directory: {config['output_dir']}")
    print(f"Using GPU: {torch.cuda.is_available()}")
    print("=" * 80)
    
    # Initialize and start training
    trainer = SimpleInstructPix2PixTrainer(config)
    trainer.train()
    
    print("=" * 80)
    print("TRAINING COMPLETED!")
    print("=" * 80)
    print(f"Checkpoints saved to: {config['output_dir']}")
    print(f"Sample images saved to: {config['output_dir']}/samples/")
    print("=" * 80)

if __name__ == "__main__":
    main()
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import numpy as np
import matplotlib.pyplot as plt
import os
from typing import Dict, List
import random
from torch.utils.data import DataLoader

# Import the model architecture and dataset from the training script
class SimpleUNet(nn.Module):
    """Simplified U-Net for image-to-image translation with text conditioning"""
    
    def __init__(self, vocab_size: int, text_embed_dim: int = 128, image_channels: int = 1):
        super().__init__()
        
        # Text embedding
        self.text_embedding = nn.Embedding(vocab_size, text_embed_dim)
        self.text_encoder = nn.LSTM(text_embed_dim, text_embed_dim, batch_first=True)
        
        # U-Net encoder
        self.enc1 = self.conv_block(image_channels, 64)
        self.enc2 = self.conv_block(64, 128)
        self.enc3 = self.conv_block(128, 256)
        self.enc4 = self.conv_block(256, 512)
        
        # Bottleneck with text conditioning
        self.bottleneck = self.conv_block(512 + text_embed_dim, 1024)
        
        # U-Net decoder
        self.dec4 = self.conv_block(1024 + 512, 512)
        self.dec3 = self.conv_block(512 + 256, 256)
        self.dec2 = self.conv_block(256 + 128, 128)
        self.dec1 = self.conv_block(128 + 64, 64)
        
        # Output layer
        self.output = nn.Conv2d(64, image_channels, 1)
        
        # Pooling and upsampling
        self.pool = nn.MaxPool2d(2)
        self.upsample = nn.Upsample(scale_factor=2, mode='bilinear', align_corners=True)
    
    def conv_block(self, in_channels: int, out_channels: int) -> nn.Module:
        """Convolutional block with batch norm and ReLU"""
        return nn.Sequential(
            nn.Conv2d(in_channels, out_channels, 3, padding=1),
            nn.BatchNorm2d(out_channels),
            nn.ReLU(inplace=True),
            nn.Conv2d(out_channels, out_channels, 3, padding=1),
            nn.BatchNorm2d(out_channels),
            nn.ReLU(inplace=True)
        )
    
    def forward(self, x: torch.Tensor, text_tokens: torch.Tensor) -> torch.Tensor:
        # Encode text
        text_embeds = self.text_embedding(text_tokens)  # [B, seq_len, embed_dim]
        text_output, (h_n, c_n) = self.text_encoder(text_embeds)
        text_features = h_n[-1]  # [B, embed_dim]
        
        # U-Net encoder with pooling
        e1 = self.enc1(x)  # [B, 64, 256, 256]
        e1_pooled = self.pool(e1)  # [B, 64, 128, 128]
        
        e2 = self.enc2(e1_pooled)  # [B, 128, 128, 128]
        e2_pooled = self.pool(e2)  # [B, 128, 64, 64]
        
        e3 = self.enc3(e2_pooled)  # [B, 256, 64, 64]
        e3_pooled = self.pool(e3)  # [B, 256, 32, 32]
        
        e4 = self.enc4(e3_pooled)  # [B, 512, 32, 32]
        e4_pooled = self.pool(e4)  # [B, 512, 16, 16]
        
        # Bottleneck with text conditioning
        # Broadcast text features to match the spatial dimensions of e4_pooled
        text_spatial = text_features.unsqueeze(-1).unsqueeze(-1)  # [B, embed_dim, 1, 1]
        text_spatial = text_spatial.expand(-1, -1, e4_pooled.size(2), e4_pooled.size(3))  # [B, embed_dim, 16, 16]
        
        bottleneck_input = torch.cat([e4_pooled, text_spatial], dim=1)  # [B, 512+embed_dim, 16, 16]
        bottleneck = self.bottleneck(bottleneck_input)  # [B, 1024, 16, 16]
        
        # U-Net decoder with skip connections
        d4 = self.dec4(torch.cat([self.upsample(bottleneck), e4], dim=1))  # [B, 512, 32, 32]
        d3 = self.dec3(torch.cat([self.upsample(d4), e3], dim=1))  # [B, 256, 64, 64]
        d2 = self.dec2(torch.cat([self.upsample(d3), e2], dim=1))  # [B, 128, 128, 128]
        d1 = self.dec1(torch.cat([self.upsample(d2), e1], dim=1))  # [B, 64, 256, 256]
        
        # Output
        output = torch.tanh(self.output(d1))  # [B, 1, 256, 256]
        return output

class SimpleInstructPix2PixDataset:
    """Simplified dataset for InstructPix2Pix testing"""
    
    def __init__(self, data_path: str, split: str = "test", image_size: int = 256):
        self.data_path = data_path
        self.split = split
        self.image_size = image_size
        
        # Load main dataset files
        self.ref_file = os.path.join(data_path, "reference_scenes.npz")
        self.target_file = os.path.join(data_path, "target_sketches.npz")
        self.inst_file = os.path.join(data_path, "instructions.txt")
        
        # Load split indices
        split_path = os.path.join(data_path, split)
        indices_file = os.path.join(split_path, "indices.txt")
        
        if os.path.exists(indices_file):
            with open(indices_file, 'r') as f:
                self.indices = [int(line.strip()) for line in f.readlines()]
        else:
            # Fallback: use validation data for testing
            print(f"Warning: No {split} split found, using validation data")
            val_path = os.path.join(data_path, "val")
            val_indices_file = os.path.join(val_path, "indices.txt")
            if os.path.exists(val_indices_file):
                with open(val_indices_file, 'r') as f:
                    self.indices = [int(line.strip()) for line in f.readlines()]
            else:
                # Last resort: use some data
                ref_data = np.load(self.ref_file)['data']
                self.indices = list(range(min(len(ref_data), 100)))
                del ref_data
        
        # Load instructions
        with open(self.inst_file, 'r') as f:
            all_instructions = [line.strip() for line in f.readlines()]
        self.instructions = [all_instructions[i] for i in self.indices]
        
        print(f"Loaded {split} dataset: {len(self.indices)} samples")
    
    def encode_text(self, text: str, vocab: Dict[str, int], max_length: int = 32) -> torch.Tensor:
        """Simple text encoding using provided vocabulary"""
        words = text.lower().split()
        tokens = [vocab.get(word, 1) for word in words]  # 1 is <UNK>
        
        # Pad or truncate
        if len(tokens) < max_length:
            tokens.extend([0] * (max_length - len(tokens)))  # 0 is <PAD>
        else:
            tokens = tokens[:max_length]
        
        return torch.tensor(tokens, dtype=torch.long)
    
    def get_sample(self, idx: int, vocab: Dict[str, int]):
        """Get a single sample for testing"""
        actual_idx = self.indices[idx]
        
        # Load from disk
        ref_data = np.load(self.ref_file)['data']
        target_data = np.load(self.target_file)['data']
        
        # Get reference images (use first one as input image)
        ref_images = ref_data[actual_idx]  # [4, H, W]
        input_image = ref_images[0]  # Use first reference as input
        
        # Get target image
        target_image = target_data[actual_idx]  # [H, W]
        
        # Get instruction
        instruction = self.instructions[idx]
        text_tokens = self.encode_text(instruction, vocab)
        
        # Convert to tensors and normalize to [-1, 1]
        input_image = torch.from_numpy(input_image).float() / 127.5 - 1.0
        target_image = torch.from_numpy(target_image).float() / 127.5 - 1.0
        
        # Add channel dimension
        input_image = input_image.unsqueeze(0).unsqueeze(0)  # [1, 1, H, W]
        target_image = target_image.unsqueeze(0).unsqueeze(0)  # [1, 1, H, W]
        text_tokens = text_tokens.unsqueeze(0)  # [1, seq_len]
        
        return {
            "input_image": input_image,
            "target_image": target_image,
            "text_tokens": text_tokens,
            "instruction": instruction,
            "idx": actual_idx
        }

class InstructPix2PixTester:
    """Test the trained InstructPix2Pix model"""
    
    def __init__(self, checkpoint_path: str, dataset_path: str):
        self.device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
        print(f"Using device: {self.device}")
        
        # Load checkpoint
        print(f"Loading checkpoint from: {checkpoint_path}")
        self.checkpoint = torch.load(checkpoint_path, map_location=self.device)
        
        # Get vocabulary and config
        self.vocab = self.checkpoint['vocab']
        self.config = self.checkpoint['config']
        
        # Initialize model
        vocab_size = len(self.vocab)
        self.model = SimpleUNet(
            vocab_size=vocab_size,
            text_embed_dim=self.config.get("text_embed_dim", 128),
            image_channels=1
        ).to(self.device)
        
        # Load model weights
        self.model.load_state_dict(self.checkpoint['model_state_dict'])
        self.model.eval()
        
        print(f"Model loaded successfully!")
        print(f"Epoch: {self.checkpoint['epoch']}")
        print(f"Validation loss: {self.checkpoint['val_loss']:.4f}")
        print(f"Model parameters: {sum(p.numel() for p in self.model.parameters()):,}")
        
        # Initialize test dataset
        self.test_dataset = SimpleInstructPix2PixDataset(dataset_path, "test")
        
    def generate_single(self, input_image: torch.Tensor, text_tokens: torch.Tensor) -> torch.Tensor:
        """Generate a single image"""
        with torch.no_grad():
            input_image = input_image.to(self.device)
            text_tokens = text_tokens.to(self.device)
            
            generated = self.model(input_image, text_tokens)
            return generated.cpu()
    
    def test_random_samples(self, num_samples: int = 8, save_path: str = "./test_results"):
        """Test on random samples and save visualizations"""
        os.makedirs(save_path, exist_ok=True)
        
        print(f"\nTesting on {num_samples} random samples...")
        
        # Get random sample indices
        sample_indices = random.sample(range(len(self.test_dataset.indices)), num_samples)
        
        results = []
        
        for i, idx in enumerate(sample_indices):
            print(f"Processing sample {i+1}/{num_samples}...")
            
            # Get sample
            sample = self.test_dataset.get_sample(idx, self.vocab)
            
            # Generate image
            generated = self.generate_single(sample["input_image"], sample["text_tokens"])
            
            # Convert to numpy for visualization
            input_np = (sample["input_image"][0, 0].numpy() + 1) / 2  # Denormalize to [0,1]
            target_np = (sample["target_image"][0, 0].numpy() + 1) / 2
            generated_np = (generated[0, 0].numpy() + 1) / 2
            
            # Clamp values to [0, 1]
            input_np = np.clip(input_np, 0, 1)
            target_np = np.clip(target_np, 0, 1)
            generated_np = np.clip(generated_np, 0, 1)
            
            result = {
                "input": input_np,
                "target": target_np,
                "generated": generated_np,
                "instruction": sample["instruction"],
                "idx": sample["idx"]
            }
            results.append(result)
            
            # Save individual comparison
            self.save_single_comparison(result, os.path.join(save_path, f"test_sample_{i+1}.png"))
        
        # Save grid comparison
        self.save_grid_comparison(results, os.path.join(save_path, "test_grid.png"))
        
        # Calculate and print metrics
        self.calculate_metrics(results)
        
        return results
    
    def save_single_comparison(self, result: Dict, save_path: str):
        """Save a single comparison image"""
        fig, axes = plt.subplots(1, 3, figsize=(15, 5))
        
        # Input image
        axes[0].imshow(result["input"], cmap='gray', vmin=0, vmax=1)
        axes[0].set_title("Input Image", fontsize=12, fontweight='bold')
        axes[0].axis('off')
        
        # Target image
        axes[1].imshow(result["target"], cmap='gray', vmin=0, vmax=1)
        axes[1].set_title("Target (Ground Truth)", fontsize=12, fontweight='bold')
        axes[1].axis('off')
        
        # Generated image
        axes[2].imshow(result["generated"], cmap='gray', vmin=0, vmax=1)
        axes[2].set_title("Generated Output", fontsize=12, fontweight='bold')
        axes[2].axis('off')
        
        # Add instruction as title
        instruction = result['instruction']
        if len(instruction) > 80:
            instruction = instruction[:80] + "..."
        
        plt.suptitle(f"Instruction: \"{instruction}\"", fontsize=14, fontweight='bold', y=0.95)
        plt.tight_layout()
        plt.subplots_adjust(top=0.85)
        
        plt.savefig(save_path, dpi=150, bbox_inches='tight', facecolor='white')
        plt.close(fig)
    
    def save_grid_comparison(self, results: List[Dict], save_path: str):
        """Save a grid comparison of all results"""
        num_samples = len(results)
        cols = 3  # input, target, generated
        rows = num_samples
        
        fig, axes = plt.subplots(rows, cols, figsize=(15, 5 * rows))
        
        if rows == 1:
            axes = axes.reshape(1, -1)
        
        for i, result in enumerate(results):
            # Input
            axes[i, 0].imshow(result["input"], cmap='gray', vmin=0, vmax=1)
            axes[i, 0].set_title(f"Sample {i+1}: Input", fontsize=10)
            axes[i, 0].axis('off')
            
            # Target
            axes[i, 1].imshow(result["target"], cmap='gray', vmin=0, vmax=1)
            axes[i, 1].set_title("Target", fontsize=10)
            axes[i, 1].axis('off')
            
            # Generated
            axes[i, 2].imshow(result["generated"], cmap='gray', vmin=0, vmax=1)
            axes[i, 2].set_title("Generated", fontsize=10)
            axes[i, 2].axis('off')
            
            # Add instruction as ylabel
            instruction = result['instruction']
            if len(instruction) > 40:
                instruction = instruction[:40] + "..."
            axes[i, 0].set_ylabel(f"\"{instruction}\"", fontsize=9, rotation=90, labelpad=20)
        
        plt.suptitle("InstructPix2Pix Test Results Comparison", fontsize=16, fontweight='bold')
        plt.tight_layout()
        plt.subplots_adjust(top=0.95)
        
        plt.savefig(save_path, dpi=150, bbox_inches='tight', facecolor='white')
        plt.close(fig)
    
    def calculate_metrics(self, results: List[Dict]):
        """Calculate and print evaluation metrics"""
        print("\n" + "="*60)
        print("EVALUATION METRICS")
        print("="*60)
        
        mse_scores = []
        psnr_scores = []
        
        for result in results:
            target = result["target"]
            generated = result["generated"]
            
            # MSE
            mse = np.mean((target - generated) ** 2)
            mse_scores.append(mse)
            
            # PSNR
            if mse > 0:
                psnr = 20 * np.log10(1.0 / np.sqrt(mse))
            else:
                psnr = float('inf')
            psnr_scores.append(psnr)
        
        print(f"Average MSE: {np.mean(mse_scores):.6f} ± {np.std(mse_scores):.6f}")
        print(f"Average PSNR: {np.mean(psnr_scores):.2f} dB ± {np.std(psnr_scores):.2f} dB")
        print(f"Best PSNR: {np.max(psnr_scores):.2f} dB")
        print(f"Worst PSNR: {np.min(psnr_scores):.2f} dB")
        
        # Print per-sample results
        print(f"\nPer-sample results:")
        for i, (mse, psnr) in enumerate(zip(mse_scores, psnr_scores)):
            instruction = results[i]['instruction'][:50] + "..." if len(results[i]['instruction']) > 50 else results[i]['instruction']
            print(f"Sample {i+1}: MSE={mse:.6f}, PSNR={psnr:.2f} dB | \"{instruction}\"")
    
    def test_custom_instruction(self, instruction: str, sample_idx: int = 0, save_path: str = "./custom_test.png"):
        """Test with a custom instruction on a specific sample"""
        print(f"\nTesting custom instruction: \"{instruction}\"")
        
        # Get sample
        sample = self.test_dataset.get_sample(sample_idx, self.vocab)
        
        # Encode custom instruction
        custom_tokens = self.test_dataset.encode_text(instruction, self.vocab).unsqueeze(0)
        
        # Generate with custom instruction
        generated = self.generate_single(sample["input_image"], custom_tokens)
        
        # Convert to numpy
        input_np = (sample["input_image"][0, 0].numpy() + 1) / 2
        target_np = (sample["target_image"][0, 0].numpy() + 1) / 2
        generated_np = (generated[0, 0].numpy() + 1) / 2
        original_instruction = sample["instruction"]
        
        # Clamp values
        input_np = np.clip(input_np, 0, 1)
        target_np = np.clip(target_np, 0, 1)
        generated_np = np.clip(generated_np, 0, 1)
        
        # Save comparison
        fig, axes = plt.subplots(1, 4, figsize=(20, 5))
        
        axes[0].imshow(input_np, cmap='gray', vmin=0, vmax=1)
        axes[0].set_title("Input Image", fontsize=12, fontweight='bold')
        axes[0].axis('off')
        
        axes[1].imshow(target_np, cmap='gray', vmin=0, vmax=1)
        axes[1].set_title(f"Original Target\n\"{original_instruction[:30]}...\"", fontsize=10)
        axes[1].axis('off')
        
        axes[2].imshow(generated_np, cmap='gray', vmin=0, vmax=1)
        axes[2].set_title(f"Generated with Custom Instruction", fontsize=12, fontweight='bold')
        axes[2].axis('off')
        
        # Show difference
        diff = np.abs(target_np - generated_np)
        axes[3].imshow(diff, cmap='hot', vmin=0, vmax=1)
        axes[3].set_title("Difference Map", fontsize=12)
        axes[3].axis('off')
        
        plt.suptitle(f"Custom Instruction Test: \"{instruction}\"", fontsize=14, fontweight='bold')
        plt.tight_layout()
        
        plt.savefig(save_path, dpi=150, bbox_inches='tight', facecolor='white')
        plt.close(fig)
        
        print(f"Custom test result saved to: {save_path}")


def main():
    """Main testing function"""
    
    # Configuration
    checkpoint_path = r"\lightweight_instructpix2pix_checkpoints\checkpoint_epoch_4.pt"
    dataset_path = "instructpix2pix_dataset_10k"
    results_dir = "./test_results_epoch_4"
    
    print("="*80)
    print("INSTRUCTPIX2PIX MODEL TESTING")
    print("="*80)
    print(f"Checkpoint: {checkpoint_path}")
    print(f"Dataset: {dataset_path}")
    print(f"Results directory: {results_dir}")
    print("="*80)
    
    # Check if files exist
    if not os.path.exists(checkpoint_path):
        print(f"Error: Checkpoint not found at {checkpoint_path}")
        return
    
    if not os.path.exists(dataset_path):
        print(f"Error: Dataset not found at {dataset_path}")
        return
    
    # Initialize tester
    tester = InstructPix2PixTester(checkpoint_path, dataset_path)
    
    # Test on random samples
    print("\n" + "="*60)
    print("TESTING ON RANDOM SAMPLES")
    print("="*60)
    
    results = tester.test_random_samples(num_samples=8, save_path=results_dir)
    
    # Test custom instructions
    print("\n" + "="*60)
    print("TESTING CUSTOM INSTRUCTIONS")
    print("="*60)
    
    custom_instructions = [
        "make it look like a simple line drawing",
        "convert to a sketch with clean lines",
        "turn this into a minimalist outline",
        "create a basic wireframe version"
    ]
    
    for i, instruction in enumerate(custom_instructions):
        save_path = os.path.join(results_dir, f"custom_instruction_{i+1}.png")
        tester.test_custom_instruction(instruction, sample_idx=i, save_path=save_path)
    
    print("\n" + "="*80)
    print("TESTING COMPLETED!")
    print("="*80)
    print(f"Results saved to: {results_dir}")
    print(f"Check the following files:")
    print(f"  - test_grid.png: Overview of all test samples")
    print(f"  - test_sample_*.png: Individual comparisons")
    print(f"  - custom_instruction_*.png: Custom instruction tests")
    print("="*80)

if __name__ == "__main__":
    main()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Circle, Rectangle, Polygon, Ellipse
import os
import random
from typing import List, Tuple, Dict
from tqdm import tqdm
import json

class GeometricCompositionDatasetGenerator:
    """Generate dataset for image composition tasks with geometric shapes"""
    
    def __init__(self, image_size: int = 256):
        self.image_size = image_size
        self.colors = {
            'white': 255,
            'black': 0,
            'gray': 128,
            'light_gray': 192,
            'dark_gray': 64
        }
        
        # Define specific composition instructions
        self.composition_templates = [
            # Spatial arrangements
            "Place {obj1} on top of {obj2}",
            "Place {obj1} to the left of {obj2}",
            "Place {obj1} to the right of {obj2}", 
            "Place {obj1} below {obj2}",
            "Center {obj1} inside {obj2}",
            "Arrange {obj1} and {obj2} side by side",
            "Stack {obj1} and {obj2} vertically",
            
            # Size modifications
            "Make {obj1} larger than {obj2} and place them together",
            "Make {obj1} smaller than {obj2} and place them together",
            "Resize {obj1} to match {obj2} size and align them",
            
            # Overlapping and intersection
            "Overlap {obj1} and {obj2} partially",
            "Make {obj1} intersect with {obj2}",
            "Place {obj1} inside {obj2}",
            "Surround {obj1} with {obj2}",
            
            # Pattern combinations
            "Combine the pattern from {obj1} with the shape of {obj2}",
            "Apply the texture of {obj1} to {obj2}",
            "Merge {obj1} and {obj2} into a single object",
            
            # Multi-object arrangements
            "Arrange {obj1}, {obj2}, and {obj3} in a line",
            "Create a triangle formation with {obj1}, {obj2}, and {obj3}",
            "Place {obj1} in the center with {obj2} and {obj3} around it"
        ]
        
        self.shape_names = {
            'circle': 'circle',
            'rectangle': 'rectangle', 
            'triangle': 'triangle',
            'square': 'square',
            'oval': 'oval'
        }
        
        self.pattern_names = {
            'solid': 'solid shape',
            'striped': 'striped pattern',
            'dotted': 'dotted pattern',
            'checkered': 'checkered pattern',
            'grid': 'grid pattern'
        }
    
    def create_shape(self, shape_type: str, size: int, pattern: str = 'solid') -> np.ndarray:
        """Create a single geometric shape with pattern"""
        img = np.zeros((self.image_size, self.image_size), dtype=np.uint8)
        
        # Random position (ensuring shape fits)
        margin = size // 2 + 10
        center_x = random.randint(margin, self.image_size - margin)
        center_y = random.randint(margin, self.image_size - margin)
        
        if shape_type == 'circle':
            self._draw_circle(img, center_x, center_y, size // 2, pattern)
        elif shape_type == 'rectangle':
            self._draw_rectangle(img, center_x, center_y, size, size // 2, pattern)
        elif shape_type == 'square':
            self._draw_rectangle(img, center_x, center_y, size, size, pattern)
        elif shape_type == 'triangle':
            self._draw_triangle(img, center_x, center_y, size, pattern)
        elif shape_type == 'oval':
            self._draw_oval(img, center_x, center_y, size, size // 2, pattern)
        
        return img
    
    def _draw_circle(self, img: np.ndarray, cx: int, cy: int, radius: int, pattern: str):
        """Draw circle with pattern"""
        y, x = np.ogrid[:self.image_size, :self.image_size]
        mask = (x - cx)**2 + (y - cy)**2 <= radius**2
        
        if pattern == 'solid':
            img[mask] = self.colors['white']
        elif pattern == 'striped':
            img[mask] = self.colors['white']
            # Add stripes
            for i in range(cy - radius, cy + radius, 8):
                if 0 <= i < self.image_size:
                    stripe_mask = mask & (y == i)
                    img[stripe_mask] = self.colors['black']
        elif pattern == 'dotted':
            img[mask] = self.colors['white']
            # Add dots
            for i in range(cx - radius, cx + radius, 12):
                for j in range(cy - radius, cy + radius, 12):
                    if 0 <= i < self.image_size and 0 <= j < self.image_size:
                        if (i - cx)**2 + (j - cy)**2 <= radius**2:
                            if 0 <= i-2 < self.image_size and 0 <= j-2 < self.image_size:
                                img[j-2:j+2, i-2:i+2] = self.colors['black']
    
    def _draw_rectangle(self, img: np.ndarray, cx: int, cy: int, width: int, height: int, pattern: str):
        """Draw rectangle with pattern"""
        x1, x2 = max(0, cx - width//2), min(self.image_size, cx + width//2)
        y1, y2 = max(0, cy - height//2), min(self.image_size, cy + height//2)
        
        if pattern == 'solid':
            img[y1:y2, x1:x2] = self.colors['white']
        elif pattern == 'striped':
            img[y1:y2, x1:x2] = self.colors['white']
            for i in range(y1, y2, 8):
                if i < y2:
                    img[i:min(i+4, y2), x1:x2] = self.colors['black']
        elif pattern == 'checkered':
            img[y1:y2, x1:x2] = self.colors['white']
            for i in range(y1, y2, 16):
                for j in range(x1, x2, 16):
                    if (i//16 + j//16) % 2 == 0:
                        end_i, end_j = min(i+16, y2), min(j+16, x2)
                        img[i:end_i, j:end_j] = self.colors['black']
        elif pattern == 'grid':
            img[y1:y2, x1:x2] = self.colors['white']
            # Vertical lines
            for j in range(x1, x2, 12):
                if j < x2:
                    img[y1:y2, j:min(j+2, x2)] = self.colors['black']
            # Horizontal lines  
            for i in range(y1, y2, 12):
                if i < y2:
                    img[i:min(i+2, y2), x1:x2] = self.colors['black']
    
    def _draw_triangle(self, img: np.ndarray, cx: int, cy: int, size: int, pattern: str):
        """Draw triangle with pattern"""
        # Create triangle vertices
        points = np.array([
            [cx, cy - size//2],  # top
            [cx - size//2, cy + size//2],  # bottom left
            [cx + size//2, cy + size//2]   # bottom right
        ])
        
        # Fill triangle
        y, x = np.ogrid[:self.image_size, :self.image_size]
        
        # Check if points are inside triangle using barycentric coordinates
        def point_in_triangle(px, py, p1, p2, p3):
            denom = (p2[1] - p3[1]) * (p1[0] - p3[0]) + (p3[0] - p2[0]) * (p1[1] - p3[1])
            if abs(denom) < 1e-10:
                return False
            a = ((p2[1] - p3[1]) * (px - p3[0]) + (p3[0] - p2[0]) * (py - p3[1])) / denom
            b = ((p3[1] - p1[1]) * (px - p3[0]) + (p1[0] - p3[0]) * (py - p3[1])) / denom
            c = 1 - a - b
            return 0 <= a <= 1 and 0 <= b <= 1 and 0 <= c <= 1
        
        for i in range(self.image_size):
            for j in range(self.image_size):
                if point_in_triangle(j, i, points[0], points[1], points[2]):
                    if pattern == 'solid':
                        img[i, j] = self.colors['white']
                    elif pattern == 'striped' and i % 8 < 4:
                        img[i, j] = self.colors['white']
                    elif pattern == 'striped' and i % 8 >= 4:
                        img[i, j] = self.colors['black']
                    else:
                        img[i, j] = self.colors['white']
    
    def _draw_oval(self, img: np.ndarray, cx: int, cy: int, width: int, height: int, pattern: str):
        """Draw oval with pattern"""
        y, x = np.ogrid[:self.image_size, :self.image_size]
        mask = ((x - cx)**2 / (width//2)**2) + ((y - cy)**2 / (height//2)**2) <= 1
        
        if pattern == 'solid':
            img[mask] = self.colors['white']
        elif pattern == 'striped':
            img[mask] = self.colors['white']
            for i in range(cy - height//2, cy + height//2, 6):
                if 0 <= i < self.image_size:
                    stripe_mask = mask & (y == i)
                    img[stripe_mask] = self.colors['black']
    
    def apply_composition_instruction(self, ref_images: List[np.ndarray], instruction: str) -> np.ndarray:
        """Apply composition instruction to reference images"""
        target = np.zeros((self.image_size, self.image_size), dtype=np.uint8)
        
        # Parse instruction type and apply accordingly
        if "on top of" in instruction:
            target = self._place_on_top(ref_images[0], ref_images[1])
        elif "to the left of" in instruction:
            target = self._place_to_left(ref_images[0], ref_images[1])
        elif "to the right of" in instruction:
            target = self._place_to_right(ref_images[0], ref_images[1])
        elif "below" in instruction:
            target = self._place_below(ref_images[0], ref_images[1])
        elif "inside" in instruction and "center" in instruction:
            target = self._place_inside(ref_images[0], ref_images[1])
        elif "side by side" in instruction:
            target = self._place_side_by_side(ref_images[0], ref_images[1])
        elif "stack" in instruction and "vertically" in instruction:
            target = self._stack_vertically(ref_images[0], ref_images[1])
        elif "overlap" in instruction:
            target = self._overlap_shapes(ref_images[0], ref_images[1])
        elif "larger than" in instruction:
            target = self._resize_and_place(ref_images[0], ref_images[1], scale_first=1.5)
        elif "smaller than" in instruction:
            target = self._resize_and_place(ref_images[0], ref_images[1], scale_first=0.7)
        elif "triangle formation" in instruction and len(ref_images) >= 3:
            target = self._triangle_formation(ref_images[0], ref_images[1], ref_images[2])
        elif "in a line" in instruction and len(ref_images) >= 3:
            target = self._line_formation(ref_images[0], ref_images[1], ref_images[2])
        elif "combine" in instruction or "merge" in instruction:
            target = self._merge_shapes(ref_images[0], ref_images[1])
        else:
            # Default: simple combination
            target = self._simple_combination(ref_images[0], ref_images[1])
        
        return target
    
    def _place_on_top(self, img1: np.ndarray, img2: np.ndarray) -> np.ndarray:
        """Place first shape on top of second shape"""
        result = np.zeros_like(img1)
        
        # Find centers of shapes
        center1 = self._find_shape_center(img1)
        center2 = self._find_shape_center(img2)
        
        if center1 and center2:
            # Place img2 at bottom
            offset_y = 60
            new_center2 = (center2[0], min(self.image_size - 50, center2[1] + offset_y))
            result = self._place_shape_at_position(result, img2, center2, new_center2)
            
            # Place img1 on top
            new_center1 = (center1[0], max(50, new_center2[1] - offset_y))
            result = self._place_shape_at_position(result, img1, center1, new_center1)
        
        return result
    
    def _place_to_left(self, img1: np.ndarray, img2: np.ndarray) -> np.ndarray:
        """Place first shape to the left of second shape"""
        result = np.zeros_like(img1)
        
        center1 = self._find_shape_center(img1)
        center2 = self._find_shape_center(img2)
        
        if center1 and center2:
            # Place img2 on the right
            new_center2 = (min(self.image_size - 50, self.image_size // 2 + 40), center2[1])
            result = self._place_shape_at_position(result, img2, center2, new_center2)
            
            # Place img1 on the left
            new_center1 = (max(50, self.image_size // 2 - 40), center1[1])
            result = self._place_shape_at_position(result, img1, center1, new_center1)
        
        return result
    
    def _place_to_right(self, img1: np.ndarray, img2: np.ndarray) -> np.ndarray:
        """Place first shape to the right of second shape"""
        result = np.zeros_like(img1)
        
        center1 = self._find_shape_center(img1)
        center2 = self._find_shape_center(img2)
        
        if center1 and center2:
            # Place img2 on the left
            new_center2 = (max(50, self.image_size // 2 - 40), center2[1])
            result = self._place_shape_at_position(result, img2, center2, new_center2)
            
            # Place img1 on the right
            new_center1 = (min(self.image_size - 50, self.image_size // 2 + 40), center1[1])
            result = self._place_shape_at_position(result, img1, center1, new_center1)
        
        return result
    
    def _place_below(self, img1: np.ndarray, img2: np.ndarray) -> np.ndarray:
        """Place first shape below second shape"""
        result = np.zeros_like(img1)
        
        center1 = self._find_shape_center(img1)
        center2 = self._find_shape_center(img2)
        
        if center1 and center2:
            # Place img2 at top
            offset_y = 60
            new_center2 = (center2[0], max(50, center2[1] - offset_y))
            result = self._place_shape_at_position(result, img2, center2, new_center2)
            
            # Place img1 below
            new_center1 = (center1[0], min(self.image_size - 50, new_center2[1] + offset_y))
            result = self._place_shape_at_position(result, img1, center1, new_center1)
        
        return result
    
    def _place_inside(self, img1: np.ndarray, img2: np.ndarray) -> np.ndarray:
        """Place first shape inside second shape (centered)"""
        result = np.zeros_like(img1)
        
        # First place the outer shape (img2)
        center2 = self._find_shape_center(img2)
        if center2:
            result = self._place_shape_at_position(result, img2, center2, (self.image_size//2, self.image_size//2))
        
        # Then place inner shape (img1) at center, but smaller
        center1 = self._find_shape_center(img1)
        if center1:
            # Scale down img1 
            scaled_img1 = self._scale_shape(img1, 0.6)
            result = self._place_shape_at_position(result, scaled_img1, center1, (self.image_size//2, self.image_size//2))
        
        return result
    
    def _place_side_by_side(self, img1: np.ndarray, img2: np.ndarray) -> np.ndarray:
        """Place shapes side by side"""
        return self._place_to_left(img1, img2)
    
    def _stack_vertically(self, img1: np.ndarray, img2: np.ndarray) -> np.ndarray:
        """Stack shapes vertically"""
        return self._place_on_top(img1, img2)
    
    def _overlap_shapes(self, img1: np.ndarray, img2: np.ndarray) -> np.ndarray:
        """Overlap two shapes partially"""
        result = np.zeros_like(img1)
        
        center1 = self._find_shape_center(img1)
        center2 = self._find_shape_center(img2)
        
        if center1 and center2:
            # Place first shape
            new_center1 = (self.image_size//2 - 20, self.image_size//2)
            result = self._place_shape_at_position(result, img1, center1, new_center1)
            
            # Place second shape overlapping
            new_center2 = (self.image_size//2 + 20, self.image_size//2)
            temp_result = np.zeros_like(img2)
            temp_result = self._place_shape_at_position(temp_result, img2, center2, new_center2)
            
            # Combine with overlap
            result = np.maximum(result, temp_result)
        
        return result
    
    def _resize_and_place(self, img1: np.ndarray, img2: np.ndarray, scale_first: float = 1.0) -> np.ndarray:
        """Resize first image and place with second"""
        result = np.zeros_like(img1)
        
        # Scale first image
        scaled_img1 = self._scale_shape(img1, scale_first)
        
        # Place side by side
        center1 = self._find_shape_center(scaled_img1)
        center2 = self._find_shape_center(img2)
        
        if center1 and center2:
            new_center1 = (self.image_size//2 - 40, self.image_size//2)
            new_center2 = (self.image_size//2 + 40, self.image_size//2)
            
            result = self._place_shape_at_position(result, scaled_img1, center1, new_center1)
            result = self._place_shape_at_position(result, img2, center2, new_center2)
        
        return result
    
    def _triangle_formation(self, img1: np.ndarray, img2: np.ndarray, img3: np.ndarray) -> np.ndarray:
        """Arrange three shapes in triangle formation"""
        result = np.zeros_like(img1)
        
        center1 = self._find_shape_center(img1)
        center2 = self._find_shape_center(img2)
        center3 = self._find_shape_center(img3)
        
        if all([center1, center2, center3]):
            # Triangle positions
            pos1 = (self.image_size//2, self.image_size//2 - 40)  # top
            pos2 = (self.image_size//2 - 40, self.image_size//2 + 40)  # bottom left
            pos3 = (self.image_size//2 + 40, self.image_size//2 + 40)  # bottom right
            
            result = self._place_shape_at_position(result, img1, center1, pos1)
            result = self._place_shape_at_position(result, img2, center2, pos2)
            result = self._place_shape_at_position(result, img3, center3, pos3)
        
        return result
    
    def _line_formation(self, img1: np.ndarray, img2: np.ndarray, img3: np.ndarray) -> np.ndarray:
        """Arrange three shapes in a line"""
        result = np.zeros_like(img1)
        
        center1 = self._find_shape_center(img1)
        center2 = self._find_shape_center(img2)
        center3 = self._find_shape_center(img3)
        
        if all([center1, center2, center3]):
            # Line positions
            pos1 = (self.image_size//2 - 60, self.image_size//2)
            pos2 = (self.image_size//2, self.image_size//2)
            pos3 = (self.image_size//2 + 60, self.image_size//2)
            
            result = self._place_shape_at_position(result, img1, center1, pos1)
            result = self._place_shape_at_position(result, img2, center2, pos2)
            result = self._place_shape_at_position(result, img3, center3, pos3)
        
        return result
    
    def _merge_shapes(self, img1: np.ndarray, img2: np.ndarray) -> np.ndarray:
        """Merge two shapes into one"""
        result = np.zeros_like(img1)
        
        center1 = self._find_shape_center(img1)
        center2 = self._find_shape_center(img2)
        
        if center1 and center2:
            # Place both at center with slight offset
            pos1 = (self.image_size//2 - 10, self.image_size//2)
            pos2 = (self.image_size//2 + 10, self.image_size//2)
            
            result = self._place_shape_at_position(result, img1, center1, pos1)
            temp_result = np.zeros_like(img2)
            temp_result = self._place_shape_at_position(temp_result, img2, center2, pos2)
            
            # Merge by taking maximum
            result = np.maximum(result, temp_result)
        
        return result
    
    def _simple_combination(self, img1: np.ndarray, img2: np.ndarray) -> np.ndarray:
        """Simple side-by-side combination"""
        return self._place_side_by_side(img1, img2)
    
    def _find_shape_center(self, img: np.ndarray) -> Tuple[int, int]:
        """Find center of shape in image"""
        white_pixels = np.where(img > 0)
        if len(white_pixels[0]) > 0:
            center_y = int(np.mean(white_pixels[0]))
            center_x = int(np.mean(white_pixels[1]))
            return (center_x, center_y)
        return None
    
    def _place_shape_at_position(self, target: np.ndarray, source: np.ndarray, 
                                old_center: Tuple[int, int], new_center: Tuple[int, int]) -> np.ndarray:
        """Place shape at new position"""
        if old_center is None:
            return target
        
        dx = new_center[0] - old_center[0]
        dy = new_center[1] - old_center[1]
        
        # Create shifted version
        white_pixels = np.where(source > 0)
        for y, x in zip(white_pixels[0], white_pixels[1]):
            new_x = x + dx
            new_y = y + dy
            if 0 <= new_x < self.image_size and 0 <= new_y < self.image_size:
                target[new_y, new_x] = source[y, x]
        
        return target
    
    def _scale_shape(self, img: np.ndarray, scale: float) -> np.ndarray:
        """Scale shape size"""
        if scale == 1.0:
            return img.copy()
        
        center = self._find_shape_center(img)
        if center is None:
            return img.copy()
        
        result = np.zeros_like(img)
        white_pixels = np.where(img > 0)
        
        for y, x in zip(white_pixels[0], white_pixels[1]):
            # Scale relative to center
            dx = x - center[0]
            dy = y - center[1]
            
            new_x = int(center[0] + dx * scale)
            new_y = int(center[1] + dy * scale)
            
            if 0 <= new_x < self.image_size and 0 <= new_y < self.image_size:
                result[new_y, new_x] = img[y, x]
        
        return result
    
    def create_shape_with_distractors(self, main_shape_type: str, main_size: int, main_pattern: str = 'solid', add_distractors: bool = True) -> Tuple[np.ndarray, str]:
        """Create a shape with optional distractor objects in the same image"""
        img = np.zeros((self.image_size, self.image_size), dtype=np.uint8)
        
        # Create main shape
        margin = main_size // 2 + 20
        center_x = random.randint(margin, self.image_size - margin)
        center_y = random.randint(margin, self.image_size - margin)
        
        if main_shape_type == 'circle':
            self._draw_circle(img, center_x, center_y, main_size // 2, main_pattern)
        elif main_shape_type == 'rectangle':
            self._draw_rectangle(img, center_x, center_y, main_size, main_size // 2, main_pattern)
        elif main_shape_type == 'square':
            self._draw_rectangle(img, center_x, center_y, main_size, main_size, main_pattern)
        elif main_shape_type == 'triangle':
            self._draw_triangle(img, center_x, center_y, main_size, main_pattern)
        elif main_shape_type == 'oval':
            self._draw_oval(img, center_x, center_y, main_size, main_size // 2, main_pattern)
        
        main_description = f"{self.pattern_names[main_pattern]} {self.shape_names[main_shape_type]}"
        
        # Add distractors (30% chance, 1-2 small objects)
        if add_distractors and random.random() < 0.3:
            num_distractors = random.randint(1, 2)
            distractor_names = []
            
            for _ in range(num_distractors):
                # Create smaller distractor
                dist_shape = random.choice(list(self.shape_names.keys()))
                dist_pattern = random.choice(list(self.pattern_names.keys()))
                dist_size = random.randint(20, 35)  # Smaller than main shape
                
                # Find position that doesn't overlap with main shape
                attempts = 0
                while attempts < 20:
                    dist_margin = dist_size // 2 + 10
                    dist_x = random.randint(dist_margin, self.image_size - dist_margin)
                    dist_y = random.randint(dist_margin, self.image_size - dist_margin)
                    
                    # Check if far enough from main shape
                    distance = ((dist_x - center_x) ** 2 + (dist_y - center_y) ** 2) ** 0.5
                    if distance > (main_size // 2 + dist_size // 2 + 20):
                        break
                    attempts += 1
                
                # Draw distractor with lower intensity to distinguish from main shape
                temp_img = np.zeros_like(img)
                if dist_shape == 'circle':
                    self._draw_circle(temp_img, dist_x, dist_y, dist_size // 2, dist_pattern)
                elif dist_shape == 'rectangle':
                    self._draw_rectangle(temp_img, dist_x, dist_y, dist_size, dist_size // 2, dist_pattern)
                elif dist_shape == 'square':
                    self._draw_rectangle(temp_img, dist_x, dist_y, dist_size, dist_size, dist_pattern)
                elif dist_shape == 'triangle':
                    self._draw_triangle(temp_img, dist_x, dist_y, dist_size, dist_pattern)
                elif dist_shape == 'oval':
                    self._draw_oval(temp_img, dist_x, dist_y, dist_size, dist_size // 2, dist_pattern)
                
                # Add distractor with reduced intensity (gray instead of white)
                distractor_mask = temp_img > 0
                img[distractor_mask] = self.colors['gray']  # Gray for distractors
                
                distractor_names.append(f"small {self.pattern_names[dist_pattern]} {self.shape_names[dist_shape]}")
            
            if distractor_names:
                return img, f"{main_description} (with {', '.join(distractor_names)} to ignore)"
        
        return img, main_description

    def generate_sample(self) -> Tuple[List[np.ndarray], np.ndarray, str]:
        """Generate a single training sample with improved reference usage"""
        
        # Choose composition strategy with more variety
        composition_types = [
            "two_object",           # Use 2 objects, ignore 2
            "three_object",         # Use 3 objects, ignore 1
            "four_object",          # Use all 4 objects
            "two_plus_background",  # Use 2 main + 1 background, ignore 1
            "selective_complex"     # Complex selection with within-image distractors
        ]
        
        composition_type = random.choice(composition_types)
        
        # Generate 4 reference images with varying distractor complexity
        ref_images = []
        shape_descriptions = []
        
        for i in range(4):
            shape_type = random.choice(list(self.shape_names.keys()))
            pattern = random.choice(list(self.pattern_names.keys()))
            size = random.randint(45, 85)
            
            # Add distractors to some references (more likely for complex compositions)
            add_distractors = composition_type in ["selective_complex", "four_object"] or random.random() < 0.4
            
            ref_img, shape_description = self.create_shape_with_distractors(
                shape_type, size, pattern, add_distractors
            )
            
            ref_images.append(ref_img)
            shape_descriptions.append(shape_description)
        
        # Generate instruction and target based on composition type
        if composition_type == "two_object":
            return self._generate_two_object_composition(ref_images, shape_descriptions)
            
        elif composition_type == "three_object":
            return self._generate_three_object_composition(ref_images, shape_descriptions)
            
        elif composition_type == "four_object":
            return self._generate_four_object_composition(ref_images, shape_descriptions)
            
        elif composition_type == "two_plus_background":
            return self._generate_background_composition(ref_images, shape_descriptions)
            
        else:  # selective_complex
            return self._generate_selective_complex_composition(ref_images, shape_descriptions)
    
    def _generate_two_object_composition(self, ref_images: List[np.ndarray], shape_descriptions: List[str]) -> Tuple[List[np.ndarray], np.ndarray, str]:
        """Generate 2-object composition with 2 ignored references"""
        used_indices = random.sample(range(4), 2)
        ignored_indices = [i for i in range(4) if i not in used_indices]
        
        main_obj = shape_descriptions[used_indices[0]].split(' (')[0]  # Remove distractor info for instruction
        second_obj = shape_descriptions[used_indices[1]].split(' (')[0]
        ignored_obj1 = shape_descriptions[ignored_indices[0]].split(' (')[0]
        ignored_obj2 = shape_descriptions[ignored_indices[1]].split(' (')[0]
        
        base_instructions = [
            f"Place the {main_obj} on top of the {second_obj}, ignore the {ignored_obj1} and {ignored_obj2}",
            f"Position the {main_obj} to the left of the {second_obj}, do not use the {ignored_obj1} or {ignored_obj2}",
            f"Put the {main_obj} below the {second_obj}, ignore other shapes",
            f"Arrange the {main_obj} and {second_obj} side by side, exclude the {ignored_obj1} and {ignored_obj2}",
            f"Overlap the {main_obj} with the {second_obj}, ignore the remaining shapes"
        ]
        
        instruction = random.choice(base_instructions)
        target_img = self._compose_two_objects(
            ref_images[used_indices[0]], 
            ref_images[used_indices[1]], 
            instruction
        )
        
        return ref_images, target_img, instruction
    
    def _generate_three_object_composition(self, ref_images: List[np.ndarray], shape_descriptions: List[str]) -> Tuple[List[np.ndarray], np.ndarray, str]:
        """Generate 3-object composition with 1 ignored reference"""
        used_indices = random.sample(range(4), 3)
        ignored_index = [i for i in range(4) if i not in used_indices][0]
        
        obj1 = shape_descriptions[used_indices[0]].split(' (')[0]
        obj2 = shape_descriptions[used_indices[1]].split(' (')[0]
        obj3 = shape_descriptions[used_indices[2]].split(' (')[0]
        ignored_obj = shape_descriptions[ignored_index].split(' (')[0]
        
        base_instructions = [
            f"Arrange the {obj1}, {obj2}, and {obj3} in a triangle formation, ignore the {ignored_obj}",
            f"Place the {obj1}, {obj2}, and {obj3} in a horizontal line, do not use the {ignored_obj}",
            f"Put the {obj1} in center with {obj2} and {obj3} around it, exclude the {ignored_obj}",
            f"Stack the {obj1} on {obj2} and place {obj3} beside them, ignore the {ignored_obj}",
            f"Create a cluster with {obj1}, {obj2}, and {obj3}, do not include the {ignored_obj}"
        ]
        
        instruction = random.choice(base_instructions)
        target_img = self._compose_three_objects(
            ref_images[used_indices[0]], 
            ref_images[used_indices[1]], 
            ref_images[used_indices[2]], 
            instruction
        )
        
        return ref_images, target_img, instruction
    
    def _generate_four_object_composition(self, ref_images: List[np.ndarray], shape_descriptions: List[str]) -> Tuple[List[np.ndarray], np.ndarray, str]:
        """Generate composition using all 4 references"""
        obj1 = shape_descriptions[0].split(' (')[0]
        obj2 = shape_descriptions[1].split(' (')[0]
        obj3 = shape_descriptions[2].split(' (')[0]
        obj4 = shape_descriptions[3].split(' (')[0]
        
        base_instructions = [
            f"Arrange all four objects: {obj1}, {obj2}, {obj3}, and {obj4} in a 2x2 grid formation",
            f"Create a diamond pattern with {obj1} at top, {obj2} at bottom, {obj3} at left, and {obj4} at right",
            f"Place {obj1} and {obj2} in the center with {obj3} and {obj4} surrounding them",
            f"Form a line with {obj1}, {obj2}, {obj3}, and {obj4} from left to right",
            f"Stack {obj1} on {obj2} vertically and {obj3} on {obj4} vertically, then place both stacks side by side",
            f"Create a cross formation with {obj1} at center and {obj2}, {obj3}, {obj4} at the arms"
        ]
        
        instruction = random.choice(base_instructions)
        target_img = self._compose_four_objects(ref_images, instruction)
        
        return ref_images, target_img, instruction
    
    def _generate_background_composition(self, ref_images: List[np.ndarray], shape_descriptions: List[str]) -> Tuple[List[np.ndarray], np.ndarray, str]:
        """Generate composition with background element"""
        used_indices = random.sample(range(4), 3)
        ignored_index = [i for i in range(4) if i not in used_indices][0]
        
        main_indices = used_indices[:2]
        bg_index = used_indices[2]
        
        main_obj1 = shape_descriptions[main_indices[0]].split(' (')[0]
        main_obj2 = shape_descriptions[main_indices[1]].split(' (')[0]
        bg_obj = shape_descriptions[bg_index].split(' (')[0]
        ignored_obj = shape_descriptions[ignored_index].split(' (')[0]
        
        base_instructions = [
            f"Place the {main_obj1} and {main_obj2} on top of the {bg_obj} background, ignore the {ignored_obj}",
            f"Use the {bg_obj} as base and arrange {main_obj1} and {main_obj2} on it, exclude the {ignored_obj}",
            f"Layer {main_obj1} and {main_obj2} over the {bg_obj} backdrop, do not use the {ignored_obj}"
        ]
        
        instruction = random.choice(base_instructions)
        target_img = self._compose_with_background(
            ref_images[main_indices[0]], 
            ref_images[main_indices[1]], 
            ref_images[bg_index], 
            instruction
        )
        
        return ref_images, target_img, instruction
    
    def _generate_selective_complex_composition(self, ref_images: List[np.ndarray], shape_descriptions: List[str]) -> Tuple[List[np.ndarray], np.ndarray, str]:
        """Generate complex composition with explicit distractor handling"""
        num_to_use = random.choice([2, 3])
        used_indices = random.sample(range(4), num_to_use)
        ignored_indices = [i for i in range(4) if i not in used_indices]
        
        used_objects = [shape_descriptions[i].split(' (')[0] for i in used_indices]
        ignored_objects = [shape_descriptions[i].split(' (')[0] for i in ignored_indices]
        
        # Create instruction that explicitly mentions within-image distractors
        if num_to_use == 2:
            instruction = f"Focus only on the main {used_objects[0]} and {used_objects[1]}, arrange them side by side. Ignore all small distractor objects and the {', '.join(ignored_objects)}."
            target_img = self._compose_two_objects(
                ref_images[used_indices[0]], 
                ref_images[used_indices[1]], 
                instruction
            )
        else:  # num_to_use == 3
            instruction = f"Use only the primary {used_objects[0]}, {used_objects[1]}, and {used_objects[2]} in a triangle formation. Ignore any small objects within the images and exclude the {', '.join(ignored_objects)}."
            target_img = self._compose_three_objects(
                ref_images[used_indices[0]], 
                ref_images[used_indices[1]], 
                ref_images[used_indices[2]], 
                instruction
            )
        
        return ref_images, target_img, instruction
    
    def _compose_four_objects(self, ref_images: List[np.ndarray], instruction: str) -> np.ndarray:
        """Compose all four objects based on instruction"""
        result = np.zeros_like(ref_images[0])
        
        centers = [self._find_shape_center(img) for img in ref_images]
        if not all(centers):
            return result
        
        if "2x2 grid" in instruction:
            positions = [
                (self.image_size//2 - 40, self.image_size//2 - 40),  # top-left
                (self.image_size//2 + 40, self.image_size//2 - 40),  # top-right
                (self.image_size//2 - 40, self.image_size//2 + 40),  # bottom-left
                (self.image_size//2 + 40, self.image_size//2 + 40)   # bottom-right
            ]
        elif "diamond" in instruction:
            positions = [
                (self.image_size//2, self.image_size//2 - 50),      # top
                (self.image_size//2, self.image_size//2 + 50),      # bottom
                (self.image_size//2 - 50, self.image_size//2),      # left
                (self.image_size//2 + 50, self.image_size//2)       # right
            ]
        elif "cross formation" in instruction:
            positions = [
                (self.image_size//2, self.image_size//2),           # center
                (self.image_size//2, self.image_size//2 - 60),      # top
                (self.image_size//2 - 60, self.image_size//2),      # left
                (self.image_size//2 + 60, self.image_size//2)       # right
            ]
        elif "line" in instruction:
            positions = [
                (self.image_size//2 - 75, self.image_size//2),
                (self.image_size//2 - 25, self.image_size//2),
                (self.image_size//2 + 25, self.image_size//2),
                (self.image_size//2 + 75, self.image_size//2)
            ]
        else:  # default: 2x2 grid
            positions = [
                (self.image_size//2 - 40, self.image_size//2 - 40),
                (self.image_size//2 + 40, self.image_size//2 - 40),
                (self.image_size//2 - 40, self.image_size//2 + 40),
                (self.image_size//2 + 40, self.image_size//2 + 40)
            ]
        
        # Scale objects smaller for 4-object compositions
        for i, (ref_img, center, pos) in enumerate(zip(ref_images, centers, positions)):
            scaled_img = self._scale_shape(ref_img, 0.6)
            result = self._place_shape_at_position(result, scaled_img, center, pos)
        
        return result
    
    def _compose_two_objects(self, img1: np.ndarray, img2: np.ndarray, instruction: str) -> np.ndarray:
        """Compose two objects based on instruction"""
        result = np.zeros_like(img1)
        
        if "on top of" in instruction:
            result = self._place_on_top(img1, img2)
        elif "to the left of" in instruction:
            result = self._place_to_left(img1, img2)
        elif "below" in instruction:
            result = self._place_below(img1, img2)
        elif "side by side" in instruction:
            result = self._place_side_by_side(img1, img2)
        elif "overlap" in instruction:
            result = self._overlap_shapes(img1, img2)
        elif "inside" in instruction:
            result = self._place_inside(img1, img2)
        else:
            # Default: side by side
            result = self._place_side_by_side(img1, img2)
        
        return result
    
    def _compose_three_objects(self, img1: np.ndarray, img2: np.ndarray, img3: np.ndarray, instruction: str) -> np.ndarray:
        """Compose three objects based on instruction"""
        result = np.zeros_like(img1)
        
        if "triangle formation" in instruction:
            result = self._triangle_formation(img1, img2, img3)
        elif "horizontal line" in instruction or "in a line" in instruction:
            result = self._line_formation(img1, img2, img3)
        elif "center" in instruction and "around" in instruction:
            result = self._center_with_surrounding(img1, img2, img3)
        elif "stack" in instruction and "beside" in instruction:
            result = self._stack_and_beside(img1, img2, img3)
        elif "cluster" in instruction:
            result = self._create_cluster(img1, img2, img3)
        else:
            # Default: triangle formation
            result = self._triangle_formation(img1, img2, img3)
        
        return result
    
    def _compose_with_background(self, main1: np.ndarray, main2: np.ndarray, background: np.ndarray, instruction: str) -> np.ndarray:
        """Compose with background object"""
        result = np.zeros_like(main1)
        
        # First place background (scaled and centered)
        bg_center = self._find_shape_center(background)
        if bg_center:
            # Scale background to be larger
            scaled_bg = self._scale_shape(background, 1.3)
            result = self._place_shape_at_position(result, scaled_bg, bg_center, (self.image_size//2, self.image_size//2))
        
        # Then place main objects on top
        center1 = self._find_shape_center(main1)
        center2 = self._find_shape_center(main2)
        
        if center1 and center2:
            # Position main objects relative to center
            pos1 = (self.image_size//2 - 30, self.image_size//2 - 20)
            pos2 = (self.image_size//2 + 30, self.image_size//2 + 20)
            
            # Scale main objects to be smaller
            scaled_main1 = self._scale_shape(main1, 0.8)
            scaled_main2 = self._scale_shape(main2, 0.8)
            
            result = self._place_shape_at_position(result, scaled_main1, center1, pos1)
            result = self._place_shape_at_position(result, scaled_main2, center2, pos2)
        
        return result
    
    def _center_with_surrounding(self, center_img: np.ndarray, surround1: np.ndarray, surround2: np.ndarray) -> np.ndarray:
        """Place center object with two surrounding objects"""
        result = np.zeros_like(center_img)
        
        center_pos = (self.image_size//2, self.image_size//2)
        surround_pos1 = (self.image_size//2 - 60, self.image_size//2 - 40)
        surround_pos2 = (self.image_size//2 + 60, self.image_size//2 + 40)
        
        center1 = self._find_shape_center(center_img)
        center2 = self._find_shape_center(surround1)
        center3 = self._find_shape_center(surround2)
        
        if all([center1, center2, center3]):
            result = self._place_shape_at_position(result, center_img, center1, center_pos)
            result = self._place_shape_at_position(result, surround1, center2, surround_pos1)
            result = self._place_shape_at_position(result, surround2, center3, surround_pos2)
        
        return result
    
    def _stack_and_beside(self, stack1: np.ndarray, stack2: np.ndarray, beside: np.ndarray) -> np.ndarray:
        """Stack two objects and place third beside the stack"""
        result = np.zeros_like(stack1)
        
        # Stack positions (left side)
        stack_pos1 = (self.image_size//2 - 50, self.image_size//2 - 30)
        stack_pos2 = (self.image_size//2 - 50, self.image_size//2 + 30)
        
        # Beside position (right side)
        beside_pos = (self.image_size//2 + 50, self.image_size//2)
        
        center1 = self._find_shape_center(stack1)
        center2 = self._find_shape_center(stack2)
        center3 = self._find_shape_center(beside)
        
        if all([center1, center2, center3]):
            result = self._place_shape_at_position(result, stack1, center1, stack_pos1)
            result = self._place_shape_at_position(result, stack2, center2, stack_pos2)
            result = self._place_shape_at_position(result, beside, center3, beside_pos)
        
        return result
    
    def _create_cluster(self, img1: np.ndarray, img2: np.ndarray, img3: np.ndarray) -> np.ndarray:
        """Create a tight cluster of three objects"""
        result = np.zeros_like(img1)
        
        # Cluster positions (close together)
        pos1 = (self.image_size//2 - 25, self.image_size//2 - 25)
        pos2 = (self.image_size//2 + 25, self.image_size//2 - 25)
        pos3 = (self.image_size//2, self.image_size//2 + 30)
        
        center1 = self._find_shape_center(img1)
        center2 = self._find_shape_center(img2)
        center3 = self._find_shape_center(img3)
        
        if all([center1, center2, center3]):
            # Scale objects to be smaller for clustering
            scaled1 = self._scale_shape(img1, 0.7)
            scaled2 = self._scale_shape(img2, 0.7)
            scaled3 = self._scale_shape(img3, 0.7)
            
            result = self._place_shape_at_position(result, scaled1, center1, pos1)
            result = self._place_shape_at_position(result, scaled2, center2, pos2)
            result = self._place_shape_at_position(result, scaled3, center3, pos3)
        
        return result
    
    def generate_dataset(self, num_samples: int = 10000, output_dir: str = "composition_dataset_10k", batch_size: int = 500):
        """Generate complete dataset in batches to avoid memory issues"""
        
        print("="*80)
        print("GENERATING INSTRUCTPIX2PIX COMPOSITION DATASET")
        print("="*80)
        print(f"Number of samples: {num_samples}")
        print(f"Output directory: {output_dir}")
        print(f"Image size: {self.image_size}x{self.image_size}")
        print(f"Batch size: {batch_size}")
        print("="*80)
        
        # Create output directory
        try:
            os.makedirs(output_dir, exist_ok=True)
            print(f"✓ Created output directory: {output_dir}")
        except Exception as e:
            print(f"✗ Failed to create output directory: {e}")
            return
        
        # Calculate number of batches
        num_batches = (num_samples + batch_size - 1) // batch_size
        print(f"Will generate {num_batches} batches")
        print("🔄 Saving each batch immediately to avoid data loss")
        
        # Initialize batch files directory
        batch_dir = os.path.join(output_dir, "batches")
        os.makedirs(batch_dir, exist_ok=True)
        
        # Collect all instructions
        all_instructions = []
        
        # Generate and save each batch immediately
        for batch_idx in range(num_batches):
            start_idx = batch_idx * batch_size
            end_idx = min(start_idx + batch_size, num_samples)
            current_batch_size = end_idx - start_idx
            
            print(f"\nProcessing batch {batch_idx + 1}/{num_batches} (samples {start_idx}-{end_idx-1})")
            
            # Storage for current batch
            batch_ref_images = []
            batch_target_images = []
            batch_instructions = []
            
            # Generate current batch
            try:
                for i in tqdm(range(current_batch_size), desc=f"Batch {batch_idx + 1}", leave=False):
                    ref_images, target_img, instruction = self.generate_sample()
                    
                    batch_ref_images.append(np.stack(ref_images))  # Shape: [4, H, W]
                    batch_target_images.append(target_img)  # Shape: [H, W]
                    batch_instructions.append(instruction)
                
                print(f"✓ Generated {current_batch_size} samples for batch {batch_idx + 1}")
                
            except Exception as e:
                print(f"✗ Error generating batch {batch_idx + 1}: {e}")
                continue
            
            # Convert batch to numpy arrays
            try:
                batch_ref_array = np.stack(batch_ref_images)  # Shape: [batch_size, 4, H, W]
                batch_target_array = np.stack(batch_target_images)  # Shape: [batch_size, H, W]
                
                print(f"✓ Converted batch {batch_idx + 1} to numpy arrays: ref {batch_ref_array.shape}, target {batch_target_array.shape}")
                
            except Exception as e:
                print(f"✗ Error converting batch {batch_idx + 1} to numpy: {e}")
                continue
            
            # Save batch immediately
            try:
                batch_ref_file = os.path.join(batch_dir, f"ref_batch_{batch_idx:03d}.npz")
                batch_target_file = os.path.join(batch_dir, f"target_batch_{batch_idx:03d}.npz")
                batch_inst_file = os.path.join(batch_dir, f"instructions_batch_{batch_idx:03d}.txt")
                
                # Save reference images for this batch
                np.savez_compressed(batch_ref_file, data=batch_ref_array)
                print(f"✓ Saved batch {batch_idx + 1} references: {os.path.getsize(batch_ref_file)/(1024*1024):.1f} MB")
                
                # Save target images for this batch
                np.savez_compressed(batch_target_file, data=batch_target_array)
                print(f"✓ Saved batch {batch_idx + 1} targets: {os.path.getsize(batch_target_file)/(1024*1024):.1f} MB")
                
                # Save instructions for this batch
                with open(batch_inst_file, 'w') as f:
                    for inst in batch_instructions:
                        f.write(inst + '\n')
                print(f"✓ Saved batch {batch_idx + 1} instructions: {len(batch_instructions)} items")
                
                # Store instructions for final file
                all_instructions.extend(batch_instructions)
                
            except Exception as e:
                print(f"✗ Error saving batch {batch_idx + 1}: {e}")
                continue
            
            # Clear memory
            try:
                del batch_ref_images, batch_target_images, batch_instructions
                del batch_ref_array, batch_target_array
                print(f"✓ Batch {batch_idx + 1} saved and memory cleared.")
                
            except Exception as e:
                print(f"⚠ Warning: Error clearing memory for batch {batch_idx + 1}: {e}")
        
        # Now combine all saved batches into final files
        print(f"\n🔄 Combining {num_batches} saved batches into final files...")
        
        try:
            # Load and combine all reference batches
            print("Combining reference images...")
            ref_arrays = []
            for batch_idx in range(num_batches):
                batch_ref_file = os.path.join(batch_dir, f"ref_batch_{batch_idx:03d}.npz")
                if os.path.exists(batch_ref_file):
                    batch_data = np.load(batch_ref_file)['data']
                    ref_arrays.append(batch_data)
                    print(f"  ✓ Loaded batch {batch_idx + 1}: {batch_data.shape}")
                else:
                    print(f"  ✗ Missing batch {batch_idx + 1}")
            
            if ref_arrays:
                final_ref_array = np.concatenate(ref_arrays, axis=0)
                ref_file = os.path.join(output_dir, "reference_scenes.npz")
                np.savez_compressed(ref_file, data=final_ref_array)
                print(f"✓ Final reference file: {final_ref_array.shape} -> {os.path.getsize(ref_file)/(1024*1024):.1f} MB")
                del ref_arrays, final_ref_array
            
            # Load and combine all target batches
            print("Combining target images...")
            target_arrays = []
            for batch_idx in range(num_batches):
                batch_target_file = os.path.join(batch_dir, f"target_batch_{batch_idx:03d}.npz")
                if os.path.exists(batch_target_file):
                    batch_data = np.load(batch_target_file)['data']
                    target_arrays.append(batch_data)
                    print(f"  ✓ Loaded batch {batch_idx + 1}: {batch_data.shape}")
                else:
                    print(f"  ✗ Missing batch {batch_idx + 1}")
            
            if target_arrays:
                final_target_array = np.concatenate(target_arrays, axis=0)
                target_file = os.path.join(output_dir, "target_sketches.npz")
                np.savez_compressed(target_file, data=final_target_array)
                print(f"✓ Final target file: {final_target_array.shape} -> {os.path.getsize(target_file)/(1024*1024):.1f} MB")
                
                # Keep first 5 for visualization
                visualization_targets = final_target_array[:5].copy()
                del target_arrays, final_target_array
            
        except Exception as e:
            print(f"✗ Error combining batches: {e}")
            return
        
        # Save instructions
        print("Saving instructions...")
        with open(inst_file, 'w') as f:
            for instruction in all_instructions:
                f.write(instruction + '\n')
        
        print(f"Number of instructions: {len(all_instructions)}")
        
        # Create train/val/test splits
        print("Creating data splits...")
        indices = list(range(num_samples))
        random.shuffle(indices)
        
        # 70% train, 15% val, 15% test
        train_size = int(0.7 * num_samples)
        val_size = int(0.15 * num_samples)
        
        train_indices = indices[:train_size]
        val_indices = indices[train_size:train_size + val_size]
        test_indices = indices[train_size + val_size:]
        
        # Save splits
        for split_name, split_indices in [("train", train_indices), ("val", val_indices), ("test", test_indices)]:
            split_dir = os.path.join(output_dir, split_name)
            os.makedirs(split_dir, exist_ok=True)
            
            with open(os.path.join(split_dir, "indices.txt"), 'w') as f:
                for idx in split_indices:
                    f.write(f"{idx}\n")
            
            print(f"{split_name.upper()} split: {len(split_indices)} samples")
        
        # Generate sample visualizations (using saved data to avoid memory issues)
        print("Creating sample visualizations...")
        self.create_sample_visualizations_from_disk(output_dir, visualization_targets, all_instructions[:5])
        
        # Save dataset info
        dataset_info = {
            "num_samples": num_samples,
            "image_size": self.image_size,
            "num_reference_images": 4,
            "batch_size": batch_size,
            "splits": {
                "train": len(train_indices),
                "val": len(val_indices), 
                "test": len(test_indices)
            },
            "instruction_templates": len(self.composition_templates),
            "shape_types": list(self.shape_names.keys()),
            "pattern_types": list(self.pattern_names.keys())
        }
        
        with open(os.path.join(output_dir, "dataset_info.json"), 'w') as f:
            json.dump(dataset_info, f, indent=2)
        
        print("="*80)
        print("DATASET GENERATION COMPLETED!")
        print("="*80)
        print(f"Files created:")
        print(f"  - reference_scenes.npz: [N, 4, 256, 256] reference images")
        print(f"  - target_sketches.npz: [N, 256, 256] target images")
        print(f"  - instructions.txt: {len(all_instructions)} instructions")
        print(f"  - train/val/test splits with indices")
        print(f"  - sample_*.png: Visualization examples")
        print(f"  - dataset_info.json: Dataset metadata")
        print(f"Total samples: {num_samples}")
        print("="*80)
    
    def create_sample_visualizations_from_disk(self, output_dir: str, target_samples: np.ndarray, instructions: List[str]):
        """Create sample visualizations using minimal data"""
        
        # Load only first 5 samples from reference file for visualization
        ref_file = os.path.join(output_dir, "reference_scenes.npz")
        ref_data = np.load(ref_file)
        ref_samples = ref_data['data'][:5]  # Only load first 5
        
        # Show first 5 samples
        for i in range(min(5, len(instructions))):
            fig, axes = plt.subplots(1, 5, figsize=(20, 4))
            
            # Show 4 reference images
            for j in range(4):
                axes[j].imshow(ref_samples[i, j], cmap='gray', vmin=0, vmax=255)
                axes[j].set_title(f"Reference {j+1}")
                axes[j].axis('off')
            
            # Show target
            axes[4].imshow(target_samples[i], cmap='gray', vmin=0, vmax=255)
            axes[4].set_title("Target Output")
            axes[4].axis('off')
            
            # Add instruction as suptitle
            instruction = instructions[i]
            if len(instruction) > 80:
                instruction = instruction[:80] + "..."
            plt.suptitle(f"Sample {i}: {instruction}", fontsize=12, fontweight='bold')
            
            plt.tight_layout()
            plt.savefig(os.path.join(output_dir, f"sample_{i}.png"), dpi=150, bbox_inches='tight')
            plt.close()
        
        # Create overview grid
        fig, axes = plt.subplots(5, 5, figsize=(20, 20))
        
        for i in range(5):
            # Reference images
            for j in range(4):
                axes[i, j].imshow(ref_samples[i, j], cmap='gray', vmin=0, vmax=255)
                if i == 0:
                    axes[i, j].set_title(f"Ref {j+1}")
                axes[i, j].axis('off')
            
            # Target
            axes[i, 4].imshow(target_samples[i], cmap='gray', vmin=0, vmax=255)
            if i == 0:
                axes[i, 4].set_title("Target")
            axes[i, 4].axis('off')
            
            # Add instruction as ylabel
            instruction = instructions[i]
            if len(instruction) > 40:
                instruction = instruction[:40] + "..."
            axes[i, 0].set_ylabel(f"{instruction}", rotation=90, fontsize=8)
        
        plt.suptitle("Dataset Overview - First 5 Samples", fontsize=16, fontweight='bold')
        plt.tight_layout()
        plt.savefig(os.path.join(output_dir, "dataset_overview.png"), dpi=150, bbox_inches='tight')
        plt.close()
        
        # Clean up
        del ref_data, ref_samples
    
    def create_sample_visualizations(self, output_dir: str, ref_array: np.ndarray, 
                                   target_array: np.ndarray, instructions: List[str]):
        """Create sample visualizations"""
        
        # Show first 5 samples
        for i in range(min(5, len(instructions))):
            fig, axes = plt.subplots(1, 5, figsize=(20, 4))
            
            # Show 4 reference images
            for j in range(4):
                axes[j].imshow(ref_array[i, j], cmap='gray', vmin=0, vmax=255)
                axes[j].set_title(f"Reference {j+1}")
                axes[j].axis('off')
            
            # Show target
            axes[4].imshow(target_array[i], cmap='gray', vmin=0, vmax=255)
            axes[4].set_title("Target Output")
            axes[4].axis('off')
            
            # Add instruction as suptitle
            instruction = instructions[i]
            if len(instruction) > 80:
                instruction = instruction[:80] + "..."
            plt.suptitle(f"Sample {i}: {instruction}", fontsize=12, fontweight='bold')
            
            plt.tight_layout()
            plt.savefig(os.path.join(output_dir, f"sample_{i}.png"), dpi=150, bbox_inches='tight')
            plt.close()
        
        # Create overview grid
        fig, axes = plt.subplots(5, 5, figsize=(20, 20))
        
        for i in range(5):
            # Reference images
            for j in range(4):
                axes[i, j].imshow(ref_array[i, j], cmap='gray', vmin=0, vmax=255)
                if i == 0:
                    axes[i, j].set_title(f"Ref {j+1}")
                axes[i, j].axis('off')
            
            # Target
            axes[i, 4].imshow(target_array[i], cmap='gray', vmin=0, vmax=255)
            if i == 0:
                axes[i, 4].set_title("Target")
            axes[i, 4].axis('off')
            
            # Add instruction as ylabel
            instruction = instructions[i]
            if len(instruction) > 40:
                instruction = instruction[:40] + "..."
            axes[i, 0].set_ylabel(f"{instruction}", rotation=90, fontsize=8)
        
        plt.suptitle("Dataset Overview - First 5 Samples", fontsize=16, fontweight='bold')
        plt.tight_layout()
        plt.savefig(os.path.join(output_dir, "dataset_overview.png"), dpi=150, bbox_inches='tight')
        plt.close()


def main():
    """Generate the composition dataset"""
    
    # Set random seed for reproducibility
    random.seed(42)
    np.random.seed(42)
    
    # Configuration - Use absolute path
    config = {
        "num_samples": 10000,
        "image_size": 256,
        "output_dir": r"\composition_dataset_10k",  # Absolute path
        "batch_size": 500  # Process 500 samples at a time to avoid memory issues
    }
    
    print("Starting dataset generation with configuration:")
    for key, value in config.items():
        print(f"  {key}: {value}")
    
    # Check if output directory path is valid
    import os
    output_parent = os.path.dirname(config["output_dir"])
    if not os.path.exists(output_parent):
        print(f"ERROR: Parent directory does not exist: {output_parent}")
        print("Please create the parent directory first or change the output_dir path")
        return
    
    print(f"\nOutput will be saved to: {config['output_dir']}")
    print(f"Current working directory: {os.getcwd()}")
    
    print(f"\nMemory management:")
    print(f"  - Processing in batches of {config['batch_size']} samples")
    num_batches = (config['num_samples'] + config['batch_size'] - 1) // config['batch_size']
    print(f"  - Will require {num_batches} batches")
    print(f"  - Each batch uses ~{4 * config['batch_size'] * config['image_size'] * config['image_size'] / (1024*1024):.1f} MB for reference images")
    print(f"  - Memory is cleared after each batch")
    
    # Generate dataset
    generator = GeometricCompositionDatasetGenerator(image_size=config["image_size"])
    generator.generate_dataset(
        num_samples=config["num_samples"],
        output_dir=config["output_dir"],
        batch_size=config["batch_size"]
    )
    
    print(f"\nDataset ready for training!")
    print(f"Use dataset_path = r'{config['output_dir']}' in your training script")
    print(f"Files should be located at: {config['output_dir']}")

if __name__ == "__main__":
    main()
```
