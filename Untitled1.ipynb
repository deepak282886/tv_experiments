```python
import numpy as np
import matplotlib.pyplot as plt
import os

def view_batch_outputs(dataset_dir, batch_idx=0, num_samples=5):
    """View outputs from a specific batch"""
    
    print("="*80)
    print(f"VIEWING BATCH {batch_idx} OUTPUTS")
    print("="*80)
    
    batch_dir = os.path.join(dataset_dir, "batches")
    
    # File paths for the batch
    ref_file = os.path.join(batch_dir, f"ref_batch_{batch_idx:03d}.npz")
    target_file = os.path.join(batch_dir, f"target_batch_{batch_idx:03d}.npz")
    inst_file = os.path.join(batch_dir, f"instructions_batch_{batch_idx:03d}.txt")
    
    # Check if files exist
    if not os.path.exists(ref_file):
        print(f"❌ Reference file not found: {ref_file}")
        return
    if not os.path.exists(target_file):
        print(f"❌ Target file not found: {target_file}")
        return
    if not os.path.exists(inst_file):
        print(f"❌ Instructions file not found: {inst_file}")
        return
    
    print(f"✅ Found batch {batch_idx} files:")
    print(f"   📁 Reference: {os.path.getsize(ref_file)/(1024*1024):.1f} MB")
    print(f"   📁 Target: {os.path.getsize(target_file)/(1024*1024):.1f} MB")
    print(f"   📁 Instructions: {os.path.getsize(inst_file)/1024:.1f} KB")
    
    # Load the data
    print(f"\n📊 Loading batch {batch_idx} data...")
    ref_data = np.load(ref_file)['data']
    target_data = np.load(target_file)['data']
    
    with open(inst_file, 'r') as f:
        instructions = [line.strip() for line in f.readlines()]
    
    print(f"✅ Loaded successfully:")
    print(f"   📊 Reference shape: {ref_data.shape}")
    print(f"   📊 Target shape: {target_data.shape}")
    print(f"   📊 Instructions: {len(instructions)}")
    print(f"   📊 Data range - Ref: [{ref_data.min()}, {ref_data.max()}], Target: [{target_data.min()}, {target_data.max()}]")
    
    # Show some sample instructions
    print(f"\n📝 Sample instructions from batch {batch_idx}:")
    for i in range(min(10, len(instructions))):
        print(f"   {i:2d}: {instructions[i]}")
    
    # Visualize some samples
    print(f"\n🖼️  Creating visualizations for {num_samples} samples...")
    
    for sample_idx in range(min(num_samples, len(instructions))):
        fig, axes = plt.subplots(1, 5, figsize=(20, 4))
        
        # Show 4 reference images
        for ref_idx in range(4):
            ref_img = ref_data[sample_idx, ref_idx]
            axes[ref_idx].imshow(ref_img, cmap='gray', vmin=0, vmax=255)
            axes[ref_idx].set_title(f"Reference {ref_idx + 1}")
            axes[ref_idx].axis('off')
            
            # Add info about the reference image
            non_zero = np.count_nonzero(ref_img)
            if non_zero > 0:
                axes[ref_idx].text(0.02, 0.98, f"{non_zero} pixels", 
                                  transform=axes[ref_idx].transAxes, 
                                  bbox=dict(boxstyle="round,pad=0.3", facecolor="yellow", alpha=0.7),
                                  verticalalignment='top', fontsize=8)
            else:
                axes[ref_idx].text(0.02, 0.98, "Empty", 
                                  transform=axes[ref_idx].transAxes,
                                  bbox=dict(boxstyle="round,pad=0.3", facecolor="red", alpha=0.7),
                                  verticalalignment='top', fontsize=8)
        
        # Show target image
        target_img = target_data[sample_idx]
        axes[4].imshow(target_img, cmap='gray', vmin=0, vmax=255)
        axes[4].set_title("Target Output")
        axes[4].axis('off')
        
        target_pixels = np.count_nonzero(target_img)
        axes[4].text(0.02, 0.98, f"{target_pixels} pixels", 
                    transform=axes[4].transAxes,
                    bbox=dict(boxstyle="round,pad=0.3", facecolor="green", alpha=0.7),
                    verticalalignment='top', fontsize=8)
        
        # Add instruction as title
        instruction = instructions[sample_idx]
        if len(instruction) > 80:
            instruction = instruction[:80] + "..."
        
        plt.suptitle(f"Batch {batch_idx}, Sample {sample_idx}: {instruction}", 
                    fontsize=12, fontweight='bold')
        plt.tight_layout()
        
        # Save visualization
        save_path = f"batch_{batch_idx}_sample_{sample_idx}.png"
        plt.savefig(save_path, dpi=150, bbox_inches='tight', facecolor='white')
        plt.show()
        
        print(f"   ✅ Saved: {save_path}")
    
    # Create overview
    print(f"\n🔍 Creating batch overview...")
    fig, axes = plt.subplots(num_samples, 5, figsize=(20, 4 * num_samples))
    
    if num_samples == 1:
        axes = axes.reshape(1, -1)
    
    for sample_idx in range(num_samples):
        # Reference images
        for ref_idx in range(4):
            ref_img = ref_data[sample_idx, ref_idx]
            axes[sample_idx, ref_idx].imshow(ref_img, cmap='gray', vmin=0, vmax=255)
            if sample_idx == 0:
                axes[sample_idx, ref_idx].set_title(f"Ref {ref_idx + 1}")
            axes[sample_idx, ref_idx].axis('off')
        
        # Target image
        target_img = target_data[sample_idx]
        axes[sample_idx, 4].imshow(target_img, cmap='gray', vmin=0, vmax=255)
        if sample_idx == 0:
            axes[sample_idx, 4].set_title("Target")
        axes[sample_idx, 4].axis('off')
        
        # Add instruction as ylabel
        instruction = instructions[sample_idx]
        if len(instruction) > 50:
            instruction = instruction[:50] + "..."
        axes[sample_idx, 0].set_ylabel(f"Sample {sample_idx}\n{instruction}", 
                                      rotation=90, fontsize=9, ha='right')
    
    plt.suptitle(f"Batch {batch_idx} Overview - First {num_samples} Samples", 
                fontsize=16, fontweight='bold')
    plt.tight_layout()
    
    overview_path = f"batch_{batch_idx}_overview.png"
    plt.savefig(overview_path, dpi=150, bbox_inches='tight', facecolor='white')
    plt.show()
    
    print(f"   ✅ Saved overview: {overview_path}")
    
    # Summary statistics
    print(f"\n📈 Batch {batch_idx} Statistics:")
    print(f"   🎯 Total samples: {len(instructions)}")
    print(f"   📏 Image dimensions: {ref_data.shape[2]}x{ref_data.shape[3]}")
    print(f"   🎨 Reference images per sample: {ref_data.shape[1]}")
    
    # Count non-empty reference images
    non_empty_refs = []
    for ref_idx in range(4):
        non_empty = np.sum(np.count_nonzero(ref_data[:, ref_idx], axis=(1, 2)) > 0)
        non_empty_refs.append(non_empty)
        print(f"   📊 Reference {ref_idx + 1}: {non_empty}/{len(instructions)} samples have content ({non_empty/len(instructions)*100:.1f}%)")
    
    # Target statistics
    non_empty_targets = np.sum(np.count_nonzero(target_data, axis=(1, 2)) > 0)
    print(f"   🎯 Targets: {non_empty_targets}/{len(instructions)} samples have content ({non_empty_targets/len(instructions)*100:.1f}%)")
    
    # Instruction analysis
    unique_instructions = len(set(instructions))
    print(f"   📝 Unique instructions: {unique_instructions}/{len(instructions)} ({unique_instructions/len(instructions)*100:.1f}%)")
    
    print("="*80)
    print("BATCH VIEWING COMPLETED")
    print("="*80)

def main():
    """View batch outputs"""
    
    # Configuration
    dataset_dir = r"\composition_dataset_10k"
    batch_idx = 0  # Which batch to view (0 = first batch)
    num_samples = 5  # How many samples to visualize
    
    print(f"Viewing batch {batch_idx} from: {dataset_dir}")
    print(f"Will show {num_samples} sample visualizations")
    
    view_batch_outputs(dataset_dir, batch_idx, num_samples)

if __name__ == "__main__":
    main()
```
